## Identity Operators — `is` and `is not`

`is` checks whether two variables **point to the same object in memory** (same identity), not just equal values.

**`==` vs `is`:**

| Operator | Checks | Example |
|----------|--------|---------|
| `==` | Value equality | `[1,2] == [1,2]` → `True` |
| `is` | Memory identity (same object) | `[1,2] is [1,2]` → `False` |

**Note:** Small integers (-5 to 256) and short strings are **interned** (cached) by Python, so `is` may return `True` even when created separately.

````python
lst1 = [1, 2, 3]
lst2 = [1, 2, 3]
lst3 = lst1

str1 = "hello world"
str2 = "hello world"

print(lst1 is lst2)       # False — different objects
print(lst1 is lst3)       # True  — same object
print(str1 is str2)       # True  — string interning
print(str1 is not str2)   # False
````

## Walrus Operator `:=`

The walrus operator (`:=`) assigns a value to a variable **as part of an expression**. Introduced in Python 3.8, it reduces redundant calls by combining assignment and evaluation in one step.

**Syntax:** `variable := expression`

```python
a = [1, 2, 3, 4, 5]

while (x := len(a)) > 2:
    a.pop()
    print(x)
# Output: 5, 4, 3
```

**Common use:** Avoiding duplicate function calls inside `while` conditions or list comprehensions.

## `global` Keyword

The `global` keyword lets a function **modify a variable from the global scope**. Without it, assigning to a name inside a function creates a new local variable instead.

**Rules:**
- Only needed when **assigning** to a global variable inside a function
- Variables only **read** inside a function are automatically global
- Has no effect at the module level

```python
count = 0

def increment():
    global count
    count += 1

increment()
print(count)  # 1
```

## LEGB Scope Rule

Python resolves variable names using the **LEGB** rule — searching scopes in this order:

| Scope | Description |
|-------|-------------|
| **L** — Local | Inside the current function |
| **E** — Enclosing | In any outer (enclosing) function — applies to closures |
| **G** — Global | At the module (top) level |
| **B** — Built-in | Python's built-ins (`len`, `print`, `range`, etc.) |

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print(x)     # local
    inner()
    print(x)         # enclosing

outer()
print(x)             # global
```

**`nonlocal` keyword:** Used inside a nested function to modify a variable from the **enclosing** (not global) scope.

## Inheritance

Inheritance allows a **child class** to acquire attributes and methods from a **parent class**, enabling code reuse.

**Benefits:**
1. Represents real-world hierarchical relationships (e.g., `Animal → Dog`)
2. Promotes code reuse — avoids rewriting existing logic
3. Transitive — if `B` inherits from `A`, all subclasses of `B` also inherit from `A`

**Types of Inheritance:**

| Type | Description | Example |
|------|-------------|---------|
| **Single** | One child from one parent | `Dog(Animal)` |
| **Multilevel** | Chain of inheritance | `Labrador(Dog)` → `Dog(Animal)` |
| **Hierarchical** | Multiple children from one parent | `Dog(Animal)`, `Cat(Animal)` |
| **Multiple** | One child from multiple parents | `FlyingFish(Fish, Bird)` |
| **Hybrid** | Combination of the above | Mix of any types |


**Single Inheritance:** Single-level inheritance enables a derived class to inherit characteristics from a single-parent class.

**Multilevel Inheritance:** Multi-level inheritance enables a derived class to inherit properties from an immediate parent class which in turn inherits properties from his parent class.

**Hierarchical Inheritance:** Hierarchical-level inheritance enables more than one derived class to inherit properties from a parent class.

**Multiple Inheritance:**  Multiple-level inheritance enables one derived class to inherit properties from more than one base class.

````python
class Person(object):
    def __init__(self, name, idnumber):
        self.name = name
        self.idnumber = idnumber

    def display(self):
        print(self.name)
        print(self.idnumber)

    def details(self):
        print("My name is {}".format(self.name))
        print("IdNumber: {}".format(self.idnumber))

class Employee(Person):
    def __init__(self, name, idnumber, salary, post):
        self.salary = salary
        self.post = post
        Person.__init__(self, name, idnumber)

    def details(self):
        print("My name is {}".format(self.name))
        print("IdNumber: {}".format(self.idnumber))
        print("Post: {}".format(self.post))

a = Employee('Rahul', 886012, 200000, "Intern")
a.display()
a.details()
````

## `super()`

`super()` returns a **proxy object of the parent class**, allowing the child class to call parent methods without hardcoding the parent class name. It is essential for clean multiple inheritance.

**Why use `super()` over `ParentClass.method(self)`?**
- Avoids hardcoding the parent class name
- Works correctly with MRO — ensures each class in the hierarchy is called exactly once

````python
class Parent():
    def show(self):
        print("Inside Parent")

class Child(Parent):
    def show(self):
        super().show()         # calls Parent's show()
        print("Inside Child")

obj = Child()
obj.show()
# Inside Parent
# Inside Child
````

## Method Resolution Order (MRO)

MRO defines the **order in which Python searches for a method or attribute** across the class hierarchy when multiple inheritance is involved.

Python uses the **C3 Linearization algorithm**:
1. Starts from the class itself
2. Searches base classes **left to right**
3. A subclass always appears before its parents
4. View with `ClassName.__mro__` or `ClassName.mro()`

````python
class A:
    def show(self): print("A")

class B(A):
    def show(self): print("B")

class C(A):
    def show(self): print("C")

class D(B, C):
    pass

d = D()
d.show()           # B  (MRO: D -> B -> C -> A)
print(D.__mro__)   # full resolution order
````

## Diamond Problem

The diamond problem occurs in multiple inheritance when a class inherits from **two classes sharing a common ancestor**, creating ambiguity about which parent's method to call.

```
      A
     / \
    B   C
     \ /
      D
```

**Python's solution:** C3 linearization (MRO) produces a consistent order, and `super()` ensures each class in the chain is called **exactly once**.

```python
class A:
    def hello(self): print("A")

class B(A):
    def hello(self):
        super().hello()
        print("B")

class C(A):
    def hello(self):
        super().hello()
        print("C")

class D(B, C):    # MRO: D -> B -> C -> A
    def hello(self):
        super().hello()
        print("D")

D().hello()
# A  C  B  D
```

## Polymorphism

Polymorphism means **"many forms"** — the same method name behaves differently depending on the object.

**1. Method Overriding (Runtime Polymorphism):**
A child class provides its own implementation of a method already defined in the parent.The method name, parameters, and return type must be the same. It is mainly used to achieve runtime polymorphism.

**2. Method Overloading:**
Method overloading refers to the ability to have multiple methods with the same name but different parameters within the same class. Python does not support method overloading by default. However, we can achieve method overloading-like behavior by providing default values for parameters or by using variable-length arguments.

**3. Duck Typing:**
If an object has the required method, it can be used regardless of its class — *"If it walks like a duck, it's a duck."*

**4. Operator Overloading:**
Custom behaviour for operators using dunder methods (`__add__`, `__len__`, `__eq__`, etc.).

````python
class India():
    def capital(self): print("New Delhi is the capital of India.")
    def language(self): print("Hindi is the most widely spoken language.")

class USA():
    def capital(self): print("Washington, D.C. is the capital of USA.")
    def language(self): print("English is the primary language.")

def func(obj):        # works with any object having these methods — duck typing
    obj.capital()
    obj.language()

func(India())
func(USA())
````

## Encapsulation

Encapsulation **bundles data and methods together** inside a class while restricting direct external access to internal data. It exposes only a controlled public interface.

**Access Modifiers in Python:**

| Type | Syntax | Accessible From |
|------|--------|----------------|
| **Public** | `self.name` | Anywhere |
| **Protected** | `self._name` | Class and subclasses (convention only) |
| **Private** | `self.__name` | Within the class only (name-mangled) |

**Name Mangling:** `self.__attr` is internally renamed to `self._ClassName__attr`. This makes it harder but not impossible to access from outside.

````python
class Base:
    def __init__(self):
        self.a = "public"
        self._b = "protected"
        self.__c = "private"

obj = Base()
print(obj.a)            # OK
print(obj._b)           # OK — convention only
print(obj._Base__c)     # OK — name-mangled access
# print(obj.__c)        # AttributeError
````

## Abstraction

Abstraction hides **internal implementation details** and exposes only the essential interface to the user. In Python, it is implemented via the `abc` module.

**Key concepts:**
- **Abstract Class:** Has at least one `@abstractmethod`. Cannot be instantiated directly.
- **Abstract Method:** Declared with no body — every child class **must** override it.
- **Concrete Method:** A regular method with a body inside an abstract class.

**Benefits:** Enforces a contract, promotes reusability, reduces duplication, and improves maintainability.

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Human(Animal):
    def move(self): print("I can walk and run")

class Snake(Animal):
    def move(self): print("I can crawl")

Human().move()
Snake().move()
````

## Context Managers (`with` statement)

A **context manager** manages setup and teardown of resources automatically, even if an exception occurs.

**Dunder methods:**
- `__enter__`: Called at the start of `with` — sets up the resource, returns the object
- `__exit__`: Called at the end — handles cleanup and optionally suppresses exceptions

```python
# Built-in usage
with open("file.txt", "r") as f:
    data = f.read()
# File is closed automatically — even if an error occurred
```

**Custom context manager:**
```python
class ManagedDB:
    def __enter__(self):
        print("Opening DB connection")
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        print("Closing DB connection")
        return False   # False = do not suppress exceptions

with ManagedDB() as db:
    print("Working with DB")
```

**`@contextlib.contextmanager`:** A simpler way to create context managers using a generator with `yield`.

## List vs Tuple

Lists are mutable, which means their elements can be changed, added, or removed after creation, while tuples are immutable and cannot be modified once created.

Tuples are generally faster and use less memory than lists because they do not need extra space or mechanisms to support modification.

Lists provide more built-in methods such as append, remove, and sort because they are designed for dynamic data manipulation, whereas tuples have only a few methods.

Tuples can be used as dictionary keys because they are immutable and hashable, while lists cannot be used as keys since they are mutable.

## Pass by Object Reference

Python uses **pass-by-object-reference** (also called pass-by-assignment).

- The function receives a **reference** to the object, not a copy
- **Mutable objects** (list, dict): changes inside the function affect the original
- **Immutable objects** (int, str, tuple): rebinding inside the function creates a new object; original is unchanged

```python

    x = x + 1         # creates new int — original unchanged

print(my_list)         # [1, 2, 99]

n = 5
rebind(n)
print(n)               # 5
```

## Shallow Copy vs Deep Copy


Object Level: Shallow copy duplicates only the outer object, while deep copy duplicates both the outer object and all nested objects.

Effect of Changes: In shallow copy, changes to nested mutable objects affect both copies, whereas in deep copy, changes do not affect the original.

Performance & Memory: Shallow copy is faster and memory-efficient, while deep copy is slower and consumes more memory because it creates full independent copies.

| Aspect | Shallow Copy | Deep Copy |
|--------|-------------|-----------|
| **Creates** | New outer object; **shares** nested objects | New outer object **and** new nested objects |
| **Nested changes** | Affect both copies | Only affect the copy |
| **Performance** | Faster, less memory | Slower, more memory |
| **Method** | `copy.copy()`, `.copy()`, `[:]` | `copy.deepcopy()` |

```python
import copy
original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
shallow[0].append(99)
print(original)    # [[1, 2, 99], [3, 4]] -- nested affected

deep = copy.deepcopy(original)
deep[0].append(99)
print(original)    # [[1, 2, 99], [3, 4]] -- original unchanged
```

**`copy()` vs `[:]`:** Both create a shallow copy. `copy()` is more explicit; `[:]` is concise. Behavior is identical.

## Iterator

An **iterator** implements the **iterator protocol** with two methods:

```python
        return self.current

for n in CountUp(3):
```


## Closure

Closures in Python are like “memory-equipped” functions. They allow a function to remember values from the environment in which it was created even if that environment no longer exists. 

**Three conditions for a closure:**
1. There must be a nested (inner) function
2. The inner function must reference a variable from the enclosing scope
3. The outer function must return the inner function

**Use cases:** Factory functions, decorators, stateful callbacks, memoization.

````python
def counter():
    count = 0
    def increment():
        nonlocal count      # refers to enclosing scope variable
        count += 1
        return count
    return increment

c = counter()
print(c())  # 1
print(c())  # 2
print(c())  # 3
````

## Decorators

are one of the most helpful and powerful tools of Python. These are used to modify the behavior of the function. Decorators provide the flexibility to wrap another function to expand the working of wrapped function, without permanently modifying it.

In Decorators, functions are passed as an argument into another function and then called inside the wrapper function.

**How it works:**
1. Decorator receives the original function as argument
2. Defines a wrapper adding new behaviour
3. Returns the wrapper

`@decorator` is syntactic sugar for `func = decorator(func)`.

```python
def log(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Done")
        return result
    return wrapper

@log
def add(a, b):
    return a + b

add(2, 3)
```

````python
def my_decorator(func):
    def wrapper():
        print("Before the function is called.")
        func()
        print("After the function is called.")
    return wrapper

@my_decorator
def hello_world():
    print('hello')

hello_world()
````

## Built-in Decorators

### `@staticmethod`

Defines a method that belongs to a class but does not access class or instance variables.

When the method performs a utility task related to the class but does not need self or cls.

### `@classmethod`

Defines a method that takes the class (cls) as the first parameter instead of the instance.

When you need to access or modify class-level variables.

### `@property`

Converts a method into a read-only attribute.

To implement getter functionality and encapsulation.

### `@abstractmethod`
Forces child classes to **implement the method**. The abstract class itself cannot be instantiated.

### `@dataclass`

is a Python decorator that automatically generates boilerplate methods like __init__, __repr__, and __eq__ for classes used to store data.
It makes code shorter, cleaner, and more readable.
You can make instances immutable with frozen=True and provide default values for fields.

````python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b
````

````python
class Employee:
    company = "ABC"

    @classmethod
    def change_company(cls, name):
        cls.company = name
````

````python
class Person:
    def __init__(self, age):
        self._age = age

    @property
    def age(self):            # getter
        return self._age

    @age.setter
    def age(self, value):     # setter
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value

    @age.deleter
    def age(self):            # deleter
        del self._age

p = Person(25)
print(p.age)   # 25 — no parentheses
p.age = 30     # calls setter
del p.age      # calls deleter
````

## Generator


a generator is a special type of iterator that allows you to iterate over a sequence of values lazily. Unlike regular functions that return a single value and terminate, generators use the yield keyword to produce a series of values, one at a time, which can be iterated over. They are useful when you need to handle large datasets or streams of data without loading everything into memory at once.


**Advantages:**
- Memory efficient — values generated on demand
- Simpler than writing a full iterator class
- Supports infinite sequences

**Generator expression:** `(expr for item in iterable)` — lazy version of list comprehension.

````python
def count_up_to(max):
    count = 1
    while count <= max:
        yield count
        count += 1

for i in count_up_to(5):
    print(i)
````

## `yield` vs `return`

| Aspect | `return` | `yield` |
|--------|----------|---------|
| **Function terminates** | Yes — completely | No — paused at yield |
| **State preserved** | No | Yes — resumes from where it paused |
| **Output** | Single value | Series of values (one per `next()`) |
| **Memory** | All values at once | One value at a time (lazy) |
| **Function type** | Regular function | Generator function |

```python
def use_yield():
    for i in [1, 2, 3]:
        yield i        # pauses here, resumes on next call

gen = use_yield()
print(next(gen))   # 1
print(next(gen))   # 2
```

## Recursion

A function that **calls itself** to solve a smaller subproblem. Every recursive function needs a **base case** (stopping condition) to prevent infinite recursion.

```python
def factorial(n):
    if n == 0 or n == 1:      # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print(factorial(5))   # 120
```

**Limitations:**
- Deep recursion causes `RecursionError` (default limit: 1000)
- Often less efficient than iteration without memoization

**`functools.lru_cache`:** Caches results to avoid redundant calls (memoization).
```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib(n):
    if n <= 1: return n
    return fib(n-1) + fib(n-2)

print(fib(50))   # instant — cached results reused
```

## Exceptions

an exception is an event that disrupts the normal flow of a program's execution. It occurs when the interpreter encounters an error during runtime, which causes the program to stop executing the current block of code.

# Concurrency

## Multiprocessing

Multiprocessing in Python allows parallel execution using multiple processes, each with its own memory space. It is mainly used for CPU-bound tasks to bypass the GIL and utilize multiple CPU cores efficiently.

**Key advantages:**
- Bypasses the GIL — each process has its own interpreter
- Can fully utilize multiple CPU cores
- Process crashes are isolated

````python
from multiprocessing import Process
import os

def print_process():
    print("Process ID:", os.getpid())

if __name__ == "__main__":
    p1 = Process(target=print_process)
    p2 = Process(target=print_process)
    p1.start()
    p2.start()
    p1.join()
    p2.join()
````

## Multithreading

Multithreading is a technique where multiple threads run within the same process to perform tasks concurrently.All threads share the same memory space.

Best for **I/O-bound tasks** (file I/O, network, database).

**Key characteristics:**
- All threads share the same memory
- Threads are lightweight
- Limited by the GIL for CPU-bound work — use multiprocessing instead

````python
import threading
import time

def task():
    print("Task started")
    time.sleep(2)
    print("Task finished")

t1 = threading.Thread(target=task)
t2 = threading.Thread(target=task)
t1.start()
t2.start()
t1.join()
t2.join()
````

## Thread Safety

Thread safety ensures that shared resources are accessed correctly when multiple threads run concurrently. It prevents race conditions by using synchronization mechanisms like locks.

**Synchronization mechanisms:**

| Tool | Purpose |
|------|---------|
| `threading.Lock()` | Only one thread holds the lock at a time |
| `threading.RLock()` | Re-entrant lock — same thread can acquire multiple times |
| `threading.Event()` | Signal between threads |
| `threading.Semaphore(n)` | Limit concurrent access to n threads |

```python
import threading

lock = threading.Lock()
counter = 0

def increment():
    global counter
    with lock:       # acquire and auto-release
        counter += 1

threads = [threading.Thread(target=increment) for _ in range(100)]
for t in threads: t.start()
for t in threads: t.join()
print(counter)   # always 100 — thread-safe
```

## Race Condition

Occurs when **two or more threads simultaneously access and modify shared data**, and the final result depends on unpredictable execution order.

```python
# Without lock -- race condition
counter = 0

def increment():
    global counter
    temp = counter     # read
    temp += 1          # modify
    counter = temp     # write  <- another thread may have changed counter here

# Result: inconsistent (e.g., 87 instead of 100)
```

**Solution:** Use `threading.Lock()` to ensure atomic read-modify-write operations.

## `asyncio` — Asynchronous Programming

`asyncio` enables **concurrent I/O-bound tasks** using a **single-threaded event loop** — no real parallelism, but high efficiency through cooperative multitasking.

**Key concepts:**
- `async def` — defines a coroutine
- `await` — pauses the coroutine and yields control to the event loop
- `asyncio.gather()` — runs multiple coroutines concurrently
- `asyncio.run()` — starts the event loop

````python
import asyncio

async def task(name):
    print(f"{name} started")
    await asyncio.sleep(2)    # yields control -- other tasks run here
    print(f"{name} finished")

async def main():
    await asyncio.gather(
        task("Task 1"),
        task("Task 2")
    )

asyncio.run(main())
````

Multiprocessing is used for CPU-bound tasks like machine learning or image processing because it runs tasks in parallel using multiple processes and bypasses the GIL.

Multithreading is used for I/O-bound tasks such as file operations, web requests, or database calls where threads wait for external resources.

Asyncio is used for highly concurrent I/O tasks like web servers or APIs by using an event loop to handle many requests efficiently.

# Memory Management

## Reference Counting

Python uses reference counting for memory management. Each object keeps track of how many references point to it. When the reference count drops to zero, the object is immediately deleted. To handle circular references, Python also uses a generational garbage collector.

```python
import sys
a = [1, 2, 3]
b = a                      # reference count -> 2
print(sys.getrefcount(a))  # usually +1 for the temp ref in getrefcount
del b                      # count -> 1
del a                      # count -> 0 -> memory freed
```

**Limitation:** Cannot handle **circular references** (e.g., `a.ref = b; b.ref = a`). Handled by the generational garbage collector.

````python
import sys

a = [1, 2, 3]
print(sys.getrefcount(a))   # reference count
````

## Garbage Collection

Python's **garbage collector (gc)** works alongside reference counting to free memory from **circular references** that reference counting cannot resolve.

**Generational GC:** Objects are grouped into 3 generations by age:

| Generation | Description | Collected |
|------------|-------------|-----------|
| 0 | Newly created objects | Most frequently |
| 1 | Survived one GC cycle | Occasionally |
| 2 | Long-lived objects | Least frequently |

```python
import gc
gc.collect()           # manually trigger GC
gc.get_threshold()     # (700, 10, 10) — collection thresholds
```

## Stack vs Heap Memory

What is Stored: Stack stores function calls and reference variables, while heap stores the actual objects and data.

Memory Management: Stack is automatically managed in LIFO order, whereas heap memory is managed by reference counting and garbage collection.

Speed: Stack access is faster because it follows a fixed allocation pattern, while heap allocation is slightly slower due to dynamic memory management.

Size & Lifetime: Stack memory is limited and exists only during function execution, while heap memory is dynamic in size and objects remain until they are garbage collected.

| Aspect | Stack | Heap |
|--------|-------|------|
| **Stores** | Function frames, local variable references | Actual Python objects and data |
| **Management** | Automatic — LIFO order | Reference counting + GC |
| **Speed** | Faster | Slightly slower |
| **Size** | Limited, fixed | Dynamic |
| **Lifetime** | During function execution | Until reference count reaches zero |

**Note:** Python variables on the stack are **references (pointers)** to objects on the heap.

## How is Memory Managed in Python?


Python manages memory automatically through the use of a private heap containing all Python objects and data structures. The key components of Python’s memory management include automatic garbage collection, memory pools, and dynamic typing.

Python utilizes reference counting to detect and reclaim memory from unused objects, along with a cyclic garbage collector to free up circular references.

## Global Interpreter Lock (GIL)

The GIL, or Global Interpreter Lock, is a mechanism used in CPython (the standard Python implementation) to ensure that only one thread executes Python bytecode at a time.

This lock is necessary because CPython’s memory management is not thread-safe.
which could lead to race conditions and corrupt data structures.

The main implication of the GIL is that it limits the execution of multi-threaded Python programs on multi-core processors.

**Implications:**
- **CPU-bound multithreaded programs** do not benefit from multiple cores — threads take turns
- **I/O-bound programs** are less affected — threads release the GIL while waiting for I/O
- **Multiprocessing bypasses the GIL** — each process has its own interpreter and GIL

**Python 3.13+:** Experimental support for per-interpreter GIL removal.

# Advanced Topics

## Metaclass

A **metaclass** is a class of a class — it defines how a class itself is **created and behaves**. Just as a class is a blueprint for objects, a metaclass is a blueprint for classes.

The default metaclass in Python is `type`.

```python
# type() creates classes dynamically
MyClass = type('MyClass', (object,), {'greet': lambda self: "Hello"})
print(MyClass().greet())   # Hello
```

**Custom metaclass:**
```python
class Meta(type):
    def __new__(mcs, name, bases, namespace):
        print(f"Creating class: {name}")
        return super().__new__(mcs, name, bases, namespace)

class MyClass(metaclass=Meta):
    pass
# Output: Creating class: MyClass
```

**Use cases:** Enforcing interface contracts, auto-registering subclasses, adding logging or validation to all class methods.

## Interpreted Language

An **interpreted language** executes code **line by line** at runtime without a prior compilation to machine code.

**Python's execution flow:**
`Source (.py)` → `Bytecode (.pyc)` → `Python Virtual Machine (PVM)` → `Output`

**Note:** Python compiles to bytecode first, then the PVM interprets it — often described as both compiled and interpreted.

**Compiled vs Interpreted:**

| Aspect | Compiled (C, C++) | Interpreted (Python) |
|--------|-------------------|---------------------|
| Execution | Directly by CPU | Through interpreter |
| Speed | Faster | Slower |
| Error detection | Compile time | Runtime |
| Portability | Platform-specific binary | Platform-independent |

## Pickling and Unpickling

Pickling:

Pickling is the name of the serialization process in Python. Any object in Python can be serialized into a byte stream and dumped as a file in the memory.

The function used for the above process is pickle.dump()

UnPickling:

Unpickling is the complete inverse of pickling. It deserializes the byte stream to recreate the objects stored in the file and loads the object to memory.
The function used for the above process is pickle.load().

```python
import pickle

# Pickling
data = {"name": "Alice", "scores": [90, 85, 92]}
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# Unpickling
with open("data.pkl", "rb") as f:
    loaded = pickle.load(f)
print(loaded)   # {'name': 'Alice', 'scores': [90, 85, 92]}
```

**Security warning:** Never unpickle data from untrusted sources — it can execute arbitrary code.

**Alternatives:** `json` (human-readable, cross-language), `shelve`, `marshal`.

## `__slots__`

`__slots__` restricts which instance attributes a class can have, replacing the default `__dict__` with a fixed, memory-efficient structure.

**Benefits:**
- Significant memory reduction for classes with many instances
- Faster attribute access
- Prevents accidental creation of unexpected attributes

```python
class Point:
    __slots__ = ['x', 'y']    # only x and y are allowed

    def __init__(self, x, y):
        self.x = x
        self.y = y

p = Point(1, 2)
print(p.x)         # 1
# p.z = 5          # AttributeError -- z not in __slots__
# p.__dict__       # AttributeError -- no __dict__
```

**When to use:** When creating millions of instances of a simple class (e.g., coordinate points, data records) where memory matters.

## Monkey Patching 

Monkey patching means modifying or extending a class or module at runtime without changing the original source code.

````python
class Dog:
    def sound(self):
        print("Bark")


# Original behavior
d = Dog()
d.sound()

def new_sound(self):
    print("Meow")

Dog.sound = new_sound   # Monkey patch

d = Dog()
d.sound()
````
