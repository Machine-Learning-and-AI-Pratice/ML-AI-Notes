## Linear Regression in Machine learning

Linear regression is a type of supervised machine-learning algorithm that learns from the labelled datasets and maps the data points with most optimized linear functions which can be used for prediction on new datasets. It assumes that there is a linear relationship between the input and output, meaning the output changes at a constant rate as the input changes. This relationship is represented by a straight line.

For example we want to predict a student's exam score based on how many hours they studied. We observe that as students study more hours, their scores go up. In the example of predicting exam scores based on hours studied. Here

- Independent variable (input): Hours studied because it's the factor we control or observe.
- Dependent variable (output): Exam score because it depends on how many hours were studied.





## Why Linear Regression is Important?
Here’s why linear regression is important:

Simplicity and Interpretability: It’s easy to understand and interpret, making it a starting point for learning about machine learning.
Predictive Ability: Helps predict future outcomes based on past data, making it useful in various fields like finance, healthcare and marketing.
Basis for Other Models: Many advanced algorithms, like logistic regression or neural networks, build on the concepts of linear regression.
Efficiency: It’s computationally efficient and works well for problems with a linear relationship.
Widely Used: It’s one of the most widely used techniques in both statistics and machine learning for regression tasks.
Analysis: It provides insights into relationships between variables (e.g., how much one variable influences another).

## Best Fit Line in Linear Regression

In **linear regression**, the best-fit line is the straight line that most accurately represents the relationship between the **independent variable (input)** and the **dependent variable (output)**.  
It minimizes the difference between the actual data points and the predicted values from the model.

---

## 1 Goal of the Best-Fit Line
- Find a straight line that **minimizes error** between observed data points and predicted values.
- Helps predict the dependent variable for **new, unseen data**.

### Linear Regression in Machine Learning
- **Y** → Dependent / Target variable  
- **X** → Independent / Predictor variable  
- Linear function is the simplest regression function.  
- X can be a **single feature** or **multiple features** representing the problem.




---

## 2 Equation of the Best-Fit Line 

For simple linear regression



Where:
- **y** → Predicted value (dependent variable)  
- **x** → Input (independent variable)  
- **m** → Slope (change in y per unit change in x)  
- **b** → Intercept (value of y when x = 0)  

The best-fit line will be the one that optimizes the values of m (slope) and b (intercept) so that the predicted y values are as close as possible to the actual data points.


--- 

## 3. Minimizing the Error: Least Squares Method
- Method: **Least Squares** → minimizes the sum of squared differences between actual and predicted values.  
- These differences are called **residuals**.



👉 Ensures the line best represents the data by minimizing squared residuals.

---

## 4. Interpretation of the Best-Fit Line

- **Slope (m):** Indicates how much y changes with each unit change in x.  
  - Example: If slope = 5 → For every 1-unit increase in x, y increases by 5 units.
- **Intercept (b):** Predicted value of y when x = 0.  
  - Point where the line crosses the y-axis.

## ⚠️ Limitations of Linear Regression
  - **Assumes Linearity:** Works only if the relationship between variables is linear.  
  - **Sensitivity to Outliers:** Outliers can skew slope and intercept, distorting the best-fit line.





# 📊 Hypothesis Function in Linear Regression

In **linear regression**, the hypothesis function is the equation used to make predictions about the **dependent variable (Y)** based on the **independent variables (X)**.  
It represents the relationship between the input features and the target output.

## 1. 🧮 Simple Linear Regression (One Independent Variable)



Where:
- **h(x)** or **ŷ** → Predicted value of the dependent variable (y)  
- **x** → Independent variable  
- **β₀** → Intercept (value of y when x = 0)  
- **β₁** → Slope (change in y for each unit change in x)  

---

## 2. 🔢 Multiple Linear Regression (More than One Independent Variable)



Where:
- **x₁, x₂, ..., xₖ** → Independent variables  
- **β₀** → Intercept  
- **β₁, β₂, ..., βₖ** → Coefficients representing the influence of each independent variable on the predicted output  

---

## 3. 📐 Assumptions of Linear Regression

1. **Linearity** → Relationship between inputs (X) and output (Y) is linear.  

      
2. **Independence of Errors** → Prediction errors should not influence each other.  
3. **Constant Variance (Homoscedasticity)** → Errors should have equal spread across all values of X.  
   - If spread changes (fans out/shrinks), it’s called **heteroscedasticity**.  



4. **Normality of Errors** → Errors should follow a normal (bell-shaped) distribution.  
5. **No Multicollinearity** (for multiple regression) → Input variables shouldn’t be too closely related.  
6. **No Autocorrelation** → Errors shouldn’t show repeating patterns, especially in time-series data.  
7. **Additivity** → Total effect on Y is the sum of effects from each X, with no interaction terms.

# 📊 Types of Linear Regression

Linear regression can be categorized based on the number of independent features:

- **Simple Linear Regression (Univariate Regression):** One independent variable.  
- **Multiple Linear Regression (Multivariate Regression):** More than one independent variable.

---

## 1. 🔹 Simple Linear Regression

Used when we want to predict a target value (dependent variable) using **only one input feature** (independent variable).  
It assumes a **straight-line relationship** between the two.

### Formula:


Where:
- **ŷ** → Predicted value  
- **x** → Input (independent variable)  
- **θ₀** → Intercept (value of ŷ when x = 0)  
- **θ₁** → Slope / Coefficient (how much ŷ changes with one unit of x)  

**Example:** Predicting a person’s salary (y) based on their years of experience (x).

---

## 2. 🔹 Multiple Linear Regression

Involves **more than one independent variable** and one dependent variable.  
The goal is to find the **best-fit line (or hyperplane)** that predicts values based on multiple inputs.

### Formula:


Where:
- **ŷ** → Predicted value  
- **x₁, x₂, …, xₙ** → Independent variables  
- **θ₁, θ₂, …, θₙ** → Coefficients (weights for each predictor)  
- **θ₀** → Intercept  

---

## 3. 🧠 Key Idea of Regression

- Regression uses a dataset with **X (features)** and **Y (target values)**.  
- The algorithm learns a function that maps **X → Y**.  
- Once trained, this function can predict **Y** for new, unseen **X** values.  
- Regression always predicts **continuous values** (not categories).

---


# 📊 Use Cases of Multiple Linear Regression (MLR)

Multiple Linear Regression (MLR) allows us to analyze the relationship between **multiple independent variables** and a **single dependent variable**.  
It is widely applied across industries to make predictions and understand influencing factors.

---

## 1. 🏠 Real Estate Pricing
- **Goal:** Predict property prices.  
- **Independent Variables:** Location, size, number of bedrooms, amenities, etc.  
- **Dependent Variable:** Property price.  
- **Benefit:** Helps buyers and sellers understand market trends and set competitive prices.

---

## 2. 💹 Financial Forecasting
- **Goal:** Predict stock prices or economic indicators.  
- **Independent Variables:** Interest rates, inflation rates, market trends, GDP growth, etc.  
- **Dependent Variable:** Stock price or financial metric.  
- **Benefit:** Enables better investment strategies and risk management.

---

## 3. 🌾 Agricultural Yield Prediction
- **Goal:** Estimate crop yields.  
- **Independent Variables:** Rainfall, temperature, soil quality, fertilizer usage, pest control, etc.  
- **Dependent Variable:** Crop yield.  
- **Benefit:** Helps farmers plan agricultural practices for optimal productivity.

---

## 4. 🛒 E-commerce Sales Analysis
- **Goal:** Assess factors influencing sales.  
- **Independent Variables:** Product price, marketing promotions, seasonal trends, customer demographics, etc.  
- **Dependent Variable:** Sales volume or revenue.  
- **Benefit:** Guides pricing strategies, marketing campaigns, and inventory planning.

---

## 🔑 Key Insight
MLR is powerful because it:
- Considers **multiple factors simultaneously**.  
- Provides a more **realistic model** of complex relationships.  
- Improves **prediction accuracy** compared to simple regression.
