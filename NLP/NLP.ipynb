# NLP

# What is Natural Language Processing (NLP)?

Natural Language Processing (NLP) is a field of Artificial Intelligence (AI) that combines:

- Computer Science  
- Artificial Intelligence  
- Linguistics (the study of language)  

It enables computers to understand, interpret, process, and generate human language in a meaningful and useful way.

NLP allows machines to read text, understand speech, analyze meaning, and respond intelligently.

---

# Phases of Natural Language Processing

The phases of NLP are the structured steps that computers follow to understand human language.

## 1️⃣ Lexical Analysis
Breaking text into smaller units called tokens (words or phrases).  
It includes normalization such as removing punctuation and converting text to lowercase.

## 2️⃣ Syntactic Analysis
Analyzing the grammatical structure of a sentence to understand how words are arranged.

## 3️⃣ Semantic Analysis
Understanding the meaning of words, phrases, and sentences.

## 4️⃣ Pragmatic Analysis
Understanding context, intention, and implied meaning beyond the literal words.

---

## 🧠 Summary

Together, these phases help machines:

- Convert raw text into structured data  
- Understand grammar  
- Extract meaning  
- Interpret context and intention  
- Generate accurate and meaningful responses  

This layered approach allows AI systems to understand and respond to human language effectively.



## 1️⃣ Lexical Analysis

**Definition:**  
Lexical Analysis is the first step in NLP where text is divided into meaningful units called **tokens** and classified. It involves scanning, tokenization, and normalization—such as removing whitespace and converting to lowercase—to transform text into structured data (lexemes) for further analysis like parsing and semantic processing.

### 🔑 Key Tasks
- Tokenization  
- Test Normalization  (lowercasing, removing punctuation, etc.)

### ❓ Why It Matters
Lexical analysis enables machines to convert unstructured, raw human language into structured, manageable components.  

It prepares the text for deeper, more complex stages such as:
- Syntax analysis (grammar structure)
- Semantic analysis (meaning)

---

## 2️⃣ Morphological Analysis

**Definition:**  
Morphological analysis is the process of analyzing the internal structure of words by breaking them down into their smallest meaningful units, called **morphemes**. It identifies how roots, prefixes, and suffixes combine to create a word's specific meaning and grammatical function.

### 🔑 Key Tasks
- Stemming  
- Lemmatization  

### ❓ Why It Matters
Without morphological analysis, a computer might see:

- `wait`
- `waiter`
- `waiters`

as three completely unrelated strings of text.

With morphological analysis, the system understands they share the same core concept of **"wait"**.

---

## 3️⃣ Syntactic Analysis (Parsing)

**Definition:**  
Parsing is the process of analyzing the grammatical structure of a sentence. It determines how words are arranged according to grammar rules and represents this structure using a **parse tree**, showing relationships such as subject, verb, and object.

### 🔑 Key Components
- POS Tagging  
- Ambiguity Resolution  

### ❓ Why It Matters
We use syntactic analysis to understand:
- Sentence structure  
- Relationships between words  

This helps machines:
- Correctly understand meaning  
- Improve grammar checking  
- Support translation  
- Answer questions  
- Perform text understanding  

---

## 4️⃣ Semantic Analysis

**Definition:**  
Semantic analysis is the process of understanding the meaning of words, phrases, and sentences in context to ensure the text is meaningful, logically correct, and contextually appropriate.

It focuses on understanding relationships between words and identifying the actual meaning of the sentence.

### 🔑 Key Tasks
- Named Entity Recognition (NER)  
- Word Sense Disambiguation (WSD)  

### ❓ Why It Matters
Logic is more important than structure.  

Without semantic analysis, a computer might accept a grammatically correct sentence that makes no logical sense.

**Example:**
```python
total = "Price" * 10


## 6️⃣ Pragmatic Analysis

### Definition

Pragmatic Analysis is the process of understanding the intended meaning of words and sentences beyond their literal or semantic meaning.  

It focuses on the speaker’s or writer’s intentions, tone, social context, and real-world knowledge.

Unlike semantic analysis (which examines direct meaning), pragmatic analysis interprets implied meaning based on context.

---

### 🔑 Key Tasks

- Understanding Speaker Intent
- Interpreting Figurative Language
- Detecting Sarcasm and Irony
- Handling Implicit Meaning
- Context-Based Interpretation

---

### 📌 Examples

**Example 1:**  
"Call me a cab."

- Semantic meaning → Change my name to "a cab."
- Pragmatic meaning → Please order a taxi for me.

**Example 2:**  
"It's cold in here."

- Semantic meaning → The temperature is low.
- Pragmatic meaning → Please close the window or turn off the AC.

---

### ❓ Why It Matters

**AI & Chatbots:**  
Enables systems like Siri or ChatGPT to understand user intent rather than interpreting statements literally.

**Sentiment Analysis:**  
Helps detect sarcasm or hidden emotions in customer feedback.

**Human Communication:**  
Prevents misunderstanding by interpreting social norms, indirect requests, and polite hints instead of robotic literal meanings.

---

### 🧠 Summary

Pragmatic analysis moves language understanding from:
Literal Meaning → Contextual Understanding → Intended Meaning


# Text Preprocessing in NLP

Text preprocessing is the initial step in NLP where raw text is cleaned and prepared before being fed into models.  
It improves model accuracy, reduces noise, and ensures consistent input format.

---

# 1️⃣ Text Cleaning

Text Cleaning is the preprocessing step where unwanted characters, noise, and inconsistencies are removed from text to improve NLP model performance.

## 🔹 Common Text Cleaning Techniques

- Removing extra whitespaces  
- Converting text to lowercase  
- Removing HTML tags  
- Removing URLs  
- Removing email IDs  
- Removing special characters and symbols  
- Removing numbers (if not required)  
- Removing punctuation  
- Removing emojis (or converting them to text)  
- Expanding contractions (e.g., *don’t → do not*)  
- Removing duplicate text  
- Handling misspellings  

### 🎯 Why It Matters
Clean text reduces noise and ensures that models focus only on meaningful information.

---

# 2️⃣ Tokenization

Tokenization is the process of breaking text into smaller units called **tokens**.  
Tokens can be words, phrases, characters, or sub-words.

---

## 🔹 Types of Tokenization

### 1. Word Tokenization
Breaks text into individual words.  
Works well for English and similar languages.

### 2. Character Tokenization
Breaks text into single characters.  
Useful for spelling tasks or languages without clear word boundaries.

### 3. Sub-word Tokenization
Breaks words into meaningful pieces smaller than a word but larger than a character.  
Common in modern deep learning models (e.g., handling unknown words).

### 4. Sentence Tokenization
Splits paragraphs into individual sentences.

### 5. N-gram Tokenization
Splits text into sequences of **n consecutive words or characters** to capture patterns.

Example:
- Bigram (n=2): *Natural Language*, *Language Processing*

---

### 🎯 Why We Use Tokenization

Breaking text into smaller units helps machines:

- Identify patterns  
- Understand word relationships  
- Perform tasks like translation, search, and sentiment analysis  

If entire sentences were used as single units, models would struggle to analyze structure and meaning effectively.

---

# 3️⃣ Text Corpus

A **Text Corpus** is a large collection of structured or unstructured text used to train and evaluate NLP systems.

It provides real-world language data that helps models learn vocabulary, grammar, and context patterns.

---

## 🔹 Types of Text Corpus

- **Monolingual Corpus** – Contains text in a single language  
- **Multilingual Corpus** – Contains text in multiple languages  
- **Parallel Corpus** – Same content translated into multiple languages (used in machine translation)  
- **Annotated Corpus** – Labeled with information such as POS tags, NER, sentiment, etc.  
- **Unannotated Corpus** – Raw text without labels  

### 📌 Example
For sentiment analysis, a corpus of movie reviews labeled as *positive* or *negative* is used to train the model.  
The model learns sentiment patterns and predicts sentiment for new reviews.

---

# 4️⃣ Sentence Segmentation

Sentence Segmentation is the process of dividing text into individual sentences.  
It helps NLP systems determine where one sentence ends and another begins.

---

## 🔹 Example

**Text:**

"I love NLP. It is very interesting!"

**After Sentence Segmentation:**

- "I love NLP."
- "It is very interesting!"

---

### 🎯 Why It Matters

Processing text sentence by sentence:

- Improves grammatical understanding  
- Enhances meaning extraction  
- Makes parsing and semantic analysis more accurate  

Models perform better when text is not treated as one long, unstructured block.


# Linguistic Processing in NLP

Linguistic Processing focuses on analyzing the grammatical and structural aspects of language.  
It helps machines move beyond simple word recognition toward deeper structural understanding.

---

# 1️⃣ Part-of-Speech (POS) Tagging

## Definition

Part-of-Speech (POS) Tagging is the process of automatically assigning grammatical categories (noun, verb, adjective, adverb, etc.) to each word in a sentence based on its context.

---

## 🎯 Why It Matters

- Reduces ambiguity (e.g., *book* as noun vs verb)
- Improves sentence structure understanding
- Supports downstream NLP tasks such as:
  - Parsing
  - Named Entity Recognition (NER)
  - Machine Translation
  - Text Summarization

---

## 📌 Example

Sentence:  
> "She booked a flight yesterday."

After POS Tagging:

- She → Pronoun  
- booked → Verb  
- a → Determiner  
- flight → Noun  
- yesterday → Adverb  

✔ This confirms "booked" is used as an action, not an object.

---

## 📍 Where POS Tagging Is Used

- Parsing  
- Machine Translation  
- Named Entity Recognition  
- Information Extraction  

---

# 2️⃣ Chunking (Shallow Parsing)

## Definition

Chunking groups words into meaningful phrases (chunks) such as noun phrases (NP) or verb phrases (VP), based on POS tags.

---

## 📌 Example

Sentence:  
> "She bought a red car."

Chunking Result:

- [She]  
- [bought]  
- [a red car] → Noun Phrase  

---

## 🎯 Why It Matters

Chunking provides phrase-level structure, which is more meaningful than analyzing individual words.

---

## 📍 Where It Is Used

- Information Extraction  
- Question Answering  
- Syntactic Preprocessing  

---

# 3️⃣ Constituency Parsing

## Definition

Constituency Parsing breaks a sentence into hierarchical sub-phrases (constituents) such as:

- Noun Phrase (NP)
- Verb Phrase (VP)

It represents the structure as a tree.

---

## 📌 Example

Sentence:  
> "She bought a red car."

Structure:

- [NP She]  
- [VP bought [NP a red car]]

---

## 🎯 Why It Matters

- Captures overall grammatical structure  
- Shows how words group together  
- Important for syntactic understanding  

---

## 📍 Where It Is Used

- Machine Translation  
- Question Answering  
- Text Understanding  
- Grammar Checking  

---

# 4️⃣ Dependency Parsing

## Definition

Dependency Parsing identifies grammatical relationships between words.  
It focuses on word-to-word connections instead of phrase hierarchies.

---

## 📌 Example

Sentence:  
> "She bought a red car."

Dependency Structure:

- bought → She (nsubj: subject)
- bought → car (dobj: object)
- car → red (amod: adjective modifier)

---

## 🎯 Why It Matters

Unlike bag-of-words models, dependency parsing shows:

- Who did what
- To whom
- With what attributes

From this structure:

- Actor → She  
- Action → bought  
- Object → car  
- Attribute → red  

This enables accurate semantic reasoning.

---

## 📍 Where It Is Used

- Information Extraction  
- Knowledge Graph Construction  
- Question Answering  
- Machine Translation  
- Semantic Search  

---

## ⚠ Limitations & Trade-offs

- Sensitive to noisy or informal text  
- Computationally expensive  
- Modern transformer models (BERT, GPT) implicitly learn syntax, reducing the need for explicit parsing in some pipelines  

👉 Senior Insight: Dependency parsing is often combined with transformer embeddings rather than used standalone in modern systems.

---

# 5️⃣ Named Entity Recognition (NER)

## Definition

Named Entity Recognition (NER) identifies and classifies real-world entities in text into predefined categories such as:

- Person  
- Organization  
- Location  
- Date  
- Quantity  

It converts unstructured text into structured information.

---

## 📌 Example

Sentence:

> "In 2024, Elon Musk bought a coffee at Starbucks in Berlin."

Extracted Entities:

- 2024 → DATE  
- Elon Musk → PERSON  
- Starbucks → ORG  
- Berlin → GPE (Geopolitical Entity)

---

## 🎯 Why It Matters

- Extracts important information automatically  
- Improves search precision  
- Enables structured data generation  

---

## 📍 Where It Is Used

- Search Engines  
- Customer Support Automation  
- Resume Screening  
- Sentiment Analysis  
- Enterprise Data Pipelines  

---

# 6️⃣ Word Sense Disambiguation (WSD)

## Definition

Word Sense Disambiguation (WSD) identifies the correct meaning of a word that has multiple meanings, based on context.

---

## 📌 Example

Word: *bank*

- "She sat near the river bank." → river edge  
- "She deposited money in the bank." → financial institution  

WSD selects the correct meaning using context.

---

## 🎯 Why It Matters

- Prevents translation errors  
- Improves search accuracy  
- Enhances question-answering systems  

---

# 7️⃣ Anaphora Resolution

## Definition

Anaphora Resolution identifies what a pronoun refers to in a text.

---

## 📌 Example

Text:

> "Sanjay went to the store. He bought apples."

"He" → refers to "Sanjay"

---

## 🎯 Why It Matters

- Maintains context across sentences  
- Essential for:
  - Chatbots
  - Question Answering
  - Summarization
  - Machine Translation  

---

## 🧠 Interview Insight

- Part of Coreference Resolution  
- Can be rule-based, ML-based, or deep learning-based  

---

# 8️⃣ Coreference Resolution

## Definition

Coreference Resolution identifies when two or more expressions refer to the same real-world entity.

---

## 📌 Example

> "Sanjay said he would finish his work."

- Sanjay  
- he  
- his  

All refer to the same entity.

---

## 🎯 Why It Matters

- Enables document-level understanding  
- Improves discourse coherence  
- Critical for advanced NLP systems  

---

# 🧠 Final Summary

Linguistic Processing helps NLP systems move from:

Words → Phrases → Structure → Relationships → Meaning → Context

It bridges the gap between basic preprocessing and deep semantic understanding.


# The Hashing Trick (Feature Hashing)

## Definition

The Hashing Trick is a technique used in Natural Language Processing (NLP) to convert text into fixed-size numerical vectors using a hash function — without explicitly storing a vocabulary.

It is also known as **Feature Hashing**.

Instead of maintaining a growing dictionary of words, features are directly mapped into a fixed-length vector space using hashing.

---

# ⚙ How It Works

1. Tokenize the text into words or features.
2. Apply a hash function to each token.
3. Map the hash value to an index in a fixed-size vector.
4. Update the value at that index (e.g., increment frequency or apply TF-IDF weighting).

---

## 📌 Example

Assume vector size = 5

Sentence:
> "cat dog cat"

Hash results (example):
- hash("cat") → index 2  
- hash("dog") → index 4  

Final vector:
[0, 0, 2, 0, 1]

- Index 2 = 2 occurrences of "cat"
- Index 4 = 1 occurrence of "dog"

---

# 🎯 Why We Use the Hashing Trick

- Avoids storing large vocabularies  
- Reduces memory consumption  
- Enables fast and scalable processing  
- Works well for streaming data  

Especially useful when:
- Dataset is extremely large  
- Vocabulary is dynamic  
- Memory constraints exist  

---

# ✅ Advantages

- Fixed-size vector representation  
- Memory efficient  
- Fast computation  
- Scales well to massive datasets  
- No need to update vocabulary dictionary  

---

# ⚠ Limitations

- Hash collisions may occur  
  (Different words may map to the same index)

- Loss of interpretability  
  (Cannot directly know which word maps to which index)

- Irreversible transformation  
  (Original words cannot be recovered from hashed vectors)

---

# 🧠 When to Use It

- Large-scale machine learning systems  
- Online learning / streaming pipelines  
- High-dimensional sparse text data  
- When vocabulary storage is impractical  

---

# 🔥 Interview Insight

Hashing Trick is commonly compared with:

- Bag of Words (BoW)
- TF-IDF
- CountVectorizer

Key difference:
Hashing Trick trades interpretability for scalability and memory efficiency.

# 🟢 Bag of Words (BoW)

Bag of Words (BoW)
is a text representation technique in Natural Language Processing where a document is represented as a bag (collection) of words, ignoring grammar and word order, and keeping only word frequency.

OR

Bag of Words is a text representation technique that converts documents into numerical vectors based on word frequency, ignoring grammar and word order.

---

## 🔹 How it works (Simple)

Create a vocabulary of all unique words from the corpus

Count how many times each word appears in a document

Represent each document as a vector of word counts

---

## 🔹 Mathematical View

If vocabulary size = V  
Each document is represented as a vector of length V.

BoW(d) = [count(w1), count(w2), ..., count(wV)]

---

## 🔹 Example

Sentences:

“I love NLP”

“I love AI”

Vocabulary: [I, love, NLP, AI]

Vectors:

“I love NLP” → [1, 1, 1, 0]

“I love AI” → [1, 1, 0, 1]

---

## 🔹 Why we use BoW (2 lines)

BoW is used because it is simple, fast, and easy to implement.  
It converts text into numerical form, making it suitable for machine learning algorithms.

---

## 🔹 When we use BoW (2 lines)

BoW is used in basic NLP tasks like text classification and sentiment analysis.  
It is suitable when word frequency matters more than word order or context.

---

## 🔹 Advantages

- Easy to understand
- Simple implementation
- Works well for small datasets
- Good baseline model

---

## 🔹 Limitations

- Ignores grammar and word order
- No semantic understanding
- High dimensional sparse vectors
- Cannot handle synonyms properly

---

# 🟢 N-grams

N-grams
are contiguous sequences of N words or tokens extracted from a text, used to capture local word order and context in Natural Language Processing.

---

## 🔹 Types of N-grams

Unigram (N=1): single word

Bigram (N=2): two consecutive words

Trigram (N=3): three consecutive words

---

## 🔹 Example

Sentence: “I love NLP”

Unigrams: [I, love, NLP]

Bigrams: [I love, love NLP]

Trigrams: [I love NLP]

---

## 🔹 How N-grams work

Split text into tokens

Generate sequences of N consecutive tokens

Convert them into numerical features (often using BoW or TF-IDF)

---

## 🔹 Mathematical Representation

For a sentence with T tokens:

Number of N-grams = T - N + 1

---

## 🔹 Why we use N-grams (2 lines)

N-grams help capture word order and local context that Bag of Words misses.  
They improve model understanding of phrases and common patterns.

---

## 🔹 When we use N-grams (2 lines)

N-grams are used when phrases matter, such as sentiment or intent detection.  
They are effective in text classification and language modeling tasks.

---

## 🔹 Limitations

High dimensionality for large N

Data sparsity problem

Cannot capture long-range dependencies

Computationally expensive for large datasets

---

# 🟢 TF-IDF (Term Frequency–Inverse Document Frequency)

TF-IDF
is a statistical method used in natural language processing and information retrieval to evaluate how important a word is to a document in relation to a larger collection of documents.

TF-IDF combines two components:

TF (Term Frequency)

IDF (Inverse Document Frequency)

---

## 🔹 Formula

TF(t,d) = (Number of times term t appears in document d) / (Total terms in document d)

IDF(t) = log ( Total number of documents / Number of documents containing term t )

TF-IDF(t,d) = TF(t,d) × IDF(t)

---

## 🔹 Example (Simple)

Documents:

D1: “I love NLP”

D2: “I love AI”

Common word “love” appears in both → low IDF  
Unique word “NLP” appears once → high IDF

---

## 🔹 Why we use TF-IDF (2 lines)

TF-IDF highlights important and unique words in a document.  
It reduces the effect of frequently occurring but less informative words.

---

## 🔹 When we use TF-IDF (2 lines)

Used when word importance matters, not just frequency.  
Commonly applied in text classification, search engines, and document similarity.

---

## 🔹 Advantages

- Reduces impact of stopwords
- Highlights rare but important words
- Better than plain BoW in many tasks

---

## 🔹 Limitations

Ignores word order and context

Cannot capture semantic meaning

High dimensional for large vocabularies

---

# 🟣 One-Hot Encoding

One-Hot Encoding
is a technique to convert categorical variables into a binary vector representation.

Each category is represented by a vector of length equal to the number of unique categories, with 1 for the category present and 0 for all others.

---

## 🔹 How it works

Identify all unique categories in a feature

Create a binary vector for each category

Mark 1 for the active category and 0 for others

Color   Red   Blue   Green  
Red     1     0      0  
Blue    0     1      0  
Green   0     0      1  

---

## 🔹 Why we use One-Hot Encoding (2 lines)

Machine learning models cannot work directly with categorical data, so one-hot encoding converts it into numerical form.  
It prevents models from assuming ordinal relationships among categories.

---

## 🔹 When we use One-Hot Encoding (2 lines)

Used when categorical features have no intrinsic order (nominal variables).  
Commonly applied in classification, regression, and neural networks.

---

## 🔹 Limitations

Increases dimensionality for features with many categories

Sparse representation can be memory inefficient

Not suitable for high-cardinality categorical features (use embeddings instead)

---

# 🟣 Word Embeddings

Word embeddings
are dense vector representations of words in a continuous vector space.

They capture semantic and syntactic meaning, so words with similar meanings are closer in the vector space.

---

## 🔹 How it works

Each word is mapped to a fixed-size dense vector (usually 50–300 dimensions)

Trained on large corpora using methods like Word2Vec, GloVe, or FastText

Word similarity is captured using cosine similarity

---

## 🔹 Example

Words like “king”, “queen”, “man”, “woman”

Vector(king) − Vector(man) + Vector(woman) ≈ Vector(queen)

Shows semantic relationships captured in vector space

---

## 🔹 Why we use Word Embeddings (2 lines)

They capture semantic meaning and relationships between words, unlike Bag of Words or TF-IDF.  
They reduce dimensionality and improve performance of NLP models.

---

## 🔹 When we use Word Embeddings (2 lines)

Used in deep learning NLP tasks like text classification, sentiment analysis, machine translation.  
Essential when context and word meaning matter, not just frequency.

---

## 🔹 Applications

Text classification

Sentiment analysis

Machine translation

Named Entity Recognition (NER)

Chatbots and question answering

---

## 🔹 Limitations

Requires large corpus for training

Pre-trained embeddings may not fit domain-specific words

Static embeddings don’t handle polysemy (same word, multiple meanings)

---

# 🟢 Word2Vec

Word2Vec
is a neural network-based model that converts words into dense vectors (word embeddings), capturing semantic and syntactic relationships.

It places similar words close together in a continuous vector space.

---

## 🔹 Key Architectures

CBOW (Continuous Bag of Words):

Predicts a word based on its context words.

Faster and works well for frequent words.

Skip-gram:

Predicts context words given a target word.

Performs better on rare words and captures richer semantic information.

---

## 🔹 Example

Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")

---

## 🔹 Why Word2Vec?

Captures semantic meaning and relationships between words.

Generates dense, low-dimensional vectors suitable for ML models.

---

## 🔹 Limitations

Static embeddings: vector for a word is the same in all contexts

Needs large datasets for good quality vectors

Does not capture long-range context

---

# 🟢 CBOW

CBOW
is a Word2Vec architecture that predicts the current (target) word from its surrounding context words.

---

## 🔹 How it works

Take a window of context words around the target word

Use the context to predict the target word

Update word vectors through a neural network

---

## 🔹 Example

Sentence: “I love natural language processing”

Context words = [“I”, “love”, “language”, “processing”]

Target word = “natural”

CBOW predicts “natural” from its context.

---

## 🔹 Characteristics

Faster than Skip-gram

Performs well on frequent words

Context size can be adjusted

---

# 🟢 Skip-gram

Skip-gram
is a Word2Vec architecture that predicts the context words surrounding a given target word.

---

## 🔹 How it works

Take a target word from the text

Use it to predict all words within a context window

Train a neural network to update word vectors

---

## 🔹 Example

Sentence: “I love natural language processing”

Target word = “language”

Context window = 2 words

Context words = [“natural”, “processing”]

Skip-gram predicts “natural” and “processing” from “language”.

---

## 🔹 Characteristics

Works better for rare words than CBOW

Slower to train compared to CBOW

Captures semantic and syntactic relationships effectively

---

## 🔹 Why we use Skip-gram (2 lines)

# 🟢 GloVe – NLP (Global Vectors for Word Representation)

GloVe (Global Vectors for Word Representation)
is a word embedding technique that generates dense vector representations of words by analyzing global word co-occurrence statistics from a corpus.

It captures semantic and syntactic relationships between words in a vector space.

GloVe was developed by Stanford University researchers.

---

## 🔹 How it Works

Construct a word-word co-occurrence matrix from the corpus.

Each element X(i,j) represents how often word j appears in the context of word i.

Learn word vectors by factorizing this matrix so that the dot product of two word vectors approximates the logarithm of their co-occurrence probability.

Words appearing in similar contexts will have similar vector representations.

---

## 🔹 Mathematical Intuition

GloVe optimizes the following objective:

w_i^T w_j + b_i + b_j ≈ log(X_ij)

Where:

- w_i, w_j → word vectors  
- b_i, b_j → bias terms  
- X_ij → co-occurrence count of word i and j  

The model minimizes the difference between predicted and actual co-occurrence statistics.

---

## 🔹 Example

Semantic relationships captured:

Vector("King")
−
Vector("Man")
+
Vector("Woman")
≈
Vector("Queen")

Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")

Similar to Word2Vec, but GloVe uses global corpus statistics rather than local context windows.

---

## 🔹 GloVe vs Word2Vec

Word2Vec:
- Uses local context windows
- Predictive model (neural network)

GloVe:
- Uses global co-occurrence matrix
- Count-based model with matrix factorization

Both generate dense word embeddings but use different training strategies.

---

## 🔹 Why we use GloVe (2 lines)

It captures semantic meaning of words using corpus-wide statistics.  
Produces dense, low-dimensional vectors for downstream NLP tasks.

---

## 🔹 When we use GloVe (2 lines)

Used in text classification, sentiment analysis, semantic similarity, and NLP deep learning models.  
Ideal when we need pre-trained embeddings for general-purpose NLP.

---

## 🔹 Advantages

- Captures global statistical information
- More stable training compared to pure predictive models
- Efficient for large corpora
- Well-known pre-trained embeddings available (50d, 100d, 200d, 300d)

---

## 🔹 Limitations

Static embeddings – cannot handle context-dependent meanings (use BERT for contextual embeddings)

Requires large corpus for high-quality embeddings

Less flexible for domain-specific vocabulary unless retrained

High memory usage due to co-occurrence matrix

---

## 🔹 Applications

Text classification

Sentiment analysis

Semantic similarity

Question answering systems

Chatbots

Named Entity Recognition (NER)

# 🟢 FastText

FastText
is a word embedding technique developed by Facebook that represents words as vectors of character n-grams.

It captures subword information, allowing it to handle rare words, misspellings, and morphologically rich languages effectively.

FastText extends Word2Vec by incorporating character-level information.

---

## 🔹 How it Works

Break each word into character-level n-grams.

Example: for “apple”, bi-grams: “ap”, “pp”, “pl”, “le”

Represent a word as the sum (or average) of its n-gram vectors.

Train embeddings using a Skip-gram-like or CBOW model.

The resulting word vector captures both word-level and subword-level information.

---

## 🔹 Mathematical View

Word Vector(w) = Sum of all character n-gram vectors of w

If G(w) = set of n-grams in word w,

V(w) = ∑ g ∈ G(w) Zg

Where Zg = vector representation of n-gram g.

---

## 🔹 Example

Word: “apple”

Character bi-grams: [ap, pp, pl, le]

Word vector = sum of its n-gram vectors

FastText can generate embeddings for:

- Unseen words like “apples”
- Misspelled words like “applle”

Because subword components are known.

---

## 🔹 Why we use FastText (2 lines)

It handles rare and out-of-vocabulary (OOV) words using subword information.  
Produces robust embeddings for morphologically rich languages.

---

## 🔹 When we use FastText (2 lines)

Used in text classification, sentiment analysis, and NLP tasks with OOV words.  
Effective when working with large vocabularies and rare words.

---

## 🔹 Advantages

- Handles OOV words
- Works well for morphologically rich languages
- More robust than Word2Vec for rare words
- Captures prefix/suffix meaning

---

## 🔹 Limitations

Slightly more computationally expensive than Word2Vec

Still produces static embeddings, so context-dependent meanings are not captured

Larger memory usage compared to Word2Vec

# 🟢 Contextual Word Embeddings

Contextual word embeddings
are dynamic word representations that change depending on the context in which a word appears.

Unlike static embeddings (Word2Vec, GloVe, FastText), the same word can have different vectors in different sentences.

OR

Contextual word embeddings are dynamic word vectors that change based on the surrounding context, capturing the meaning of words in different sentences.

---

## 🔹 How it Works

Uses deep learning models like:

- ELMo  
- BERT  
- GPT  

Each word’s embedding is computed based on the entire sentence or surrounding context.

Typically built using Transformer architectures and attention mechanisms.

Captures polysemy (same word, multiple meanings) and nuanced semantics.

---

## 🔹 Example

Word: “bank”

Sentence 1:  
“I went to the bank to deposit money”  
→ embedding reflects financial institution

Sentence 2:  
“The river bank was full of trees”  
→ embedding reflects river edge

The same word gets two different embeddings.

---

## 🔹 Why we use Contextual Word Embeddings (2 lines)

They capture word meaning based on context, solving limitations of static embeddings.  
Essential for understanding polysemy and complex language structures.

---

## 🔹 When we use Contextual Word Embeddings (2 lines)

Used in modern NLP tasks like question answering, machine translation, sentiment analysis, and NER.  
Critical when word meaning changes depending on context.

---

## 🔹 Advantages

- Handles polysemy
- Captures deep contextual meaning
- State-of-the-art performance
- Works well for complex NLP tasks

---

## 🔹 Limitations

Computationally expensive

Requires large pre-trained models and GPU for training

Higher memory requirements

Longer inference time compared to static embeddings

---

# 🟢 Interview One-Liner – Sentence Embeddings

Sentence embeddings
are dense vector representations of entire sentences, capturing semantic meaning beyond individual words.

They encode the overall context and meaning of a sentence in a fixed-length vector.

---

## 🔹 How it Works

Use pre-trained models like Universal Sentence Encoder (USE), Sentence-BERT, or InferSent.

Encode the sentence into a single dense vector.

Vectors can be compared using cosine similarity or other distance metrics.

---

## 🔹 Example

Sentence 1:  
“I love machine learning.”

Sentence 2:  
“Machine learning is my passion.”

Their vectors are close in vector space, reflecting similar meaning.

---

## 🔹 Why we use Sentence Embeddings (2 lines)

To capture the meaning of an entire sentence, not just individual words.  
Useful for semantic similarity, clustering, and information retrieval.

---

## 🔹 Limitations

Loss of word-level nuance: Aggregating sentence meaning may ignore subtle word-level differences.

Context dependence: Some models may fail for very long or complex sentences.

Data requirements: Pre-trained models may not work well for domain-specific language.

Computational cost compared to simpler techniques like TF-IDF.

# Evaluation Metrics for NLP

---

## Accuracy

Accuracy evaluates how well a machine learning model performs. It represents the percentage of correct predictions made by the model. While simple to calculate and understand, accuracy is most effective when the dataset is balanced.

### How to Measure Accuracy?

Accuracy is easy to understand and works well when the data is balanced (i.e., all classes in the dataset are equally represented). However, accuracy isn’t always the best measure of performance.

In datasets where one class dominates (e.g., 95% "negative" and 5% "positive"), a model predicting only the majority class would have high accuracy but wouldn’t solve the problem effectively.

To calculate accuracy:

- Compare the model's predictions to the actual values  
- Count how many predictions were correct  
- Divide that by the total number of predictions  

Breakdown:

- **True Positives (TP):** Correctly predicted as positive  
- **True Negatives (TN):** Correctly predicted as negative  
- **False Positives (FP):** Predicted positive but actually negative  
- **False Negatives (FN):** Predicted negative but actually positive  

### Formula

Accuracy = (TP + TN) / (TP + TN + FP + FN)

### Practical Example

If a model is tested on 100 cases and identifies 85 correctly:

Accuracy = 85 / 100 = 85%

---

## Precision

Precision is the ratio between the True Positives and all the Positives predicted by the model. It measures how many of the "yes" predictions were actually correct. It is a key metric for reducing False Positives (FP).

### Formula

Precision = TP / (TP + FP)

### Example: Bird Detection

Imagine you build a model to find birds in photos.

- **True Positive (TP):** The model marks a photo as a "bird," and there is actually a bird in it.  
- **False Positive (FP):** The model marks a photo as a "bird," but it is actually a cat.  

### When to Use Precision

Precision is vital when the cost of a false alarm is high.

**Spam Detection:**  
You don't want a real email from your boss to be sent to the Spam folder (False Positive). You would rather see a bit of spam in your inbox than miss an important message.

### Advantage

High precision means the model is trustworthy when it says "yes."

### Limitation

A model can be 100% precise by making only one prediction but missing many actual cases (low recall).

---

## Recall

Recall tells us how well a model finds all the actual positive cases in the data. It measures how many real "yes" cases the model correctly identified out of everyone who actually should have been a "yes."

### Formula

Recall = TP / (TP + FN)

### Example: Bird Detection

- **True Positive (TP):** The model finds the bird in the photo.  
- **False Negative (FN):** There is a bird in the photo, but the model says "no bird."  

### When to Use Recall

Recall is vital when missing a case is dangerous or costly.

**Medical Diagnosis:**  
If a patient has a serious illness, it is better to flag them for more tests than to tell a sick person they are fine.

**Fraud Detection:**  
It is better to temporarily block a real credit card transaction than to let a thief steal money.

### Advantage

High recall ensures that very few actual cases are missed.

### Limitation

Trying to catch everything may increase False Positives (low precision).

---

## F1-Score

The F1-Score balances Precision and Recall. It is the harmonic mean of Precision and Recall.

### Formula

F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

### Why Use the Harmonic Mean?

- Penalizes extreme differences between precision and recall  
- Better suited for handling rates  
- Provides better insight for imbalanced datasets  

---

## Confusion Matrix

A Confusion Matrix is a summary table used to measure the performance of a classification model. It compares predictions against actual ground truth.

### The Four Categories

- **True Positive (TP)** – Correct positive  
- **True Negative (TN)** – Correct negative  
- **False Positive (FP)** – Type I Error  
- **False Negative (FN)** – Type II Error  

### Why Is This Used?

- Spot specific errors  
- Detect bias  
- Handle imbalanced data  
- Assist risk management decisions  

### Limitations

- Difficult to interpret for multi-class classification  
- Shows performance at only one threshold  

---


## BLEU

is an evaluation metric designed to measure the quality of text that has been machine-translated from one natural language to another.

How it Works

The score is calculated by comparing n-grams (sequences of $n$ words) of the machine-translated output to a set of high-quality human reference translations. It essentially measures Precision: "How many of the words/phrases the model generated actually appear in the human reference?"

The FormulaThe BLEU score

is the product of a Brevity Penalty ($BP$) and the geometric mean of the modified n-gram precisions:$$BLEU = BP \cdot \exp\left(\sum_{n=1}^{N} w_n \log p_n\right)$$$p_n$: Precision for n-grams.$BP$: Prevents the model from getting a high score by outputting very short, "safe" sentences.

Why is it used?   
Speed: It is much faster and cheaper than hiring humans to grade every translation.

Benchmarking: It allows researchers to compare different models (like Transformer vs. RNN) on the same dataset.

Limitations   
Meaning vs. Words: It does not account for synonyms. If a model uses "huge" instead of the reference word "large," the BLEU score drops, even though the meaning is identical.

Fluency: A sentence can have a high BLEU score but still be grammatically awkward or nonsensical to a native speaker.

## ROUGE

## ROUGE: Standard Metric for Text Summarization

ROUGE is the standard metric for evaluating **Text Summarization**. Since a summary is usually much shorter than the original text, ROUGE focuses on how much of the **essential information** from the human reference was captured by the model.


## Key ROUGE Metrics

- **ROUGE-N:** Measures the n-gram overlap.  
  - ROUGE-1 refers to unigrams (individual words).  
  - ROUGE-2 refers to bigrams (word pairs).  

- **ROUGE-L:** Uses the **Longest Common Subsequence (LCS)**.  
  - Identifies the longest sequence of words that appear in both the reference and the candidate in the same relative order.  

- **ROUGE-W:** A weighted version of LCS that gives more credit to consecutive word matches.  

- **ROUGE-S:** Measures **Skip-bigrams**.  
  - Allows for matches between word pairs even if they are separated by other words in the sentence.

## Formulas

### 1. ROUGE-N (Recall focus)

The most common version of ROUGE-N is calculated as follows:

$$
ROUGE\text{-}N = \frac{\sum_{S \in \{References\}} \sum_{n\text{-}gram \in S} Count_{match}(n\text{-}gram)}{\sum_{S \in \{References\}} \sum_{n\text{-}gram \in S} Count(n\text{-}gram)}
$$

**In simpler terms:**

$$
ROUGE\text{-}N = \frac{\text{Number of matching n-grams}}{\text{Total n-grams in the reference}}
$$


### 2. ROUGE-L (The F-measure)

To ensure a balance between the length of the summary and its accuracy, ROUGE-L is often calculated as an F-score:

$$
ROUGE\text{-}L = \frac{(1 + \beta^2) \cdot P \cdot R}{\beta^2 \cdot P + R}
$$

Where:  
- **P (Precision):** How much of the model's summary was relevant?  
- **R (Recall):** How much of the human reference was captured?  
- **β:** A parameter to weight importance. Typically, **β = 1**, which makes it a standard F1-score.


## Why is it used?

- **Summarization Tasks:** It is the primary metric for news summaries, meeting notes, and abstract generation.  
- **Information Density:** Rewards models that don't leave out crucial details.


## Limitations

- **Word-for-Word Matching:** Like BLEU, it struggles with synonyms.  
  - If the model paraphrases a sentence perfectly using different words, ROUGE will give it a low score.  
- **Coherence:** A model could extract all the right keywords but put them in a confusing order; ROUGE (especially ROUGE-N) might still give it a high score.


---

## Perplexity

Perplexity is a metric used to evaluate **language models**. It measures how well a model predicts a sequence of words.

### Key Points

- **Definition:** Perplexity is the exponentiation of the average negative log-likelihood of a sequence.  
- **Interpretation:** Lower perplexity means the model is better at predicting the next word.  
- **Use Cases:** Commonly used in evaluating models like GPT, LSTM, and other NLP language models.

### Formula

For a sequence of words \( w_1, w_2, ..., w_N \) with probability \( P(w_i) \) predicted by the model:

$$
PP(W) = P(w_1, w_2, ..., w_N)^{-\frac{1}{N}} = \exp \left( -\frac{1}{N} \sum_{i=1}^{N} \log P(w_i) \right)
$$

**Interpretation:**  
- **Lower PPL:** Model predictions are more confident and accurate.  
- **Higher PPL:** Model is uncertain about the next word.

### Why it is used

- Evaluates **language model quality** in predicting sequences.  
- Useful for comparing different models on the same dataset.


---

## Word Error Rate (WER)

WER is a metric used to evaluate **automatic speech recognition (ASR)** systems. It measures the difference between a predicted transcript and a reference transcript.

### Key Points

- **Definition:** WER calculates the minimum number of word-level edits needed to change the hypothesis into the reference.  
- **Components:**  
  - **Substitutions (S):** Wrong words replaced.  
  - **Insertions (I):** Extra words added.  
  - **Deletions (D):** Words missing from the prediction.  

### Formula

$$
WER = \frac{S + D + I}{N}
$$

Where:  
- **N:** Total words in the reference transcript.  
- **S:** Substitutions  
- **D:** Deletions  
- **I:** Insertions

### Why it is used

- Measures **accuracy of speech-to-text systems**.  
- Provides a standardized way to compare ASR systems.  
- Lower WER indicates better transcription quality.



### Example

Reference: "I love natural language processing"  
Hypothesis: "I love natural processing"

- **Substitutions (S):** 0  
- **Deletions (D):** 1 ("language" is missing)  
- **Insertions (I):** 0  
- **WER:** \( \frac{0 + 1 + 0}{5} = 0.2 \) (20% error)

| Complexity            | Simpler compared to chunking.                                                                             | Slightly more complex because it uses POS tags to build phrases.                           |
| Dependency            | Independent task.                                                                                         | Usually depends on POS tagging results.                                                    |

| Aspect                   | Sparse Vectorization                                                  | Dense Vectorization                                                                 |
| ------------------------ | --------------------------------------------------------------------- | ----------------------------------------------------------------------------------- |
| Definition               | A text representation method where most vector values are **zero**.   | A text representation method where most vector values are **non-zero and compact**. |
| Core Idea                | Represents text using large vocabulary-based vectors with many zeros. | Represents text using low-dimensional embeddings that capture semantic meaning.     |

| Value Distribution       | Mostly zeros, few non-zero values.                                    | Mostly meaningful floating-point values.                                            |

| Interpretability         | Easier to interpret (based on word frequency).                        | Harder to interpret (abstract learned features).                                    |


| Aspect                        | GloVe (Global Vectors)                                                                              | FastText                                                                                                          |
| ----------------------------- | --------------------------------------------------------------------------------------------------- | ----------------------------------------------------------------------------------------------------------------- |
| Developed By                  | Stanford University (2014)                                                                          | Facebook AI Research (2016)                                                                                       |
| Core Idea                     | Count-based model that generates word embeddings by factorizing a global word co-occurrence matrix. | Predictive model that represents words as **bags of character n-grams**, generating embeddings for subwords.      |
| Word Representation           | Each word has a **single embedding vector**.                                                        | Words are represented as the sum of their **subword (character n-gram) embeddings**.                              |
| Handling Rare Words           | Rare words may have poor embeddings if not frequent in corpus.                                      | Handles rare and unseen words effectively because it uses subword information.                                    |
| Context Consideration         | Focuses on **global co-occurrence statistics**.                                                     | Focuses on **local context prediction**, similar to Word2Vec, but adds subword modeling.                          |
| Training Approach             | Matrix factorization of co-occurrence matrix.                                                       | Predictive, similar to Skip-Gram/CBOW but on character n-grams.                                                   |
| Semantic Relationships        | Captures semantic similarity between frequent words.                                                | Captures semantic similarity and also morphological relationships (prefixes, suffixes).                           |
| Out-of-Vocabulary (OOV) Words | Cannot handle OOV words; embeddings missing for unseen words.                                       | Can generate embeddings for OOV words using subword information.                                                  |
| Computational Requirement     | Needs to build and factorize large co-occurrence matrix; moderate memory usage.                     | Computationally efficient; requires training predictive model on n-grams.                                         |
| Dimensionality                | Fixed-size embedding per word.                                                                      | Fixed-size embedding per word, but computed from n-grams.                                                         |
| Example                       | “king − man + woman ≈ queen”                                                                        | “running” shares similarity with “run”, “runner”, “ran” due to n-grams.                                           |
| Best Use Case                 | Static embeddings for large corpora where morphology is less important.                             | Morphology-rich languages, rare word handling, text classification, NLP applications requiring robust embeddings. |
| Popular Models                | Pre-trained GloVe embeddings (Common Crawl, Wikipedia).                                             | FastText pre-trained embeddings (Wikipedia, Common Crawl).                                                        |
