# Natural Language Processing (NLP) Notes

## What is Natural Language Processing
Natural Language Processing (NLP) is a field that combines computer science, artificial intelligence, and linguistics. It enables computers to understand, process, and generate human language in a meaningful and useful way.


## Phases of NLP
The phases of NLP are the sequential steps computers use to understand human language. These phases include lexical analysis (breaking text into words), syntactic analysis (checking grammar and sentence structure), semantic analysis (understanding meaning), and pragmatic analysis (understanding context and intention). Together, these steps help machines accurately understand and respond to human language.


## Lexical Analysis
Lexical analysis is the first step in NLP where text is divided into meaningful units called tokens and classified. It involves scanning, tokenization, and normalization — such as removing whitespace and converting to lowercase — to transform raw text into structured data (lexemes) for further analysis.
Key tasks: Tokenization and Part-of-Speech (POS) Tagging.

**Why:** Lexical analysis enables machines to convert unstructured human language into structured, manageable components, preparing the text for deeper stages such as syntax analysis and semantic analysis.

## Morphological Analysis
Morphological analysis is the process of analyzing the internal structure of words by breaking them down into their smallest meaningful units, called morphemes. It identifies how roots, prefixes, and suffixes combine to create a word's specific meaning and grammatical function.
Key tasks: Stemming and Lemmatization.

**Why:** Without morphological analysis, a computer would treat "wait," "waiter," and "waiters" as completely unrelated strings. With it, the computer recognizes they all share the same core concept.

## Syntactic Analysis (Parsing)
Syntactic analysis is the process of analyzing the grammatical structure of a sentence. It examines how words are arranged according to grammar rules and represents this structure using a parse tree, showing relationships such as subject, verb, and object.
Key components: POS Tagging and Ambiguity Resolution.

**Why:** Syntactic analysis helps machines correctly understand sentence meaning, improve grammar checking, and support tasks like translation, question answering, and text understanding.

## Semantic Analysis
Semantic analysis is the process of understanding the meaning of words, phrases, and sentences in context, ensuring that text is meaningful, logically correct, and contextually appropriate. It focuses on understanding relationships between words and identifying the actual meaning of a sentence.
Key tasks: Named Entity Recognition (NER) and Word Sense Disambiguation (WSD).

**Why:** Semantic analysis ensures logical correctness beyond grammatical structure. For example, the expression `total = "Price" * 10` is grammatically perfect but semantically invalid, as you cannot multiply a string by a number.

## Discourse Integration
Discourse integration is the process of understanding how sentences or parts of a text are connected and relate to each other. It ensures that meaning is consistent and coherent across the whole text, especially in longer or complex passages.
Key tasks: Anaphora Resolution and Contextual References.

**Why:** Discourse integration allows chatbots to remember prior context, helps AI create consistent summaries, and ensures pronouns and references stay accurate across an entire translated document.

## Pragmatic Analysis
Pragmatic analysis is the process of understanding the deeper meaning of words and sentences beyond their literal semantic meaning. It focuses on the speaker's or writer's intentions, tone, and context.
Key tasks: Understanding Intentions and Figurative Meaning.

**Why:** Pragmatic analysis allows AI tools like Siri or ChatGPT to understand that "Call me a cab" means "order a taxi," not "rename me 'a cab'." It also helps sentiment analysis distinguish a genuinely happy customer from a sarcastic one.


# Text Preprocessing

## Text Cleaning
Text cleaning is the preprocessing step where unwanted characters, noise, and inconsistencies are removed from text to improve NLP model performance. Common operations include removing extra whitespace, lowercasing, removing HTML tags, URLs, email addresses, special characters, numbers (when not required), punctuation, and emojis. Additional steps include expanding contractions (e.g., don't → do not), removing duplicates, and handling misspellings.


## Tokenization
Tokenization is the process of breaking text into smaller units called tokens. Tokens can be words, phrases, subwords, or individual characters, depending on the level of detail required.

**Types:**
- **Word Tokenization:** Breaks text into words. Works well for English and similar languages.
- **Character Tokenization:** Breaks text into single letters. Useful for spelling tasks or languages without clear word boundaries.
- **Subword Tokenization:** Breaks text into pieces smaller than a word but larger than a letter.
- **Sentence Tokenization:** Splits paragraphs into separate sentences.
- **N-gram Tokenization:** Groups words into sequences of n items to find patterns.

**Why:** Tokenization breaks text into smaller units, making it easier for machines to process and analyze. Without it, models would struggle to identify individual words, patterns, or relationships needed for tasks like translation, search, or sentiment analysis.


## Text Corpus
A text corpus is a large collection of structured or unstructured text used to train and evaluate NLP systems. It provides real language data that helps models learn vocabulary, grammar, and context patterns.

**Types of Text Corpus:**
- **Monolingual Corpus:** Single language
- **Multilingual Corpus:** Multiple languages
- **Parallel Corpus:** Same content in multiple languages (used in translation)
- **Annotated Corpus:** Labeled with POS, NER, etc.
- **Unannotated Corpus:** Raw text only

**Example:** For sentiment analysis, a corpus of movie reviews labeled as positive or negative is used to train the model, enabling it to learn sentiment patterns and predict sentiment for new reviews.


## Sentence Segmentation
Sentence segmentation is the process of dividing a text into individual sentences, helping NLP systems understand where one sentence ends and another begins.

**Example:**
Input: *"I love NLP. It is very interesting!"*
Output: `["I love NLP.", "It is very interesting!"]`

**Why:** NLP models work better when text is processed sentence by sentence rather than as one continuous block.


## Stop Words
Stop words are frequently occurring grammatical words — such as articles, prepositions, conjunctions, and pronouns — that help form sentences but add little semantic meaning. They are often removed in NLP to focus on important words, improve accuracy, and reduce computational cost.

**Why:** In search engines, removing stop words helps the system focus on meaningful query terms.
Example: `"How to bake a cake in London"` → Stop words removed → Keywords: `[bake, cake, London]`


## Stemming
Stemming is an NLP technique that converts words into their base or root form by removing suffixes or prefixes, helping to reduce word variations and simplify text analysis.

**Why:** Stemming reduces different forms of a word to a common base form, decreasing vocabulary size, improving processing speed, and allowing models to focus on the core meaning of words.

**Limitations:**
- **Over-stemming:** "Universal," "University," and "Universe" may all reduce to "Univers" despite having different meanings.
- **Nonsense roots:** "Business" may be reduced to "Busin," which is not a real word.

**Use Cases:** Search engines (broadening query results), spam filters (catching word variations), and sentiment analysis (grouping related word forms).

## Lemmatization
Lemmatization is an NLP preprocessing technique that converts words into their correct dictionary base form (lemma) by considering the word's meaning and part of speech, producing more accurate results than stemming.

**Why:** Lemmatization preserves the true meaning of a word by considering its context and part of speech. For example, it correctly identifies that "saw" is a tool (noun) in one sentence and the past tense of "see" (verb) in another.

**Use Cases:** Chatbots, sentiment analysis, and medical or legal search where precise terminology is required.


## Handling Punctuation
Handling punctuation involves removing symbols like commas and exclamation marks that do not add meaning, reducing noise and improving text consistency.

## Handling Numbers
Handling numbers involves removing or converting numeric values based on their importance in context. Numbers are retained when they carry useful information, such as prices or dates.

## Handling Special Characters
Special characters such as `@`, `#`, and `$` are cleaned to remove unwanted noise, making the text easier for models to process.

## Handling Emojis
Handling emojis involves removing or converting emojis into words that represent emotions. This is especially useful in sentiment analysis tasks.


## Spelling Correction
Spelling correction is the process of identifying and fixing misspelled words in text to their correct form.

**Why:** It reduces vocabulary noise, improves feature quality, and helps models understand the intended words, increasing accuracy in search, chatbots, and classification tasks.

**When:** Applied to user-generated text (chat, reviews, social media), search engines, chatbots, and before feature extraction (BoW, TF-IDF).


## Byte Pair Encoding (BPE)
BPE is a subword tokenization technique that splits words into smaller units (subwords) based on frequency, allowing the model to handle rare or unseen words more effectively.

**How it Works:**
1. Start with all characters as tokens.
2. Find the most frequent pair of consecutive tokens in the corpus.
3. Merge this pair into a new token.
4. Repeat until reaching a predefined vocabulary size.

**Example:** Corpus: `"lower"`, `"lowest"` → Initial tokens: `l o w e r`, `l o w e s t` → Merge frequent pairs: `lowe r`, `lowe st`

**Benefits:** Reduces vocabulary size, handles rare words and misspellings via subwords, and prevents `<UNK>` tokens for unseen words.


## OOV (Out-Of-Vocabulary) Handling
OOV words are words not present in the model's training vocabulary. Effective handling is crucial for building robust NLP systems.

**Techniques:**
- **Subword Tokenization (BPE, WordPiece, SentencePiece):** Splits OOV words into known subwords. Example: `"unhappiness"` → `"un"` + `"happi"` + `"ness"`
- **Character-level Embeddings:** Represents words as character sequences, generating embeddings for any string.
- **Unknown Token `<UNK>`:** Replaces all OOV words with a single generic token. Simple but loses semantic information.
- **FastText-style Embeddings:** Uses subword n-grams to compute embeddings for unseen words.

**Why:** Real-world text contains typos, slang, and domain-specific terms. Proper OOV handling ensures the model generalizes well.


## Text Normalization
Text normalization is the process of standardizing text into a consistent format by correcting spelling errors, expanding contractions (e.g., don't → do not), and handling or removing special characters.

**Steps:**
- **Lowercasing:** Converting all text to lowercase to avoid case-sensitive differences.
- **Removing Punctuation:** Eliminating punctuation marks to reduce noise.
- **Removing Numbers:** Removing numeric values when they do not contribute meaningful information.
- **Handling Special Characters & Whitespace:** Removing special symbols and extra spaces for a consistent format.


# Linguistic Processing

## Part-of-Speech (POS) Tagging
POS tagging is the process of automatically assigning grammatical categories — such as noun, verb, adjective, or adverb — to each word in a sentence based on its context and usage.

**Example:**
Sentence: *"She booked a flight yesterday."*
- She → Pronoun
- booked → Verb
- a → Determiner
- flight → Noun
- yesterday → Adverb

**Why:** POS tagging identifies grammatical roles, reducing ambiguity (e.g., whether "book" is a noun or a verb). It improves the performance of downstream tasks like parsing, NER, machine translation, and text summarization.


## Chunking
Chunking is the process of dividing a sentence or a large piece of text into smaller, meaningful units called chunks (usually phrases or segments) using pos tags. 

**Example:** Sentence: *"She bought a red car"*
Result: `[She] [bought] [a red car]` → `a red car` is a noun phrase.

**Why:** Chunking helps NLP systems understand phrase-level structure, which is more informative than individual words. It is used in information extraction, question answering, and syntactic parsing.


## Constituency Parsing
Constituency parsing is the process of breaking a sentence down into phrases (constituents) and showing how they are arranged in a hierarchical tree.

**Example:** Sentence: *"She bought a red car"*
Constituents: `[NP She] [VP bought [NP a red car]]`

**Why:** Constituency parsing helps NLP systems understand the overall grammatical structure of a sentence and how words group together. It is used in syntactic analysis, machine translation, question answering, and text understanding.


## Dependency Parsing
Dependency parsing identifies grammatical relationships between words in a sentence by showing how each word depends on another. It focuses on word-to-word connections rather than phrase structures.

**Why:** Dependency parsing solves the problem of shallow text understanding. Bag-of-Words and embeddings capture what words exist, but not who did what to whom. By modeling grammatical relationships, it enables semantic reasoning, improves information extraction, and reduces ambiguity in downstream NLP tasks.

**Example:** Sentence: *"She bought a red car."*
- `bought → She` (nsubj: subject)
- `bought → car` (dobj: object)
- `car → red` (amod: adjective modifier)

**Use Cases:** Information extraction, knowledge graphs, question answering systems, machine translation, and semantic search.


## Named Entity Recognition (NER)
Named Entity Recognition (NER) is an NLP technique that identifies and classifies important real-world entities in text — such as names of people, places, organizations, dates, and quantities — into predefined categories. It converts unstructured text into structured data for easier machine understanding.

**Example:** *"In 2024, Elon Musk bought a coffee at a Starbucks in Berlin."*
- `2024` → DATE
- `Elon Musk` → PERSON
- `Starbucks` → ORG
- `Berlin` → GPE (Geopolitical Entity)

**Use Cases:** Search engines (precise query results), customer support (tagging products and locations), recruitment (extracting skills from resumes), and sentiment analysis (identifying which entity a review is about).


## Word Sense Disambiguation (WSD)
Word Sense Disambiguation is the NLP process of identifying the correct meaning of a word that has multiple meanings, based on the context in which it appears. It resolves ambiguity by analyzing surrounding words and sentence structure.

**Why:** WSD is required to resolve ambiguity and ensure correct understanding of words. Without it, tasks like translation, search, and question answering may produce incorrect results.

**When:** Applied when a word has multiple possible meanings in a sentence or context, using surrounding words to identify the intended meaning.


## Anaphora Resolution
Anaphora resolution is the NLP task of identifying what a referring expression refers to in a text.

**Example:** *"Sanjay went to the store. He bought some apples."*
→ Anaphora resolution links `"He"` → `"Sanjay"`

**Why:** It is essential for tasks like question answering, machine translation, text summarization, and conversational AI, helping machines understand context across sentences.


## Coreference Resolution
Coreference resolution is the task of finding all words or phrases in a text that refer to the same real‑world entity (like a person, object, or event) and grouping them together.


# Feature Engineering

## Bag of Words (BoW)
Bag of Words is a text representation technique that converts documents into numerical vectors based on word frequency, ignoring grammar and word order.

**How it works:**
1. Create a vocabulary of all unique words from the corpus.
2. Count how many times each word appears in a document.
3. Represent each document as a vector of word counts.

**Example:** Sentences: `"I love NLP"` and `"I love AI"`
Vocabulary: `[I, love, NLP, AI]`
- `"I love NLP"` → `[1, 1, 1, 0]`
- `"I love AI"` → `[1, 1, 0, 1]`

**Why:** BoW is simple, fast, and easy to implement. It converts text into numerical form suitable for machine learning.

**When:** Used in basic NLP tasks like text classification and sentiment analysis where word frequency matters more than order.


## N-grams
N-grams are continous sequences of N words or tokens extracted from text, used to capture local word order and context in NLP.

**Types:** Unigram (N=1), Bigram (N=2), Trigram (N=3).

**Example:** Sentence: *"I love NLP"*
- Unigrams: `[I, love, NLP]`
- Bigrams: `[I love, love NLP]`
- Trigrams: `[I love NLP]`

**Why:** N-grams capture word order and local context that Bag of Words misses, improving understanding of phrases and patterns.

**Limitations:** High dimensionality for large N, data sparsity, inability to capture long-range dependencies, and high computational cost for large datasets.


## TF-IDF (Term Frequency–Inverse Document Frequency)
TF-IDF is a statistical method used in NLP and information retrieval to evaluate how important a word is to a document relative to a larger collection. It combines Term Frequency (how often a word appears in a document) and Inverse Document Frequency (how rare the word is across all documents).

**Example:**
- D1: `"I love NLP"` | D2: `"I love AI"`
- Common word `"love"` appears in both → low IDF (less important)
- Unique word `"NLP"` appears once → high IDF (more important)

**Why:** TF-IDF highlights important and unique words in a document, reducing the effect of frequently occurring but less informative words.

**Limitations:** Ignores word order and context; cannot capture semantic meaning; high dimensional for large vocabularies.


## Word Frequency
Word Frequency is the count of how often each word appears in a document or corpus, used for basic text analysis in NLP.

**Example:** Sentence: *"NLP is easy and NLP is powerful"*
- NLP → 2, is → 2, easy → 1, and → 1, powerful → 1

**Why:** Word frequency helps identify important or commonly used words in text, useful for basic analysis and feature extraction.

**Limitations:** Ignores context and word order; common words may dominate results; not effective for understanding meaning.


## Character N-grams
Character N-grams are contiguous sequences of N characters extracted from a word or text. They are used in NLP to capture subword patterns, prefixes, suffixes, and spelling variations.

**Example:** Word: `"NLP"` → Bigrams: `[NL, LP]`, Trigrams: `[NLP]`

**Why:** Character N-grams handle misspellings, morphological variations, and unseen words. They are useful for text classification, language identification, and spell correction.


# Word Representation & Embeddings

## One-Hot Encoding
One-Hot Encoding is a technique to convert categorical variables into a binary vector representation. Each category is represented by a vector of length equal to the number of unique categories, with 1 for the active category and 0 for all others.

**Example:**
| Color | Red | Blue | Green |
| ----- | --- | ---- | ----- |
| Red   | 1   | 0    | 0     |
| Blue  | 0   | 1    | 0     |
| Green | 0   | 0    | 1     |

**Why:** Machine learning models cannot work directly with categorical data; one-hot encoding converts it to numerical form without implying any ordinal relationship among categories.

**Limitations:** Increases dimensionality for high-cardinality features; sparse and memory-inefficient for large vocabularies.


## Word Embeddings
Word embeddings are dense vector representations of words in a continuous vector space. They capture semantic and syntactic meaning so that words with similar meanings are closer in the vector space.

**Example:** `Vector(king) − Vector(man) + Vector(woman) ≈ Vector(queen)`

**Why:** They capture semantic relationships between words, unlike BoW or TF-IDF, and reduce dimensionality while improving NLP model performance.

**Use Cases:** Text classification, sentiment analysis, machine translation, NER, and chatbots.

**Limitations:** Require large corpora for training; pre-trained embeddings may not fit domain-specific words; static embeddings cannot handle polysemy (use BERT for contextual embeddings).


## Word2Vec
Word2Vec is a neural network-based model that converts words into dense vectors (word embeddings), capturing semantic and syntactic relationships and placing similar words close together in vector space.

**Key Architectures:**
- **CBOW (Continuous Bag of Words):**
- **Skip-gram:**

**Example:** `Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")`

**Limitations:** Static embeddings (one vector per word regardless of context); requires large datasets; does not capture long-range context.


## GloVe (Global Vectors for Word Representation)
GloVe is a word embedding technique that generates dense vector representations of words by analyzing global word co-occurrence statistics from a corpus, capturing semantic and syntactic relationships.

**How it Works:** Construct a word-word co-occurrence matrix from the corpus, then learn word vectors by factorizing this matrix so that vector dot products approximate co-occurrence probabilities.

**Why:** It captures semantic meaning using corpus-wide statistics and produces dense, low-dimensional vectors for NLP tasks.

**Limitations:** Static embeddings (cannot handle context-dependent meanings); requires a large corpus; less flexible for domain-specific vocabulary without retraining.


## FastText
FastText is a word embedding technique developed by Facebook that represents words as vectors of character n-grams. It captures subword information, enabling it to handle rare words, misspellings, and morphologically rich languages.

**How it Works:** Break each word into character-level n-grams (e.g., `"apple"` → `[ap, pp, pl, le]`), represent the word as the sum of its n-gram vectors, then train using a Skip-gram or CBOW model.

**Why:** It handles out-of-vocabulary (OOV) words using subword information and produces robust embeddings for morphologically rich languages.

**Limitations:** Slightly more computationally expensive than Word2Vec; still produces static embeddings.


## Contextual Word Embeddings
Contextual word embeddings are dynamic word representations that change depending on the context in which a word appears. Unlike static embeddings (Word2Vec, GloVe, FastText), the same word can have different vectors in different sentences.

**Example:** Word: `"bank"`
- *"I went to the bank to deposit money"* → embedding reflects a financial institution.
- *"The river bank was full of trees"* → embedding reflects a river edge.

**Why:** They capture word meaning based on context, solving the limitations of static embeddings and handling polysemy effectively.

**Models:** ELMo, BERT, GPT.

**Limitations:** Computationally expensive; require large pre-trained models and GPU resources; higher memory requirements.


## Sentence Embeddings
Sentence embeddings are dense vector representations of entire sentences, capturing semantic meaning beyond individual words. They encode the overall context and meaning of a sentence in a fixed-length vector.

**Models:** Universal Sentence Encoder (USE), Sentence-BERT, InferSent.

**Example:**
- *"I love machine learning."*
- *"Machine learning is my passion."*
→ Their vectors are close in vector space, reflecting similar meaning.

**Why:** Captures the meaning of an entire sentence, not just individual words. Useful for semantic similarity, clustering, and information retrieval.

**Limitations:** May lose word-level nuance; some models struggle with very long or complex sentences; pre-trained models may not generalize well to domain-specific language.


# Evaluation Metrics for NLP

## Accuracy
Accuracy evaluates how well a classification model performs by measuring the percentage of correct predictions. It is most effective when the dataset is balanced (all classes are equally represented).

$$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$$

**Limitation:** In imbalanced datasets (e.g., 95% negative, 5% positive), a model predicting only the majority class achieves high accuracy without actually solving the problem.


## Precision
Precision is the ratio of true positive predictions to all positive predictions made by the model. It measures how many of the model's "positive" predictions were actually correct.

$$Precision = \frac{TP}{TP + FP}$$

**When to use:** When the cost of a false alarm (False Positive) is high — e.g., spam detection, where marking a real email as spam is harmful.

**Limitation:** A model can achieve 100% precision by making very few, highly confident predictions, while missing many actual positive cases (Low Recall).


## Recall
Recall measures how well a model finds all actual positive cases. It represents how many real positive instances the model correctly identified.

$$Recall = \frac{TP}{TP + FN}$$

**When to use:** When missing a positive case is dangerous — e.g., medical diagnosis or fraud detection, where failing to detect a case has serious consequences.

**Limitation:** Maximizing recall often leads to flagging many false positives (Low Precision).


## F1-Score
The F1-Score is the harmonic mean of Precision and Recall. It provides a single balanced score, ensuring neither metric is sacrificed for the other.

$$F1\text{-}Score = 2 \times \frac{Precision \times Recall}{Precision + Recall}$$

**Why the harmonic mean:** It penalizes extreme values — if either Precision or Recall is very low, the F1-Score drops significantly. This makes it ideal for imbalanced datasets.


## Confusion Matrix
A Confusion Matrix is a summary table used to measure the performance of a classification model. It compares the model's predictions against actual ground truth across four categories:

- **True Positive (TP):** Predicted "Yes," actual was "Yes." (Correct)
- **True Negative (TN):** Predicted "No," actual was "No." (Correct)
- **False Positive (FP):** Predicted "Yes," actual was "No." (Type I Error)
- **False Negative (FN):** Predicted "No," actual was "Yes." (Type II Error)

**Why:** It reveals whether a model is biased, handles imbalanced data well, and helps decide whether to minimize Type I or Type II errors based on the application.

**Limitation:** Becomes difficult to interpret for multi-class classification problems.


## BLEU (Bilingual Evaluation Understudy)
BLEU is an evaluation metric that measures the quality of machine-translated text by comparing n-gram overlaps between the machine output and human reference translations.

$$BLEU = BP \cdot \exp\left(\sum_{n=1}^{N} w_n \log p_n\right)$$

Where $p_n$ is the n-gram precision and $BP$ is a brevity penalty to prevent rewarding very short outputs.

**Why:** Faster and cheaper than human evaluation; allows standardized model benchmarking.

**Limitations:** Does not account for synonyms (using "huge" instead of "large" lowers the score); a sentence can score high but still sound grammatically awkward.


## ROUGE (Recall-Oriented Understudy for Gisting Evaluation)
ROUGE is the standard metric for evaluating text summarization. It focuses on how much essential information from a human reference is captured by the model's output.

**Key Variants:**
- **ROUGE-N:** Measures n-gram overlap (ROUGE-1 for unigrams, ROUGE-2 for bigrams).
- **ROUGE-L:** Uses the Longest Common Subsequence (LCS) between reference and candidate.
- **ROUGE-W:** A weighted version of LCS giving more credit to consecutive matches.
- **ROUGE-S:** Measures skip-bigrams, allowing matches between word pairs separated by other words.

$$ROUGE\text{-}N = \frac{\text{Number of matching n-grams}}{\text{Total n-grams in the reference}}$$


## Perplexity
Perplexity is a metric used to evaluate language models. It measures how well a model predicts a sequence of words — a lower perplexity indicates better prediction performance.

$$PP(W) = \exp \left( -\frac{1}{N} \sum_{i=1}^{N} \log P(w_i) \right)$$

**Interpretation:** Lower PPL = more confident and accurate predictions. Higher PPL = model is uncertain about next words.

**Use Cases:** Evaluating and comparing language models like GPT, LSTM, and other sequence models.


# Differences

## NLP vs NLU vs NLI
NLP (Natural Language Processing) is the broad field of computational techniques that enables computers to read, write, and process human language at a foundational level through tasks like tokenization, parsing, and embeddings.

NLU (Natural Language Understanding) is a subfield of NLP focused specifically on understanding the meaning, intent, entities, and context behind text — going deeper than surface-level processing. 

NLI (Natural Language Inference) is a subfield of NLU that determines the logical relationship between two sentences, specifically whether a hypothesis is entailed, contradicted, or neutral given a premise. 

In short, NLP handles the processing pipeline, NLU handles semantic understanding, and NLI handles logical reasoning between statements.


## Bag of Words vs TF-IDF
Bag of Words (BoW) represents text by simply counting word occurrences, treating all words equally regardless of how common or rare they are across documents. 

TF-IDF improves on this by weighting each word based on both its frequency in a document and its rarity across the corpus, giving higher scores to distinctive terms. 

BoW is simpler and computationally inexpensive, but common words can dominate the representation and dilute meaningful signals. 

TF-IDF is more complex and computationally expensive, but provides a more informative and balanced representation that is better suited for tasks like search engines and document similarity.


## CBOW vs Skip-gram
CBOW (Continuous Bag of Words) predicts a target word from its surrounding context words, making it faster to train and more effective for frequent words with smaller context windows. 

Skip-gram does the reverse — it predicts context words from a given target word — performing better for rare words and capturing richer semantic relationships with larger context windows.

CBOW is computationally more efficient and generally preferred when speed is a priority and the dataset is large. 

Skip-gram is slower to train but produces higher-quality embeddings for rare and low-frequency words, making it the preferred choice for datasets where rare words are important.


## Static Embeddings vs Contextual Embeddings
Static embeddings (Word2Vec, GloVe) assign a single fixed vector to each word regardless of the context it appears in, meaning the word "bank" has the same representation whether used in a financial or geographical sense. 

Contextual embeddings (BERT, ELMo) generate dynamic vectors that change based on the surrounding context, allowing the same word to have different representations in different sentences. 

Static embeddings are faster to train and use, require less memory, and perform well for basic tasks like word similarity and text classification. 

Contextual embeddings excel in complex tasks such as question answering, NER, and sentiment analysis, but at the cost of significantly higher computational and memory requirements.


## Stemming vs Lemmatization
Stemming reduces words to their root form by chopping off prefixes or suffixes using simple rules, often producing non-words (e.g., "flies" → "fli"). 

Lemmatization reduces words to their correct dictionary base form (lemma) by considering the word's part of speech and meaning, always producing valid words (e.g., "flies" → "fly"). 

Stemming is faster and computationally inexpensive, making it suitable for large-scale text processing where precision is less critical, such as search engines. 

Lemmatization is slower but more accurate, and is preferred for NLP tasks that require correct semantics, such as POS tagging, machine translation, and question answering.


## POS Tagging vs Chunking
POS tagging assigns a grammatical label (noun, verb, adjective, etc.) to each individual word in a sentence, operating at the word level. 

Chunking groups words with their POS tags into meaningful phrases such as noun phrases or verb phrases, operating at the phrase level. 

POS tagging is a prerequisite for chunking — chunking uses POS tags as input to identify higher-level syntactic structures. POS tagging is used for syntax analysis and NLP preprocessing, w

hile chunking is applied in named entity recognition, relation extraction, and text summarization.


## Constituency Parsing vs Dependency Parsing
Constituency parsing analyzes the hierarchical phrase structure of a sentence, breaking it into nested constituents like noun phrases (NP) and verb phrases (VP) organized in a tree.

Dependency parsing analyzes grammatical relationships between individual words, representing which words depend on which others through a directed graph structure.
 
Constituency parsing focuses on phrase hierarchy and is used in grammar checking, language understanding, and text generation.

Dependency parsing focuses on direct word-to-word relationships and is preferred for information extraction, relation extraction, semantic analysis, and machine translation.


## Sparse Vectorization vs Dense Vectorization
Sparse vectorization (e.g., BoW, TF-IDF, one-hot encoding) represents text as high-dimensional vectors where most entries are zero, making them simple to compute but memory-intensive and semantically limited. 

Dense vectorization (e.g., Word2Vec, GloVe, BERT embeddings) represents text as low-dimensional, compact vectors where most entries are non-zero, capturing rich semantic and syntactic relationships. 

Sparse methods are computationally cheap and interpretable, but they fail to capture word meaning, context, or similarity beyond simple word occurrence. 

Dense methods require more resources to generate but enable meaningful similarity computations and power modern NLP tasks like semantic search, question answering, and recommendation systems.


## Word2Vec vs GloVe vs FastText
Word2Vec is a predictive model that learns embeddings by predicting context words (via CBOW or Skip-gram) based on local context windows, but it struggles with rare and out-of-vocabulary (OOV) words since each word has a single fixed vector. 

GloVe is a count-based model that learns embeddings from a global word co-occurrence matrix, capturing corpus-wide statistical relationships, but it also fails on rare and OOV words since it assigns one fixed vector per word. 

FastText extends Word2Vec by representing words as the sum of their character n-grams, enabling it to generate embeddings for rare and unseen words, making it especially effective for morphologically rich languages. In summary, 

Word2Vec and GloVe are best for general semantic tasks with known vocabularies, while FastText is the preferred choice when robustness to rare words and OOV terms is required.

## 1. How do you handle imbalanced text datasets?

In real-world NLP problems like spam detection, fraud detection, or
sentiment analysis, class imbalance is very common. I first analyze the
distribution of classes using basic statistics and visualization to
understand how severe the imbalance is.

From a data-level approach, I may use oversampling techniques like
random oversampling or synthetic techniques such as SMOTE for text
embeddings. For text specifically, I prefer data augmentation methods
such as synonym replacement, back translation, or contextual
augmentation using transformer models.

From a model-level approach, I use class-weighted loss functions where
the minority class is penalized more heavily. In deep learning models, I
adjust the loss function using weighted cross-entropy or focal loss.

For evaluation, I avoid accuracy because it can be misleading. Instead,
I use precision, recall, F1-score, ROC-AUC, or PR-AUC depending on the
business problem.

------------------------------------------------------------------------

## 2. How do you clean and preprocess text data?

Text preprocessing depends on the use case. For classical ML models like
TF-IDF + Logistic Regression, preprocessing is more aggressive. For
transformer-based models like BERT, minimal preprocessing is preferred.

Typical steps include:

-   Lowercasing (if case is not important)
-   Removing HTML tags and special characters
-   Handling punctuation
-   Removing stopwords (for traditional ML)
-   Tokenization
-   Lemmatization or stemming (if required)
-   Handling emojis and slang if social media data

For transformer-based models, I usually keep punctuation and casing if
using cased models. Tokenization is handled using pretrained tokenizers
like WordPiece or BPE.

I also handle missing values, remove duplicates, and normalize text such
as expanding contractions (e.g., "can't" → "cannot").

Finally, I split the dataset carefully into train, validation, and test
sets without data leakage.

------------------------------------------------------------------------

## 3. How do you deal with out-of-vocabulary (OOV) words?

Out-of-vocabulary words are common in NLP, especially in domain-specific
applications.

In traditional NLP models like Bag-of-Words or Word2Vec, OOV words are
either ignored or mapped to an unknown token. This can reduce
performance in specialized domains.

Modern transformer models solve this using subword tokenization
techniques such as:

-   Byte Pair Encoding (BPE)
-   WordPiece
-   SentencePiece

These break unknown words into smaller subword units. For example,
"unhappiness" can be split into "un", "happi", "ness". This
significantly reduces the OOV problem.

In domain-specific use cases, I also consider fine-tuning the tokenizer
or training embeddings on domain-specific corpora.

For dynamic systems like chatbots, I may log unknown words and
periodically retrain the model with updated vocabulary.

## 4. How would you build a chatbot?

Building a chatbot depends on whether it is rule-based, retrieval-based,
or generative.

First, I define the use case clearly: customer support, FAQ bot,
conversational AI, etc.

High-level steps:

1.  Requirement gathering and defining intents
2.  Data collection and annotation
3.  NLP pipeline:
    -   Intent classification
    -   Entity recognition (NER)
4.  Response generation:
    -   Rule-based templates
    -   Retrieval-based system
    -   Generative model using LLM
5.  Backend integration (database, APIs)
6.  Deployment (FastAPI backend, containerized with Docker, deployed on
    cloud)
7.  Monitoring and feedback loop

For modern chatbot systems, I may use RAG architecture: - Store
documents in vector databases - Use embeddings for similarity search -
Retrieve relevant context - Pass context to LLM for response generation

Evaluation is done using metrics like intent accuracy, response
relevance, latency, and user satisfaction.

## 5. How do you reduce overfitting in NLP models?

Overfitting occurs when the model memorizes training data but performs
poorly on unseen data.

To reduce overfitting, I use:

-   More training data or data augmentation
-   Regularization techniques like L2 regularization
-   Dropout layers in deep learning models
-   Early stopping based on validation loss
-   Cross-validation
-   Simpler model architectures if dataset is small

For transformer fine-tuning: - Freeze lower layers initially - Use lower
learning rate - Apply weight decay - Use smaller batch sizes carefully

Proper train-validation-test split and monitoring validation metrics are
critical to prevent data leakage and overfitting.

------------------------------------------------------------------------

## 6. How do you handle multilingual NLP problems?

Multilingual NLP requires handling multiple languages with different
scripts and grammar.

Approaches:

1.  Translate everything into a single language using machine
    translation.
2.  Train separate models per language.
3.  Use multilingual pretrained models like mBERT or XLM-R.

I usually prefer multilingual transformer models because they share
representations across languages and perform well even in low-resource
languages.

Key considerations:

-   Language detection as a preprocessing step.
-   Unicode normalization.
-   Handling code-mixing (common in India).
-   Evaluating per-language performance.
-   Fine-tuning with balanced multilingual datasets.

For production systems, I monitor performance per language and
continuously retrain with new data.
