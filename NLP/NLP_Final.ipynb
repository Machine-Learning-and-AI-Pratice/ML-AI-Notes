# Natural Language Processing (NLP) Notes

## What is Natural Language Processing?
Natural Language Processing (NLP) is an interdisciplinary field that combines computer science, artificial intelligence, and linguistics to enable computers to understand, process, interpret, and generate human language in a meaningful way.

**Why it Matters:**
NLP bridges the gap between human communication and machine understanding. It powers a wide range of real-world applications including search engines, virtual assistants (Siri, Alexa), chatbots, machine translation (Google Translate), spam filters, sentiment analysis, and medical text analysis.

**Core Goals:**
- Enable machines to **read and understand** text (comprehension)
- Enable machines to **generate** coherent and meaningful text (generation)
- Enable machines to **translate** between languages (translation)
- Enable machines to **respond** intelligently in conversation (dialogue)


## Phases of NLP
NLP processes human language through a series of sequential stages, each building on the previous one to extract progressively deeper understanding.

| Phase | Purpose |
|-------|---------|
| **Lexical Analysis** | Breaks text into words (tokens) and identifies their basic properties |
| **Morphological Analysis** | Examines the internal structure of words (roots, prefixes, suffixes) |
| **Syntactic Analysis** | Checks grammar and sentence structure using parse trees |
| **Semantic Analysis** | Understands the meaning of words and sentences in context |
| **Discourse Integration** | Connects sentences to ensure coherence across the full text |
| **Pragmatic Analysis** | Interprets the speaker's intent, tone, and context beyond literal meaning |


## Lexical Analysis
Lexical analysis is the first stage of NLP where raw text is scanned and divided into meaningful units called **tokens**, which are then classified into categories (lexemes). It involves tokenization, normalization (lowercasing, whitespace removal), and vocabulary identification.

**Key Tasks:** Tokenization and Part-of-Speech (POS) Tagging.

**Example:**
Input: `"Running fast is fun."`
Tokens: `["Running", "fast", "is", "fun", "."]`

**Why:** It transforms unstructured raw text into structured, manageable components that all subsequent NLP stages depend on. Without lexical analysis, a machine cannot distinguish individual words from a continuous stream of characters.

**Use Cases:** All NLP pipelines — preprocessing step for parsing, embeddings, classification, and more.


## Morphological Analysis
Morphological analysis examines the **internal structure of words** by breaking them into their smallest meaningful units, called **morphemes** (roots, prefixes, and suffixes). It identifies how word forms relate to their base meaning and grammatical function.

**Key Tasks:** Stemming and Lemmatization.

**Example:**
- Word: `"unhappiness"` → Morphemes: `un-` (prefix) + `happi` (root) + `-ness` (suffix)
- `"waiting"`, `"waiter"`, `"waited"` → all share the root morpheme `"wait"`

**Why:** Without morphological analysis, a computer treats `"wait"`, `"waiter"`, and `"waiters"` as completely unrelated strings. With it, the computer recognizes they share the same core concept, reducing redundancy and improving accuracy.

**Use Cases:** Stemming/lemmatization pipelines, spell correction, morphologically rich language processing (e.g., Arabic, Finnish).


## Syntactic Analysis (Parsing)
Syntactic analysis examines the **grammatical structure** of a sentence — how words are arranged and related to each other according to grammar rules. It produces a **parse tree** that represents the hierarchical relationships between words (subject, verb, object, modifiers).

**Key Tasks:** POS Tagging, Constituency Parsing, Dependency Parsing, and Ambiguity Resolution.

**Example:**
Sentence: `"The cat sat on the mat."`
Parse tree identifies: Subject → `"The cat"`, Verb → `"sat"`, Prepositional phrase → `"on the mat"`

**Why:** Syntactic analysis ensures machines understand sentence structure, not just individual words. It enables correct interpretation of meaning, supports grammar checking, and is essential for translation, question answering, and text understanding.

**Use Cases:** Grammar checkers, machine translation, question answering systems, and dependency parsing pipelines.


## Semantic Analysis
Semantic analysis focuses on understanding the **actual meaning** of words, phrases, and sentences in context. It goes beyond grammar to ensure that text is logically correct, contextually appropriate, and meaningfully interpreted.

**Key Tasks:** Named Entity Recognition (NER), Word Sense Disambiguation (WSD), and Semantic Role Labeling.

**Example:**
- Sentence: `"I saw her duck."` — Semantic analysis determines whether "duck" is a noun (the bird) or a verb (to lower one's head) based on context.
- Code: `total = "Price" * 10` — Grammatically valid, but semantically incorrect (cannot multiply a string by a number).

**Why:** Grammar alone does not guarantee correctness. Semantic analysis ensures that the meaning conveyed is valid and coherent, preventing nonsensical or logically broken outputs.

**Use Cases:** Machine translation, question answering, information extraction, chatbots, and code analysis.


## Discourse Integration
Discourse integration is the process of understanding how **sentences connect and relate to each other** across a larger body of text. It ensures meaning is consistent and coherent throughout a paragraph, document, or conversation.

**Key Tasks:** Anaphora Resolution and Contextual Reference Tracking.

**Example:**
- `"Alice entered the room. She sat down."` → Discourse integration links `"She"` back to `"Alice"` across sentences.
- `"The company released its product. It was well received."` → `"It"` refers to `"the product"`, not `"the company"`.

**Why:** Without discourse integration, an NLP system processes each sentence in isolation and loses track of entities, pronouns, and context across the text — leading to incoherent responses and summaries.

**Use Cases:** Chatbots (maintaining conversation context), text summarization, machine translation (consistent pronoun use), and document-level question answering.


## Pragmatic Analysis
Pragmatic analysis interprets the **deeper communicative intent** of language — the meaning a speaker intends beyond the literal words used. It considers context, social norms, tone, and situational factors.

**Key Tasks:** Intent Recognition, Figurative Language Understanding (idioms, sarcasm, irony).

**Example:**
- `"Can you pass the salt?"` — Literally a question about ability; pragmatically a polite request.
- `"Call me a cab."` — Does not mean "rename me 'a cab'"; pragmatic analysis understands it as a request to order a taxi.
- `"Great, another Monday."` — Sarcastic; pragmatic analysis detects the negative sentiment behind a seemingly positive word.

**Why:** Human communication is full of implied meaning, sarcasm, idioms, and indirect requests. Without pragmatic analysis, AI systems respond robotically and literally, producing inappropriate or incorrect outputs.

**Use Cases:** Virtual assistants (Siri, Alexa, ChatGPT), sentiment analysis (detecting sarcasm), customer service bots, and social media monitoring.


# Text Preprocessing

## Text Cleaning
Text cleaning is the foundational preprocessing step that removes **unwanted characters, noise, and inconsistencies** from raw text before it is fed into an NLP model.

**Common Operations:**
- Removing extra whitespace and blank lines
- Converting text to lowercase
- Removing HTML tags, URLs, and email addresses
- Removing special characters, symbols, and punctuation
- Removing or replacing numbers (when not semantically relevant)
- Removing or converting emojis to text
- Expanding contractions (e.g., `don't` → `do not`)
- Removing duplicate sentences or records
- Correcting misspellings

**Why:** Raw, real-world text is messy. Uncleaned text introduces noise that can confuse models, inflate vocabulary size unnecessarily, and reduce prediction accuracy. Text cleaning ensures the model focuses only on meaningful content.

**Use Cases:** All NLP pipelines — applied before tokenization, feature extraction, and model training.


## Tokenization
Tokenization is the process of breaking text into smaller units called **tokens**. Tokens can be words, subwords, or individual characters, depending on the granularity required by the NLP task.

**Types:**
- **Word Tokenization:** Splits text into words. e.g., `"I love NLP"` → `["I", "love", "NLP"]`
- **Character Tokenization:** Splits text into individual characters. Useful for spelling tasks or languages without clear word boundaries.
- **Subword Tokenization:** Splits words into meaningful subunits. e.g., `"unhappy"` → `["un", "happy"]`. Used in BPE, WordPiece.
- **Sentence Tokenization:** Splits a paragraph into individual sentences.
- **N-gram Tokenization:** Groups tokens into sequences of N items to capture local patterns.

**Example:**
Input: `"NLP is powerful!"`
Word tokens: `["NLP", "is", "powerful", "!"]`

**Why:** Tokenization converts raw text into discrete units that machines can process mathematically. Without it, models cannot identify individual words, measure frequencies, or generate embeddings.

**Use Cases:** Every NLP pipeline — prerequisite for BoW, TF-IDF, embeddings, and language model training.


## Text Corpus
A text corpus is a **large, structured collection of text** used to train, validate, and evaluate NLP models. It provides real language data from which models learn vocabulary, grammar patterns, context, and semantics.

**Types:**
- **Monolingual Corpus:** Text in a single language (e.g., English Wikipedia)
- **Multilingual Corpus:** Text in multiple languages
- **Parallel Corpus:** Same content translated into multiple languages — used for machine translation training
- **Annotated Corpus:** Text labeled with linguistic information (POS tags, NER labels, sentiment, etc.)
- **Unannotated Corpus:** Raw, unlabeled text — used for unsupervised learning

**Example:** A sentiment analysis model is trained on a corpus of 50,000 movie reviews labeled as positive or negative. The model learns which words and phrases correlate with each sentiment from this corpus.

**Why:** The quality and size of the corpus directly determines the quality of the trained NLP model. A diverse, well-curated corpus leads to a model that generalizes well to real-world language.

**Use Cases:** Pre-training language models (BERT, GPT), building word embeddings, training classifiers, and evaluating NLP systems.


## Sentence Segmentation
Sentence segmentation is the process of **dividing a body of text into individual sentences**. It identifies sentence boundaries, typically marked by punctuation such as periods, question marks, or exclamation points.

**Example:**
Input: `"I love NLP. It is very powerful! Do you agree?"`
Output: `["I love NLP.", "It is very powerful!", "Do you agree?"]`

**Challenge:** Periods do not always mark sentence endings — abbreviations (`Dr.`, `U.S.A.`) and decimal numbers (`3.14`) can cause false splits.

**Why:** Many NLP tasks — such as machine translation, summarization, and sentiment analysis — work best at the sentence level. Processing the entire text as one block leads to loss of structure and context.

**Use Cases:** Machine translation, text summarization, sentiment analysis, and document-level NLP pipelines.


## Stop Words
Stop words are **high-frequency grammatical words** — such as articles (`a`, `the`), prepositions (`in`, `on`), conjunctions (`and`, `but`), and pronouns (`he`, `she`) — that appear very commonly in text but carry little semantic meaning on their own.

**Example:**
Query: `"How to bake a cake in London"`
After stop word removal: `["bake", "cake", "London"]`
The system now focuses on the three meaningful keywords instead of six total words.

**Why:** Removing stop words reduces vocabulary size, decreases computational cost, and allows models to focus on the words that carry actual meaning — improving accuracy and efficiency.

**When NOT to remove:** Tasks like sentiment analysis (`"not good"` → removing `"not"` reverses the meaning) or language modeling where grammatical words matter for structure.

**Use Cases:** Search engines, text classification, topic modeling, TF-IDF vectorization, and document similarity.


## Stemming
Stemming is an NLP technique that reduces words to their **base or root form** by mechanically removing suffixes or prefixes using heuristic rules, without considering word meaning or context.

**Example:**
- `"running"`, `"runner"`, `"runs"` → `"run"`
- `"studies"`, `"studying"`, `"studied"` → `"studi"`
- `"happiness"` → `"happi"`

**Why:** Stemming reduces vocabulary size, decreases storage requirements, and helps models treat different word forms as the same term — improving recall in search and classification tasks.

**Limitations:**
- **Over-stemming:** `"Universal"`, `"University"`, and `"Universe"` may all reduce to `"Univers"`, incorrectly merging unrelated words.
- **Under-stemming:** `"data"` and `"datum"` may not be correctly linked.
- **Nonsense roots:** `"Business"` → `"Busin"`, which is not a real word.

**Use Cases:** Search engines (broadening query results), spam detection, and large-scale text classification where speed is prioritized over precision.


## Lemmatization
Lemmatization is an NLP preprocessing technique that reduces words to their **correct dictionary base form (lemma)** by considering the word's meaning, context, and part of speech — producing linguistically valid results.

**Example:**
- `"running"` (verb) → `"run"` | `"better"` → `"good"` | `"was"` → `"be"`
- `"flies"` (noun, the insect) → `"fly"` | `"flies"` (verb) → `"fly"`
- `"saw"` (noun, the tool) → `"saw"` | `"saw"` (verb, past tense of see) → `"see"`

**Why:** Unlike stemming, lemmatization always produces a valid dictionary word and respects the word's grammatical role. This makes it more accurate for tasks where the precise meaning of words matters.

**Limitations:**
- Slower and more computationally expensive than stemming.
- Requires a linguistic dictionary (e.g., WordNet) and a POS tagger.
- May struggle with domain-specific or rare vocabulary.

**Use Cases:** Chatbots, sentiment analysis, medical and legal text analysis, question answering, and any task requiring precise semantic understanding.


## Handling Punctuation
Punctuation handling involves **removing or normalizing punctuation marks** (commas, periods, exclamation marks, etc.) that do not contribute meaningful information to the NLP task.

**Why:** Punctuation adds noise to feature vectors. For example, `"NLP!"` and `"NLP"` should be treated as the same token. Removing punctuation reduces vocabulary size and improves model consistency.

**Note:** In some tasks (e.g., sentence segmentation, named entity recognition), punctuation must be preserved as it carries structural meaning.

## Handling Numbers
Number handling involves **removing, replacing, or normalizing numeric values** based on their relevance to the task.

**Why:** Numbers are often treated as noise in text classification tasks, but they carry important meaning in domains like finance or medicine (e.g., `"blood pressure 120/80"` or `"price: $500"`). The decision to keep or remove numbers should be made based on the task context.

## Handling Special Characters
Special characters such as `@`, `#`, `$`, `%`, and `*` are typically **removed or replaced** to reduce noise in the input text.

**Why:** These characters rarely contribute semantic value in standard NLP tasks and can create inconsistent tokenization. In social media NLP, `#hashtags` and `@mentions` may be preserved or extracted separately as features.

## Handling Emojis
Emojis are either **removed or converted to their textual description** (e.g., 😊 → `"smiling face"`) during preprocessing.

**Why:** Emojis carry strong sentiment signals, especially in social media data. Converting them to text preserves this information rather than discarding it, which can significantly improve sentiment analysis accuracy.


## Spelling Correction
Spelling correction is the process of **identifying and automatically fixing misspelled words** in text to improve data quality and model performance.

**How it Works:**
- **Dictionary Lookup:** Check if a word exists in a known vocabulary; if not, suggest the closest match.
- **Edit Distance (Levenshtein):** Find the word requiring the fewest insertions, deletions, or substitutions to match a known word.
- **Language Model-based:** Use context to predict the most likely intended word (e.g., `"I went to the bech"` → `"beach"`).

**Example:**
`"I luv NLP"` → `"I love NLP"` | `"recieve"` → `"receive"`

**Why:** Misspelled words create spurious vocabulary entries. `"colour"`, `"colour"`, and `"clour"` would be treated as three different tokens without correction, reducing model accuracy and inflating vocabulary size.

**Use Cases:** User-generated content (chat, reviews, social media), search engines (query correction), chatbots, and preprocessing before feature extraction (BoW, TF-IDF).


## Byte Pair Encoding (BPE)
Byte Pair Encoding (BPE) is a **subword tokenization algorithm** that iteratively merges the most frequent pairs of characters or character sequences into new tokens, building a compact vocabulary that balances between word-level and character-level representation.

**How it Works:**
1. Initialize the vocabulary with all individual characters in the corpus.
2. Count all adjacent token pairs and find the most frequent pair.
3. Merge that pair into a single new token and add it to the vocabulary.
4. Repeat until the vocabulary reaches the predefined target size.

**Example:**
Corpus: `["lower", "lowest"]`
- Initial: `l o w e r`, `l o w e s t`
- Most frequent pair `l o` → merge to `lo`: `lo w e r`, `lo w e s t`
- Continue merging: eventually produces `lowe r`, `lowe st`

**Why:** BPE allows models to handle rare and unseen words by decomposing them into known subword units, eliminating the need for `<UNK>` (unknown) tokens. It is used in GPT, RoBERTa, and many modern transformer models.

**Use Cases:** Tokenization in transformer models (GPT, RoBERTa), machine translation, and handling multilingual text.


## OOV (Out-Of-Vocabulary) Handling
Out-of-Vocabulary (OOV) words are words that **do not appear in the model's training vocabulary**. Robust OOV handling is essential for NLP systems deployed on real-world text, which frequently contains slang, typos, brand names, and neologisms.

**Techniques:**
- **Subword Tokenization (BPE, WordPiece, SentencePiece):** Splits OOV words into known subword units.
  Example: `"unhappiness"` → `["un", "happi", "ness"]`
- **Character-level Embeddings:** Represents any word as a sequence of characters, allowing embeddings for any possible string.
- **Unknown Token `<UNK>`:** Replaces all OOV words with a single generic token. Simple but loses all semantic information about the word.
- **FastText-style Embeddings:** Computes word vectors as the sum of character n-gram vectors, enabling embeddings for entirely unseen words.

**Why:** Real-world text is unpredictable. Users coin new words, use slang, and make spelling errors. If a model cannot handle OOV words, it either crashes or loses important information — both unacceptable for production systems.

**Use Cases:** Any production NLP system, especially chatbots, search engines, and models deployed on social media or user-generated content.


## Text Normalization
Text normalization is the process of **transforming text into a standard, consistent format** to reduce variation and ensure that semantically equivalent text is represented identically.

**Steps:**
- **Lowercasing:** `"NLP"`, `"Nlp"`, and `"nlp"` become the same token → `"nlp"`
- **Expanding Contractions:** `"don't"` → `"do not"`, `"I'm"` → `"I am"`
- **Removing Punctuation:** Reduces noise for tasks where punctuation is not meaningful
- **Removing Numbers:** When numeric values are not relevant to the task
- **Removing Special Characters & Extra Whitespace:** Ensures clean, uniform input
- **Unicode Normalization:** Converts different encodings of the same character to a standard form

**Why:** Inconsistent text representation causes the same word to appear as multiple different vocabulary entries, inflating vocabulary size and reducing model generalization. Normalization ensures the model sees consistent, clean input.

**Use Cases:** All NLP preprocessing pipelines — applied before tokenization, stemming/lemmatization, and feature extraction.


# Linguistic Processing

## Part-of-Speech (POS) Tagging
POS tagging is the process of **automatically assigning a grammatical category** (noun, verb, adjective, adverb, pronoun, etc.) to each word in a sentence based on both its definition and its context within the sentence.

**Common POS Tags:**
| Tag | Category | Example |
|-----|----------|---------|
| NN | Noun | `"dog"`, `"city"` |
| VB | Verb | `"run"`, `"buy"` |
| JJ | Adjective | `"happy"`, `"red"` |
| RB | Adverb | `"quickly"`, `"very"` |
| PRP | Pronoun | `"he"`, `"she"` |

**Example:**
Sentence: *"She booked a flight yesterday."*
- `She` → Pronoun | `booked` → Verb | `a` → Determiner | `flight` → Noun | `yesterday` → Adverb

**Why:** POS tagging resolves word-level ambiguity — for example, `"book"` can be a noun (the object) or a verb (to reserve). It is a prerequisite for many downstream NLP tasks.

**Use Cases:** Dependency parsing, NER, machine translation, text summarization, grammar checking, and lemmatization.


## Chunking (Shallow Parsing)
Chunking is the process of **grouping consecutive words into meaningful syntactic phrases** (chunks) such as noun phrases (NP) and verb phrases (VP), based on their POS tags. It is also called shallow parsing because it identifies phrase boundaries without building a full parse tree.

**Example:**
Sentence: *"She bought a red car."*
POS tags: `She/PRP bought/VBD a/DT red/JJ car/NN`
Chunks: `[NP She]` `[VP bought]` `[NP a red car]`

**Why:** Chunking provides phrase-level structure, which is more informative than individual word-level analysis. It helps identify the "what" and "who" in a sentence more precisely than POS tagging alone.

**Use Cases:** Information extraction (extracting noun phrases as entities), question answering, text summarization, and named entity recognition (NER).


## Constituency Parsing
Constituency parsing breaks a sentence into **nested sub-phrases (constituents)** such as noun phrases (NP), verb phrases (VP), and prepositional phrases (PP), organized into a hierarchical tree structure called a **parse tree** or **phrase structure tree**.

**Example:**
Sentence: *"She bought a red car."*
Parse tree:
```
(S
  (NP She)
  (VP bought
    (NP a red car)))
```
Identifies: Subject NP → `"She"`, VP → `"bought a red car"`, Object NP → `"a red car"`

**Why:** Constituency parsing reveals the complete grammatical hierarchy of a sentence, enabling deeper language understanding beyond individual word roles.

**Use Cases:** Syntactic analysis, grammar checking, machine translation (preserving sentence structure), question answering, and language understanding research.


## Dependency Parsing
Dependency parsing identifies **grammatical relationships between individual words** in a sentence by creating a directed graph where each word (except the root) depends on another word. It shows who did what to whom.

**Key Relations:**
- `nsubj` — nominal subject (who performs the action)
- `dobj` — direct object (what the action is performed on)
- `amod` — adjectival modifier
- `prep` — prepositional modifier

**Example:**
Sentence: *"She bought a red car."*
- `bought → She` (nsubj: She is the subject)
- `bought → car` (dobj: car is the object)
- `car → red` (amod: red modifies car)

From this, a system confidently infers: **Actor** = She | **Action** = bought | **Object** = red car

**Why:** Bag-of-Words and basic embeddings capture *what* words exist but not *who did what to whom*. Dependency parsing enables precise semantic reasoning and information extraction.

**Use Cases:** Information extraction, knowledge graph construction, question answering, machine translation, and relation extraction.


## Named Entity Recognition (NER)
Named Entity Recognition (NER) is an NLP technique that **automatically identifies and classifies real-world entities** mentioned in text into predefined categories such as Person, Organization, Location, Date, and Quantity.

**Common Entity Categories:**
| Label | Meaning | Example |
|-------|---------|---------|
| PERSON | People's names | `Elon Musk` |
| ORG | Organizations/Companies | `Starbucks` |
| GPE | Geopolitical entity (city/country) | `Berlin` |
| DATE | Dates and time periods | `2024` |
| MONEY | Monetary values | `$500` |

**Example:**
*"In 2024, Elon Musk visited the Starbucks headquarters in Berlin."*
- `2024` → DATE | `Elon Musk` → PERSON | `Starbucks` → ORG | `Berlin` → GPE

**Why:** NER converts unstructured text into structured, machine-readable data, making it possible to automatically extract key facts from large volumes of text.

**Use Cases:** Search engines, customer support automation, recruitment (resume parsing), news article tagging, medical record analysis, and financial document processing.


## Word Sense Disambiguation (WSD)
Word Sense Disambiguation is the NLP process of **identifying the correct meaning of a word that has multiple possible meanings (polysemy)**, based on the surrounding context.

**Example:**
- *"I went to the **bank** to deposit money."* → `bank` = financial institution
- *"We sat by the river **bank**."* → `bank` = land alongside water
- *"He couldn't **bear** the cold."* → `bear` = tolerate
- *"The **bear** ran into the forest."* → `bear` = the animal

**Why:** Many common English words have multiple meanings. Without WSD, an NLP system might translate `"bank"` incorrectly, retrieve wrong search results, or generate nonsensical answers. Resolving ambiguity is fundamental to accurate language understanding.

**When:** Applied whenever a word has multiple possible meanings and the correct interpretation must be determined from context — especially in machine translation, information retrieval, and question answering.

**Use Cases:** Machine translation, semantic search, question answering, information extraction, and text-to-speech systems.


## Anaphora Resolution
Anaphora resolution is the NLP task of **identifying the specific noun or entity that a pronoun or referring expression points back to** in a text. It resolves references such as `"he"`, `"she"`, `"it"`, `"they"`, `"this"`, and `"that"` to their correct antecedent.

**Example:**
*"Sanjay went to the store. **He** bought some apples. The clerk helped **him**."*
- `"He"` → `"Sanjay"` | `"him"` → `"Sanjay"`

**Why:** Anaphora resolution allows machines to understand that different words or phrases refer to the same entity across sentences. Without it, each sentence is processed in isolation, breaking conversational continuity and causing wrong interpretations.

**Use Cases:** Chatbots and dialogue systems (maintaining conversation context), machine translation, text summarization, question answering, and coreference-based information extraction.


## Coreference Resolution
Coreference resolution is the NLP task of **identifying all expressions in a text that refer to the same real-world entity** and grouping them into coreference chains. It includes not just pronouns but also noun phrases, proper names, and descriptive references.

**Example:**
*"**Apple** launched a new product. **The company** said **it** would ship next month. **Tim Cook** praised **the device**."*
- Coreference chain 1: `Apple` → `The company`
- Coreference chain 2: `it` → `the device` → `a new product`
- Note: `Tim Cook` is a separate entity (CEO of Apple, not Apple itself)

**Difference from Anaphora Resolution:**
Anaphora resolution handles only backward-looking pronoun references, while coreference resolution handles all types of references — including forward references (cataphora) and noun phrase co-references.

**Why:** Full document understanding requires tracking entities across sentences. Coreference resolution enables machines to understand *who* is being discussed throughout a document, not just in a single sentence.

**Use Cases:** Document summarization, multi-document information extraction, question answering over long texts, knowledge graph construction, and dialogue systems.


# Feature Engineering

## Bag of Words (BoW)
Bag of Words is a text representation technique that converts a document into a **numerical vector based on word frequency**, completely ignoring grammar and word order. Each document is represented as a "bag" containing its words and their occurrence counts.

**How it Works:**
1. Build a vocabulary of all unique words across the corpus.
2. For each document, count how many times each vocabulary word appears.
3. Represent the document as a fixed-length vector of those counts.

**Example:**
Sentences: `"I love NLP"` and `"I love AI"`
Vocabulary: `[I, love, NLP, AI]`
- `"I love NLP"` → `[1, 1, 1, 0]`
- `"I love AI"` → `[1, 1, 0, 1]`

**Why:** BoW is simple, fast, and easy to implement, providing a quick numerical baseline for NLP models.

**Limitations:**
- Ignores word order — `"dog bites man"` and `"man bites dog"` produce identical vectors.
- Does not capture semantic meaning — `"good"` and `"excellent"` are treated as unrelated.
- Creates very high-dimensional, sparse vectors for large vocabularies.
- Common words (stop words) dominate the representation if not removed.

**Use Cases:** Spam detection, basic text classification, document similarity, and as a baseline before using more advanced methods.


## N-grams
N-grams are **contiguous sequences of N words or tokens** extracted from text. They extend the Bag of Words model by capturing local word order and short-range context, allowing the model to detect phrases and common word patterns.

**Types:**
- **Unigram (N=1):** Single words — `["I", "love", "NLP"]`
- **Bigram (N=2):** Word pairs — `["I love", "love NLP"]`
- **Trigram (N=3):** Word triples — `["I love NLP"]`

**Example:**
Sentence: *"New York City is big"*
Bigrams: `["New York", "York City", "City is", "is big"]`
→ Bigrams correctly capture `"New York"` as a unit, which unigrams would split.

**Why:** Unigrams (single words) miss important phrase-level patterns. N-grams capture context like `"not good"` (bigram) vs. just `"good"` (unigram), which changes meaning entirely.

**Limitations:**
- Vocabulary size grows exponentially with N (high dimensionality).
- Sparse vectors — most n-gram combinations rarely appear.
- Cannot capture long-range dependencies beyond the window of N words.

**Use Cases:** Text classification, sentiment analysis, language modeling, machine translation (phrase tables), and intent detection in chatbots.


## TF-IDF (Term Frequency–Inverse Document Frequency)
TF-IDF is a statistical weighting method that evaluates **how important a word is to a specific document relative to a collection of documents (corpus)**. It balances word frequency within a document against the word's commonality across the entire corpus.

**Components:**
- **TF (Term Frequency):** How often a word appears in a document.
  `TF(t, d) = (Number of times term t appears in document d) / (Total terms in d)`
- **IDF (Inverse Document Frequency):** How rare the word is across all documents.
  `IDF(t) = log(Total documents / Number of documents containing t)`
- **TF-IDF Score:** `TF-IDF(t, d) = TF(t, d) × IDF(t)`

**Example:**
Corpus: D1 = `"I love NLP"`, D2 = `"I love AI"`
- `"love"` appears in both documents → low IDF → low TF-IDF (common, less informative)
- `"NLP"` appears only in D1 → high IDF → high TF-IDF (distinctive, more informative)

**Why:** Unlike BoW, TF-IDF down-weights common words (like `"the"`, `"is"`) that appear everywhere and gives higher scores to distinctive terms that are meaningful for a specific document.

**Limitations:** Ignores word order and semantic meaning; cannot capture synonyms; produces high-dimensional sparse vectors.

**Use Cases:** Search engines (document ranking), text classification, keyword extraction, and document similarity.


## Word Frequency
Word Frequency is the **raw count of how often each word appears** in a document or corpus. It is the simplest form of text representation and serves as the foundation for more advanced methods like TF-IDF.

**Example:**
Sentence: *"NLP is easy and NLP is powerful"*
| Word | Count |
|------|-------|
| NLP | 2 |
| is | 2 |
| easy | 1 |
| and | 1 |
| powerful | 1 |

**Why:** Word frequency provides a quick, interpretable snapshot of which words dominate a text. It is used for exploratory text analysis and as a baseline feature for simple classification models.

**Limitations:**
- Treats all words equally — common words (stop words) dominate without removal.
- Ignores word order and context entirely.
- Not suitable for comparing documents of different lengths without normalization.

**Use Cases:** Exploratory data analysis, basic keyword extraction, word cloud generation, and corpus statistics.


## Character N-grams
Character N-grams are **contiguous sequences of N characters** extracted from a word or text. Unlike word N-grams that operate at the word level, character N-grams operate at the character level, capturing subword patterns such as prefixes, suffixes, and morphological variations.

**Example:**
Word: `"NLP"`
- Unigrams (N=1): `[N, L, P]`
- Bigrams (N=2): `[NL, LP]`
- Trigrams (N=3): `[NLP]`

Word: `"playing"`
- Trigrams: `[pla, lay, ayi, yin, ing]`
→ The `"ing"` trigram signals a present participle across many words.

**Why:** Character N-grams can capture morphological patterns and handle misspellings, rare words, and out-of-vocabulary terms more robustly than word-level features. They are especially powerful for morphologically rich languages.

**When:** Used when word-level features are insufficient — particularly for spell correction, language identification, and text processing in languages like Arabic, Turkish, or Finnish.

**Use Cases:** Spell correction, language identification, authorship detection, text classification, and subword-level NLP in low-resource languages.


## The Hashing Trick (Feature Hashing)
The Hashing Trick is a technique that converts text features into a **fixed-size numerical vector using a hash function**, without the need to store an explicit vocabulary dictionary in memory.

**How it Works:**
1. Apply a hash function to each token (word or n-gram).
2. Map the hash output to a fixed-size vector index (e.g., index in a 2^18 dimensional vector).
3. Increment the value at that index (using raw count or TF-IDF weight).

**Example:**
Token `"NLP"` → hash function → index `42315` → vector[42315] += 1
Token `"AI"` → hash function → index `87231` → vector[87231] += 1

**Why:** Traditional vocabulary-based vectorization requires storing and loading the entire vocabulary, which is memory-intensive for large corpora. The Hashing Trick eliminates this need, enabling memory-efficient and fast streaming text processing.

**Limitations:**
- **Hash collisions:** Two different words may map to the same index, causing information loss.
- Features are not interpretable — you cannot recover the original word from its hash.
- Slightly lower accuracy than vocabulary-based methods due to collisions.

**Use Cases:** Large-scale text classification, online learning systems, and real-time streaming NLP pipelines.


# Word Representation & Embeddings

## One-Hot Encoding
One-Hot Encoding is a technique that converts **categorical variables into a binary vector representation**. For NLP, each word in the vocabulary is represented as a vector of zeros with a single `1` at the position corresponding to that word.

**How it Works:**
For a vocabulary of size V, each word is a vector of length V where exactly one position is 1 and all others are 0.

**Example:**
Vocabulary: `[cat, dog, bird]` (size = 3)
| Word | cat | dog | bird |
|------|-----|-----|------|
| cat  | 1   | 0   | 0    |
| dog  | 0   | 1   | 0    |
| bird | 0   | 0   | 1    |

**Why:** Machine learning models require numerical inputs. One-hot encoding provides a simple, unambiguous numerical representation for categorical data without implying any ordinal relationship between categories.

**Limitations:**
- Produces extremely high-dimensional, sparse vectors (one dimension per vocabulary word).
- Captures no semantic relationship — `"cat"` and `"kitten"` are just as different as `"cat"` and `"airplane"`.
- Memory-inefficient for large vocabularies.
- Cannot handle out-of-vocabulary words.

**Use Cases:** Simple categorical feature encoding, as input to basic ML models, and as a starting point before learning word embeddings.


## Word Embeddings
Word embeddings are **dense, low-dimensional vector representations of words** in a continuous vector space. Unlike one-hot encoding (sparse, high-dimensional), embeddings place semantically similar words close together in this space, capturing rich linguistic relationships.

**Key Property — Semantic Arithmetic:**
`Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")`
This shows that embeddings capture meaningful relational structure, not just identity.

**How it Works:**
Each word is mapped to a dense vector (typically 50–300 dimensions) learned during training. The model learns to position words such that those appearing in similar contexts have similar vectors, measured using cosine similarity or Euclidean distance.

**Why:** Word embeddings capture semantic meaning and syntactic relationships that sparse representations like BoW and one-hot encoding completely miss. They dramatically improve the performance of downstream NLP models.

**Limitations:**
- Require large corpora for high-quality training.
- Pre-trained embeddings may not generalize to domain-specific vocabulary.
- Traditional (static) embeddings assign one fixed vector per word, failing to capture context-dependent meaning (polysemy).

**Use Cases:** Text classification, sentiment analysis, machine translation, NER, information retrieval, and as input features for deep learning NLP models.


## Word2Vec
Word2Vec is a **neural network-based model** that learns dense word embeddings by training a shallow neural network to predict words from their context (or context from words). It places semantically related words close together in a continuous vector space.

**Key Architectures:**
- **CBOW (Continuous Bag of Words):** Predicts the target word from its surrounding context words. Faster to train; works well for frequent words.
- **Skip-gram:** Predicts context words given a single target word. Slower but better for rare words and richer semantic capture.

**Example:**
`Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")`

**Why:** Word2Vec produces high-quality dense embeddings that capture both semantic similarity (`"cat"` near `"dog"`) and syntactic patterns (`"walked"` near `"talked"`), making it a powerful feature extractor for downstream NLP tasks.

**Limitations:**
- **Static embeddings:** Each word has exactly one vector regardless of context — `"bank"` has the same vector whether it means a financial institution or a river bank.
- Requires a large training corpus for quality embeddings.
- Cannot capture long-range dependencies within sentences.

**Use Cases:** Semantic similarity, document classification, recommendation systems, and as pre-trained features for NLP models.


## CBOW (Continuous Bag of Words)
CBOW is a Word2Vec architecture that **predicts the current (target) word from its surrounding context words**. It averages the embeddings of context words and uses this average to predict the missing center word.

**How it Works:**
1. Take a sliding window of size 2k+1 over the text (k words on each side of the target).
2. Use the k context words on each side as input.
3. Train the neural network to predict the center target word.
4. The learned input weight matrix becomes the word embedding matrix.

**Example:**
Sentence: *"I love natural language processing"* (window size = 2)
Context: `["I", "love", "language", "processing"]` → Predicts target: `"natural"`

**Why:** CBOW trains faster than Skip-gram because it averages multiple context signals into a single prediction, making it more efficient on large corpora with frequent words.

**Characteristics:** Faster training, better for frequent words, smaller memory footprint, and adjustable context window size.

**Use Cases:** Large-scale NLP where training speed is a priority, text classification, and general-purpose word embeddings.


## Skip-gram
Skip-gram is a Word2Vec architecture that **predicts the surrounding context words from a given target word**. It trains the model to maximize the probability of observing the actual neighboring words given the center word.

**How it Works:**
1. Take each word in the corpus as the target (center) word.
2. For each target word, generate training pairs with each word within the context window.
3. Train the neural network to predict which words are likely to appear in the context of the target.
4. The learned weight matrix becomes the word embedding matrix.

**Example:**
Sentence: *"I love natural language processing"* (window = 2)
Target: `"language"` → Predict context: `["natural", "processing"]`

**Why:** Skip-gram treats each context word as a separate training example, giving rare words more training signal compared to CBOW. This results in higher-quality embeddings for infrequent vocabulary.

**Characteristics:** Slower than CBOW; significantly better for rare words; handles larger vocabularies more effectively.

**Use Cases:** Semantic similarity, machine translation, recommendation systems, question answering, and any task where rare word quality matters.


## GloVe (Global Vectors for Word Representation)
GloVe is a **count-based word embedding technique** developed at Stanford that generates dense vector representations by analyzing global word co-occurrence statistics across the entire corpus. Unlike Word2Vec (which uses local context windows), GloVe uses corpus-wide co-occurrence frequencies.

**How it Works:**
1. Construct a **word-word co-occurrence matrix** from the corpus — each cell records how often word i co-occurs with word j across all documents.
2. Apply a weighted least-squares objective to factorize this matrix into lower-dimensional word vectors.
3. The resulting vectors are trained such that their dot product approximates the log of co-occurrence probability.

**Example:**
`Vector("King") − Vector("Man") + Vector("Woman") ≈ Vector("Queen")`
GloVe captures the same semantic analogies as Word2Vec, but learns them from global statistics rather than local context windows.

**Why:** GloVe effectively combines the benefits of count-based methods (corpus-wide statistics) with the efficiency of predictive methods, producing high-quality embeddings that capture both global and local context patterns.

**Limitations:** Produces static embeddings (one fixed vector per word); requires a large corpus; less flexible for domain-specific vocabulary without retraining.

**Use Cases:** Text classification, semantic similarity, sentiment analysis, NLP deep learning models, and anywhere pre-trained general-purpose embeddings are needed.


## FastText
FastText is a word embedding technique developed by **Facebook AI Research** that extends Word2Vec by representing words as the **sum of their character n-gram vectors** rather than treating each word as an atomic unit. This subword approach enables it to generate embeddings for rare and unseen words.

**How it Works:**
1. Decompose each word into its character n-grams.
   Example: `"apple"` (n=3) → `["<ap", "app", "ppl", "ple", "le>"]` (including boundary markers `<` and `>`)
2. Learn a separate embedding for each character n-gram.
3. Represent the full word vector as the **sum of all its n-gram embeddings**.
4. Train using Skip-gram or CBOW objectives on these subword representations.

**Why:** Traditional embeddings cannot handle words they have never seen during training (OOV problem). FastText generates embeddings for any word — even misspellings or new words — by composing known n-gram vectors.

**Example:** The model has never seen `"unhappily"`, but can estimate its vector from known n-grams like `"un"`, `"happy"`, `"ily"`.

**Limitations:** Slightly slower to train than Word2Vec; still produces static embeddings (context-independent); larger model size due to n-gram vocabulary.

**Use Cases:** Morphologically rich languages (e.g., Turkish, Finnish, Arabic), OOV-robust NLP systems, text classification, language identification, and spell-tolerant search.


## Contextual Word Embeddings
Contextual word embeddings are **dynamic word representations that change based on the surrounding context**. Unlike static embeddings (Word2Vec, GloVe, FastText) which assign one fixed vector per word regardless of usage, contextual embeddings generate a different vector for the same word in different sentences.

**How it Works:**
Deep learning models (ELMo, BERT, GPT) process the entire sentence and compute each word's embedding based on all surrounding words using mechanisms like bidirectional LSTMs or self-attention.

**Example:**
Word: `"bank"`
- *"I went to the **bank** to deposit money."* → embedding encodes financial institution meaning
- *"The river **bank** was covered in mud."* → embedding encodes geographical feature meaning

Both sentences use the same word, but produce entirely different vectors.

**Models:** ELMo (uses bidirectional LSTMs), BERT (uses bidirectional Transformers), GPT (uses unidirectional Transformer).

**Why:** Polysemy (multiple meanings for one word) is a fundamental challenge in NLP. Static embeddings cannot resolve it; contextual embeddings solve it by considering the full sentence context.

**Limitations:** Computationally expensive to generate; require large pre-trained models; high memory and GPU requirements.

**Use Cases:** Question answering, machine translation, NER, sentiment analysis, text summarization, and all modern state-of-the-art NLP tasks.


## Sentence Embeddings
Sentence embeddings are **dense vector representations of entire sentences** that encode the overall semantic meaning of a sentence in a single fixed-length vector — going beyond word-level representations to capture sentence-level context and intent.

**How it Works:**
Dedicated sentence embedding models (not just word embedding averages) encode sentences by considering:
- The meaning of each word in context
- The relationships and structure between words
- The overall intent and topic of the full sentence

**Models:** Universal Sentence Encoder (USE), Sentence-BERT (SBERT), InferSent.

**Example:**
- Sentence 1: *"I love machine learning."*
- Sentence 2: *"Machine learning is my passion."*
→ Despite using different words, their sentence vectors are close in embedding space, reflecting near-identical meaning.
→ Sentence 3: *"I enjoy cooking pasta."* → vector is far from Sentences 1 and 2.

**Why:** Word embeddings cannot represent sentence meaning — averaging word vectors loses grammatical structure, negation, and sentence-level context. Sentence embeddings capture holistic meaning for comparison and retrieval.

**Limitations:** May lose fine-grained word-level nuance; some models struggle with very long sentences; domain-specific language may not be well represented by general-purpose pre-trained models.

**Use Cases:** Semantic similarity search, document clustering, duplicate detection, paraphrase identification, question answering retrieval, and recommendation systems.


# Evaluation Metrics for NLP

## Accuracy
Accuracy is the most basic classification metric that measures the **proportion of correct predictions** out of all predictions made.

$$Accuracy = \frac{TP + TN}{TP + TN + FP + FN}$$

Where:
- **TP (True Positive):** Model predicted Positive, actual is Positive ✓
- **TN (True Negative):** Model predicted Negative, actual is Negative ✓
- **FP (False Positive):** Model predicted Positive, actual is Negative ✗ (Type I Error)
- **FN (False Negative):** Model predicted Negative, actual is Positive ✗ (Type II Error)

**Example:** A model makes 100 predictions and gets 85 correct → Accuracy = 85/100 = **85%**

**When to use:** Only reliable when the dataset is **balanced** (roughly equal class distribution).

**Limitation:** On an imbalanced dataset (e.g., 95% negative, 5% positive), a model that always predicts "Negative" achieves 95% accuracy while being completely useless for detecting the minority class.


## Precision
Precision measures **how many of the model's positive predictions were actually correct**. It answers the question: *"Of all the cases the model flagged as positive, how many truly were positive?"*

$$Precision = \frac{TP}{TP + FP}$$

**Example:**
An email spam filter flags 100 emails as spam.
- 90 are actually spam (TP = 90)
- 10 are legitimate emails incorrectly flagged (FP = 10)
- Precision = 90 / (90 + 10) = **90%**

**When to use:** When the **cost of a False Positive is high**.
- Spam detection: Sending a real email to spam is costly (False Positive is bad).
- Medical screening: Incorrectly diagnosing a healthy patient costs time and money.

**Limitation:** A model can achieve 100% precision by making only one prediction it is absolutely certain about, while missing all other positive cases (Low Recall).


## Recall (Sensitivity)
Recall measures **how many of the actual positive cases the model successfully identified**. It answers the question: *"Of all the truly positive cases, how many did the model catch?"*

$$Recall = \frac{TP}{TP + FN}$$

**Example:**
100 patients have a disease. The model detects 80 of them (TP = 80) and misses 20 (FN = 20).
Recall = 80 / (80 + 20) = **80%**

**When to use:** When the **cost of a False Negative is high**.
- Medical diagnosis: Missing a real disease case (False Negative) is dangerous.
- Fraud detection: Missing a fraudulent transaction lets the fraudster succeed.
- Safety systems: Missing a defective product reaching consumers is unacceptable.

**Limitation:** Maximizing recall (catching every positive case) inevitably increases False Positives — flagging many legitimate cases as positive (Low Precision).


## F1-Score
The F1-Score is the **harmonic mean of Precision and Recall**, providing a single balanced metric that penalizes extreme imbalance between the two.

$$F1\text{-}Score = 2 \times \frac{Precision \times Recall}{Precision + Recall}$$

**Why the harmonic mean (not arithmetic mean)?**
The harmonic mean is sensitive to low values. If Precision = 1.0 and Recall = 0.0:
- Arithmetic mean = (1.0 + 0.0) / 2 = **0.50** (misleadingly high)
- Harmonic mean (F1) = 2 × (1.0 × 0.0) / (1.0 + 0.0) = **0.00** (correctly reflects failure)

**Example:**
Precision = 0.80, Recall = 0.70
F1 = 2 × (0.80 × 0.70) / (0.80 + 0.70) = 2 × 0.56 / 1.50 ≈ **0.747**

**When to use:** When both Precision and Recall are important, especially on imbalanced datasets where Accuracy alone is misleading.

**Limitation:** Treats Precision and Recall as equally important. When one matters more than the other, use F-beta score: F_β = (1 + β²) × (Precision × Recall) / (β² × Precision + Recall).


## Confusion Matrix
A Confusion Matrix is a **summary table that visualizes the complete performance of a classification model** by showing the counts of correct and incorrect predictions broken down by each class.

**Structure (Binary Classification):**

|  | Predicted Positive | Predicted Negative |
|--|---|---|
| **Actual Positive** | TP (True Positive) | FN (False Negative — Type II Error) |
| **Actual Negative** | FP (False Positive — Type I Error) | TN (True Negative) |

**Example:**
A disease detection model tested on 200 patients:
- 90 sick patients correctly identified (TP = 90)
- 10 sick patients missed (FN = 10)
- 15 healthy patients incorrectly flagged (FP = 15)
- 85 healthy patients correctly cleared (TN = 85)

From this, we can compute: Accuracy = 87.5%, Precision = 85.7%, Recall = 90%.

**Why:** A confusion matrix reveals *where* a model makes mistakes — whether it over-predicts positives (high FP) or misses actual positives (high FN). This guides targeted improvements.

**Limitation:** Becomes difficult to interpret visually for multi-class problems (e.g., a 10-class confusion matrix has 100 cells).


## BLEU (Bilingual Evaluation Understudy)
BLEU is an automated evaluation metric used to measure the **quality of machine-generated text by comparing it against one or more human reference translations**. It is the standard metric for machine translation evaluation.

**How it Works:**
BLEU measures precision-based n-gram overlap between the machine output and reference translations. It checks what fraction of the model's n-grams appear in the reference.

$$BLEU = BP \cdot \exp\left(\sum_{n=1}^{N} w_n \log p_n\right)$$

Where:
- $p_n$ = modified n-gram precision (prevents gaming by clipping matches)
- $BP$ = Brevity Penalty (penalizes outputs shorter than the reference)
- $w_n$ = weight for each n-gram level (typically equal weights)

**Example:**
Reference: *"The cat is on the mat."*
Model output: *"The cat sat on the mat."* → High BLEU (only `"sat"` differs)
Model output: *"Cat mat."* → Low BLEU (short, few matching n-grams) + Brevity Penalty

**Why:** Manual human evaluation of translation quality is slow and expensive. BLEU provides a fast, reproducible, language-independent metric for comparing and benchmarking models.

**Limitations:**
- Does not account for synonyms — `"huge"` vs. `"large"` lowers the score even if the meaning is identical.
- A sentence can achieve high BLEU but still sound unnatural or grammatically awkward.
- Poorly correlated with human judgment at the sentence level.

**Use Cases:** Machine translation model comparison, neural machine translation (NMT) benchmarking, and text generation evaluation.


## ROUGE (Recall-Oriented Understudy for Gisting Evaluation)
ROUGE is the standard evaluation metric for **text summarization**. Unlike BLEU (which focuses on precision), ROUGE focuses on **recall** — measuring how much of the human reference summary is captured by the model's generated summary.

**Key Variants:**

| Variant | What it Measures |
|---------|-----------------|
| **ROUGE-1** | Unigram (individual word) overlap |
| **ROUGE-2** | Bigram (word pair) overlap |
| **ROUGE-L** | Longest Common Subsequence (LCS) — rewards in-order matches |
| **ROUGE-W** | Weighted LCS — gives extra credit for consecutive word matches |
| **ROUGE-S** | Skip-bigram overlap — allows gaps between matched word pairs |

**Formula (ROUGE-N):**
$$ROUGE\text{-}N = \frac{\text{Number of matching n-grams between output and reference}}{\text{Total n-grams in the reference}}$$

**Example:**
Reference: *"The cat sat on the mat."*
Model output: *"The cat is on the mat."* → ROUGE-1 = 5/6 ≈ 0.83 (5 matching unigrams out of 6 reference unigrams)

**Why:** Summarization requires capturing key information from source text. ROUGE measures how well the model preserves the essential content of the reference, making it the natural metric for recall-heavy generation tasks.

**Limitations:** Does not account for paraphrases or synonyms; high ROUGE does not guarantee grammatical fluency or factual correctness.

**Use Cases:** Abstractive and extractive text summarization, news headline generation, and document condensation evaluation.


## Perplexity
Perplexity (PPL) is a metric used to evaluate **language models** by measuring how well a model predicts an unseen sequence of words. Intuitively, it measures how "surprised" or "uncertain" the model is when it encounters the test text.

**Formula:**
$$PP(W) = P(w_1, w_2, ..., w_N)^{-\frac{1}{N}} = \exp \left( -\frac{1}{N} \sum_{i=1}^{N} \log P(w_i) \right)$$

Where N is the number of words and P(w_i) is the model's predicted probability for each word.

**Interpretation:**
- **Lower PPL = Better model** — the model assigns high probability to the actual next words (less surprised).
- **Higher PPL = Worse model** — the model is uncertain and assigns low probability to the actual words.
- A perplexity of K means the model is as confused as if it were randomly choosing from K equally likely options at each step.

**Example:**
- PPL = 10: The model is effectively choosing from ~10 candidates per word (good for a specific domain).
- PPL = 1000: The model has very little idea what word comes next (poor performance).

**Why:** Perplexity provides a direct measure of language model quality on unseen data without requiring task-specific labels. It is used to compare models before deploying them for downstream tasks.

**Use Cases:** Evaluating and comparing language models (GPT, LSTM, n-gram models), hyperparameter tuning, and model selection before fine-tuning.


# Differences

## Rule-Based NLP vs Statistical NLP vs Neural NLP
Rule-based NLP relies entirely on hand-crafted grammar rules and linguistic patterns written by experts, requiring no training data and offering full transparency, but it is rigid, difficult to scale, and breaks down on language variations it was not explicitly programmed to handle. Statistical NLP learns probabilistic patterns from data using methods like Naive Bayes, Hidden Markov Models (HMMs), and TF-IDF, offering more flexibility and adaptability at the cost of requiring a moderate amount of labeled training data. Neural NLP uses deep learning architectures — RNNs, LSTMs, and Transformers — that automatically learn complex, hierarchical representations from massive datasets, achieving state-of-the-art accuracy and scalability across virtually all NLP tasks. The key trade-off is that as you move from rule-based to neural, accuracy and flexibility increase dramatically, but so do data requirements, computational cost, and interpretability challenges.


## NLP vs NLU vs NLI
NLP (Natural Language Processing) is the broad umbrella field covering all computational techniques for processing, analyzing, and generating human language — including preprocessing (tokenization, parsing), feature extraction (embeddings), and generation. NLU (Natural Language Understanding) is a subfield of NLP that focuses specifically on making machines comprehend the meaning, intent, sentiment, and context behind text — going beyond surface-level processing to genuine semantic understanding. NLI (Natural Language Inference) is a specific reasoning task within NLU that determines the logical relationship between two sentences — whether a hypothesis is *entailed* (must be true), *contradicted* (must be false), or *neutral* (undetermined) given a premise. In summary, NLP handles the full processing pipeline, NLU handles meaning comprehension within that pipeline, and NLI is a specific reasoning capability within NLU focused on logical entailment.


## Bag of Words vs TF-IDF
Bag of Words (BoW) represents documents as raw word frequency counts, treating every word equally regardless of how common or rare it is across the entire corpus, which means very frequent words like `"the"` and `"is"` dominate the representation even though they carry no distinctive information. TF-IDF improves on BoW by weighting each word's frequency against its rarity across the corpus — words that appear in nearly every document receive very low scores, while words unique to specific documents receive high scores, highlighting what is truly distinctive about each document. BoW is simpler, faster, and requires no corpus-level statistics, making it suitable as a quick baseline, but it is easily overwhelmed by common words even after stop word removal. TF-IDF is more computationally involved but produces significantly more informative representations and is preferred for tasks like search engine ranking, keyword extraction, and document similarity where distinguishing features matter.


## CBOW vs Skip-gram
CBOW (Continuous Bag of Words) predicts a single target word from multiple surrounding context words by averaging their embeddings, making it computationally efficient and well-suited for frequent words that appear with consistent context patterns. Skip-gram does the reverse — it predicts multiple context words from a single target word, generating one training example per context-target pair, which gives rare words more training signal and results in richer semantic representations. CBOW trains faster because it makes one prediction per window, making it the preferred choice when training speed and corpus size are primary concerns. Skip-gram is slower to train but produces higher-quality embeddings for rare words and captures more nuanced semantic and syntactic relationships, making it the better choice when embedding quality for the full vocabulary — including infrequent words — is the priority.


## Static Embeddings vs Contextual Embeddings
Static embeddings (Word2Vec, GloVe, FastText) assign a single, fixed vector to each word that never changes regardless of the sentence context — the word `"bank"` always has the same representation whether it refers to a financial institution or a river bank, making them unable to resolve polysemy. Contextual embeddings (BERT, ELMo, GPT) dynamically generate a different vector for each word occurrence based on the full surrounding sentence, capturing context-dependent meaning and correctly distinguishing between different senses of the same word. Static embeddings are computationally lightweight, fast to generate, and require only standard hardware, making them practical for simple classification and similarity tasks on moderate-sized datasets. Contextual embeddings achieve superior performance on virtually all complex NLP benchmarks — including question answering, NER, and sentiment analysis — but require significantly more computational resources, larger model sizes, and GPU infrastructure.


## Tokenization vs Sentence Segmentation
Tokenization is a fine-grained process that splits text into individual units called tokens — typically words, subwords, or characters — and is the foundational step that converts raw text into discrete, processable elements for all subsequent NLP operations. Sentence segmentation is a coarser process that identifies the boundaries between complete sentences within a block of text, helping NLP systems understand where one thought ends and the next begins. Tokenization operates at the word or character level and focuses on splitting within sentences, while sentence segmentation operates at the sentence level and focuses on splitting between sentences — making them complementary but distinct operations in the preprocessing pipeline. Tokenization is relatively straightforward (splitting on whitespace and punctuation), while sentence segmentation is more challenging due to ambiguous punctuation (e.g., `"Dr. Smith visited the U.S.A."` contains periods that are not sentence boundaries).


## Stemming vs Lemmatization
Stemming reduces words to their root form by mechanically chopping off suffixes and prefixes using heuristic rules — it is fast and simple but often produces non-real words (e.g., `"studies"` → `"studi"`, `"happiness"` → `"happi"`), and it ignores the word's part of speech and actual meaning entirely. Lemmatization reduces words to their proper dictionary base form (lemma) by applying morphological analysis and considering the word's context and grammatical role — it always produces valid words (e.g., `"studies"` → `"study"`, `"better"` → `"good"`, `"was"` → `"be"`). Stemming is significantly faster and computationally cheaper, making it the preferred choice for large-scale tasks like search engine indexing where approximate matching is acceptable and speed is critical. Lemmatization is slower and requires linguistic resources (a dictionary and POS tagger), but produces semantically correct base forms that are essential for tasks requiring precise language understanding, such as chatbots, machine translation, and question answering.


## POS Tagging vs Chunking
POS tagging operates at the **word level**, assigning a grammatical category (noun, verb, adjective, etc.) to each individual word in a sentence independently, providing a flat, word-by-word annotation of grammatical roles. Chunking (shallow parsing) operates at the **phrase level**, grouping sequences of words with their POS tags into meaningful syntactic phrases — such as noun phrases (NP) or verb phrases (VP) — building a slightly higher-level structure than POS tagging alone. POS tagging is a prerequisite for chunking: chunking rules typically operate on sequences of POS tags rather than raw words, so POS tagging must be performed first. POS tagging is used broadly for syntax analysis, lemmatization, and NLP preprocessing pipelines, while chunking is applied for more specific tasks like named entity recognition, relation extraction, and identifying key phrases for information retrieval.


## Constituency Parsing vs Dependency Parsing
Constituency parsing analyzes a sentence by identifying its **hierarchical phrase structure** — breaking the sentence into nested constituents (NP, VP, PP) arranged in a tree where phrases are grouped progressively from words to clauses. Dependency parsing analyzes a sentence by identifying **grammatical relationships between individual word pairs** — representing the sentence as a directed graph where each word (except the root) has exactly one head word it depends on, labeled with the type of grammatical relation. Constituency parsing emphasizes *how words group into phrases and clauses* and produces a phrase structure tree useful for grammar checking and structural analysis, while dependency parsing emphasizes *who depends on whom* and produces a dependency graph directly useful for extracting subject-verb-object triples and semantic roles. Constituency parsing is commonly used in grammar-based machine translation and language understanding research, while dependency parsing is preferred in practical NLP systems for information extraction, relation extraction, and semantic role labeling due to its more direct representation of meaning.


## Named Entity Recognition vs Coreference Resolution
Named Entity Recognition (NER) detects and classifies **specific named entities** within individual text spans — identifying that `"Elon Musk"` is a PERSON, `"Tesla"` is an ORG, and `"Berlin"` is a GPE — operating at the word or phrase level within a single mention. Coreference resolution identifies **when multiple different expressions throughout a document refer to the same real-world entity** — for example, linking `"Elon Musk"`, `"he"`, `"the CEO"`, and `"Musk"` across multiple sentences to the same individual. NER focuses on recognizing and classifying entities at a single mention point and does not track how the same entity is referred to elsewhere in the document, while coreference resolution explicitly builds chains of references across the entire document. NER is relatively simpler and can be implemented reliably with sequence labeling models (CRF, BiLSTM-CRF, BERT), whereas coreference resolution is significantly more complex because it requires understanding context across multiple sentences and resolving ambiguous pronoun references.


## Semantic Analysis vs Syntactic Analysis
Syntactic analysis examines the **grammatical structure and form** of a sentence — how words are arranged according to language rules — using techniques like POS tagging, constituency parsing, and dependency parsing to produce structural representations such as parse trees and dependency graphs. Semantic analysis examines the **meaning and interpretation** of a sentence — what the words and their relationships actually convey — using techniques like WSD, NER, and semantic role labeling to produce meaning representations such as logical forms, entity labels, and semantic roles. A sentence can be syntactically correct but semantically nonsensical (e.g., *"Colorless green ideas sleep furiously"* is grammatically perfect but meaningless), and a sentence can also be syntactically awkward but semantically clear — showing that the two dimensions are complementary, not interchangeable. Syntactic analysis is applied in grammar checking, rule-based NLP, and as a preprocessing step for deeper analysis, while semantic analysis powers higher-level tasks such as question answering, machine translation, information extraction, and sentiment analysis.


## Sparse Vectorization vs Dense Vectorization
Sparse vectorization (BoW, TF-IDF, one-hot encoding) represents text as very high-dimensional vectors — typically with dimensionality equal to the full vocabulary size — where the vast majority of entries are zero because most words do not appear in any given document, making these representations memory-intensive and semantically impoverished. Dense vectorization (Word2Vec, GloVe, FastText, BERT embeddings) represents text as compact, low-dimensional vectors (typically 100–1024 dimensions) where most entries are non-zero and each dimension encodes a learned semantic or syntactic feature, capturing rich linguistic relationships in a memory-efficient format. Sparse methods are fast to compute, easy to interpret, and work well for traditional machine learning models (logistic regression, SVM) on simple tasks, but they completely fail to capture semantic similarity — `"car"` and `"automobile"` appear as completely unrelated in a sparse representation. Dense methods require significantly more computation to generate but enable meaningful similarity computations (cosine similarity of dense vectors is semantically informative), support transfer learning, and power all modern state-of-the-art NLP systems.


## Word2Vec vs GloVe vs FastText
Word2Vec is a **predictive model** that learns word embeddings by training a shallow neural network to predict words from their local context window (via CBOW or Skip-gram), producing high-quality embeddings for common words but struggling with rare and out-of-vocabulary (OOV) words since each word is treated as an indivisible unit with one fixed vector. GloVe is a **count-based model** that constructs a global word co-occurrence matrix from the entire corpus and learns word vectors by factorizing this matrix, capturing corpus-wide statistical relationships more efficiently than Word2Vec's local window approach — but it also assigns one fixed vector per word and cannot handle OOV terms. FastText is an **extension of Word2Vec** that represents every word as the sum of its character n-gram embeddings, enabling it to generate meaningful vectors for rare words, misspellings, and completely unseen words by composing known n-gram vectors — making it the most robust of the three for real-world noisy text. In summary, for standard NLP tasks with a clean, known vocabulary, Word2Vec and GloVe perform comparably well; when OOV robustness or morphologically rich languages are involved, FastText is the clear preferred choice.
