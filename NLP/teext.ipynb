## Tokenization

Tokenization is the process of splitting text into smaller units called tokens, which are the basic building blocks for NLP models. Tokens can be words, subwords, characters, or bytes, depending on the method used.

1️⃣ Why Tokenization is Important

- ⚡ Model Input: NLP models cannot process raw text directly; tokens are converted into numerical embeddings.

- 🧠 Context Understanding: Tokens allow the model to capture relationships and meanings in text.

- 🎯 Flexibility: Subword or byte-level tokenization can handle rare words, misspellings, and multiple languages efficiently.

2️⃣ Types of Tokenization

- Word-level: Splits text by spaces/punctuation. Example: "I love AI" → ["I", "love", "AI"]

- Subword-level: Splits words into smaller, frequent units. Example: "Transformers" → ["Trans", "form", "ers"]

- Methods include BPE, WordPiece, SentencePiece.

- Byte-level: Splits text into bytes, works with any language or symbol. Example: "AI" → [b'A', b'I']

3️⃣ Common Subword Tokenization Methods

- BPE (Byte-Pair Encoding): Merges the most frequent character pairs.

- WordPiece: Merges subwords based on likelihood in training data.

- SentencePiece: Language-agnostic, works without pre-tokenization, can operate at byte or character level.

4️⃣ Tokenization Flow

- Raw text → Tokenization → basic units (words/subwords/bytes).

- Tokens → numerical embeddings.

- Embeddings → fed into Transformers for attention, context learning, and sequence modeling.

## Stopwords

Stopwords are common words in a language (like “the”, “is”, “and”) that carry little semantic meaning in many NLP tasks.
Stopword removal is the process of filtering these words out from text to focus on meaningful content.

1️⃣ Why Stopword Removal is Important

- ⚡ Reduces Noise: Removes words that do not add value to tasks like classification, search, or clustering.

- 🧠 Improves Model Focus: Lets the model focus on keywords and important tokens.

- 💾 Reduces Dimensionality: Fewer tokens → smaller vocabulary → faster training and inference.

2️⃣ How It Works

- Start with a text corpus.

- Compare each word against a stopword list (language-specific).

- Remove words that match the stopword list.

- Keep remaining words for processing (tokenization, embeddings, etc.).

3️⃣ Example

- Original sentence:

  - "The cat is sitting on the mat"

- After stopword removal:

  - "cat sitting mat"

Words like “the” and “is” are removed.

4️⃣ When to Be Careful

- Some tasks need stopwords for meaning (e.g., language generation, sentiment analysis, NER).

- Blindly removing stopwords can sometimes hurt model performance.

## Stemming vs Lemmatization

- Stemming reduces words to their root form by chopping off suffixes or prefixes, often producing non-dictionary words. Lemmatization reduces words to their dictionary/base form using linguistic rules, producing valid words.

- Stemming is faster and simpler, while lemmatization is more accurate and context-aware, often requiring part-of-speech information.

- Stemming works well for large-scale text processing or search tasks. Lemmatization is preferred for tasks requiring precise meaning, like NLP pipelines, sentiment analysis, or question-answering.

- Stemming can be imprecise and sometimes produce unnatural roots. Lemmatization is slower but ensures outputs are valid words with correct meanings.

- Stemming is widely used in information retrieval and keyword-based search. Lemmatization is used in language understanding tasks, text normalization, and advanced NLP applications.
