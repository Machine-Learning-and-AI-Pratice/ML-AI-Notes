# Linear Algebra 

Linear algebra is a core mathematical foundation for machine learning, as most datasets and models are represented using vectors and matrices. It allows efficient computation, data manipulation and optimization, making complex tasks manageable.

## Fundamental Concepts in Linear Algebra

### 1. Vectors

Vectors are quantities that have both magnitude and direction, often represented as arrows in space.
$$
v = \begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix}
$$

### 2. Matrices

Matrices are rectangular arrays of numbers, arranged in rows and columns. Matrices are used to represent linear transformations, systems of linear equations and data transformations in machine learning.

$$
\begin{bmatrix}
1 & 4 & 7 \\
2 & 5 & 8 \\
3 & 6 & 9
\end{bmatrix}
$$

### 3. Scalars
Scalars are single numerical values, without direction, magnitude only. Scalars are just single numbers that can multiply vectors or matrices.  
In machine learning, they’re used to adjust things like the weights in a model or the learning rate during training

We have a scalar $k = 3$ and a vector  

$$
v = \begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix}
$$

Multiplying the scalar with the vector:

$$
k \cdot v = 3 \cdot 
\begin{bmatrix}
2 \\
-1 \\
4
\end{bmatrix}
=
\begin{bmatrix}
3 \cdot 2 \\
3 \cdot (-1) \\
3 \cdot 4
\end{bmatrix}
=
\begin{bmatrix}
6 \\
-3 \\
12
\end{bmatrix}
$$

## Operations in Linear Algebra

### 1. Addition & Subtraction
Add or subtract corresponding elements of vectors/matrices.

$$
u = \begin{bmatrix} 2 \\ -1 \\ 4 \end{bmatrix}, \quad 
v = \begin{bmatrix} 3 \\ 0 \\ -2 \end{bmatrix}
$$

$$
u + v = \begin{bmatrix} 5 \\ -1 \\ 2 \end{bmatrix}, \quad 
u - v = \begin{bmatrix} -1 \\ -1 \\ 6 \end{bmatrix}
$$

### 2. Scalar Multiplication
Multiply each element by a scalar.

$$
3 \cdot \begin{bmatrix} 2 \\ -1 \\ 4 \end{bmatrix}
= \begin{bmatrix} 6 \\ -3 \\ 12 \end{bmatrix}
$$

### 3. Dot Product
Measures similarity of directions by multiplying matching elements and summing.

$$
u \cdot v = u_1 v_1 + u_2 v_2 + u_3 v_3
$$

### 4. Cross Product
For 3D vectors, produces a new vector perpendicular to both.

$$
u \times v = 
\begin{bmatrix}
u_2 v_3 - u_3 v_2 \\
u_3 v_1 - u_1 v_3 \\
u_1 v_2 - u_2 v_1
\end{bmatrix}
$$


## Linear Transformations

Linear transformations are basic operations in linear algebra that change vectors and matrices while keeping important properties like straight lines and proportionality. In machine learning, they are key for tasks like preparing data, creating features and training models. 

A transformation $T$ is linear if it satisfies:

1. **Additivity**
$$
T(u + v) = T(u) + T(v)
$$

2. **Homogeneity (Scalar Multiplication)**
$$
T(kv) = k \cdot T(v)
$$

### Common Types in ML

- **Translation**: Centering data by subtracting the mean.
$$
x' = x - \mu
$$

- **Scaling**: Normalizing features so no single feature dominates.
$$
x' = \frac{x}{\|x\|}
$$

- **Rotation**: Turning data, often used in computer vision and robotics.
$$
R(\theta) = 
\begin{bmatrix}
\cos\theta & -\sin\theta \\
\sin\theta & \cos\theta
\end{bmatrix}
$$

$$
x' = R(\theta) \cdot x
$$


## Matrix Operations

Matrix operations are central to linear algebra and widely used in machine learning for data handling, transformations and model training. 

### 1. Matrix Multiplication
Combine two matrices by taking the dot product of rows and columns.

$$
A = \begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}, \quad
B = \begin{bmatrix}
3 & 1 \\
0 & 2
\end{bmatrix}
$$

$$
A \times B =
\begin{bmatrix}
7 & 5 \\
2 & 4
\end{bmatrix}
$$


### 2. Transpose
Flip a matrix across its diagonal (rows become columns). Denoted by 
A^T 
.

$$
A^T = \begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}^T
= \begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}
$$

### 3. Inverse
The matrix A^-1 satisfies:

$$
A \cdot A^{-1} = I
$$

### 4. Determinant
A scalar value indicating whether a matrix is invertible.

$$
\det(A) = ad - bc
$$

## Eigenvalues and Eigenvectors

### Eigenvalues

Eigenvalues are scalar values linked to a matrix or linear transformation. They indicate how much an eigenvector gets stretched or compressed during the transformation.

- If the eigenvalue is positive, the eigenvector’s direction is preserved.

- If the eigenvalue is negative, the eigenvector’s direction is reversed.


### Eigenvectors
Eigenvectors are non-zero vectors that, when multiplied by a matrix, change only in magnitude, not direction.


$$
A v = \lambda v
$$

Where:      

𝐴 = matrix  
𝑣 = eigenvector         
𝜆 = eigenvalue (scalar)         

### Eigen Decomposition
 
Eigen decomposition expresses a square matrix 𝐴 in terms of its eigenvalues and eigenvectors:

$$
A = Q \Lambda Q^{-1}
$$

### Applications in ML:

- Dimensionality Reduction (PCA): Keeps directions with largest eigenvalues (most variance).

- Matrix Factorization (SVD, NMF): Breaks large datasets into smaller, structured parts for feature extraction.

## Solving Linear Systems of equations

### 1. Gaussian Elimination

Transforms a matrix into row‑echelon form using row operations.

- Forward Elimination → make entries below diagonal zero

- Back Substitution → solve variables from last row upward

- Pivoting → swap rows to avoid division by zero

### 2. LU Decomposition
Splits a matrix into Lower (L) and Upper (U) triangular matrices.

$$
A = L U
$$

- 𝐿 = lower triangular matrix

- 𝑈 = upper triangular matrix

- Efficient for solving systems using forward + back substitution.

###  3. QR Decomposition
Splits a matrix into Orthogonal (Q) and Upper triangular (R) matrices.

$$
A = Q R
$$

𝑄 = orthogonal matrix (𝑄^𝑇𝑄=𝐼)

𝑅 = upper triangular matrix

Useful for least squares problems and eigenvalue computation.


# Applications of Linear Algebra in Machine Learning

Linear algebra powers many ML algorithms by enabling data manipulation, model representation, and optimization.  
Key applications include:

---

## 1. Principal Component Analysis (PCA)

Reduces dimensionality by computing covariance, eigenvalues/eigenvectors, and projecting data onto principal components.

- Covariance matrix:  
  $$
  C = \frac{1}{n} X^T X
  $$

- Eigen decomposition:  
  $$
  C v = \lambda v
  $$

- Projection onto principal components:  
  $$
  X' = X W
  $$  
  where \(W\) contains top eigenvectors.

---

## 2. Singular Value Decomposition (SVD)

Factorizes a matrix into orthogonal components, useful for dimensionality reduction, compression, and noise filtering.

- Decomposition:  
  $$
  A = U \Sigma V^T
  $$

- Applications:  
  - Latent semantic analysis  
  - Image compression  
  - Noise reduction

---

## 3. Linear Regression

Models relationships via matrix form.

- Model:  
  $$
  Y = X \beta + \epsilon
  $$

- Normal equation solution:  
  $$
  \hat{\beta} = (X^T X)^{-1} X^T Y
  $$

---

## 4. Support Vector Machines (SVM)

Uses optimization and kernel tricks to find decision boundaries.

- Optimization objective:  
  $$
  \min_{w,b} \frac{1}{2} \|w\|^2
  $$  
  subject to  
  $$
  y_i (w^T x_i + b) \geq 1
  $$

- Kernel trick:  
  Replace dot product with kernel function  
  $$
  K(x_i, x_j) = \phi(x_i)^T \phi(x_j)
  $$

---

## 5. Neural Networks

Depend heavily on matrix operations for training and inference.

- Forward pass:  
  $$
  Z = W X + b
  $$

- Activation:  
  $$
  A = \sigma(Z)
  $$

- Backpropagation (gradient descent):  
  $$
  W := W - \eta \frac{\partial L}{\partial W}
  $$
