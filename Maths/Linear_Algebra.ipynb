# Linear Algebra 

Linear Algebra in Data Science is the branch of mathematics that deals with vectors, matrices, and linear transformations, used to represent, process, and analyze data in computational models.

It is used to:

- Store data (as matrices)
- Perform calculations (like matrix multiplication)
- Build and train machine learning models

## Scalar

A scalar is a single numerical value with no dimensions, used to represent one quantity.

Example: 5, 3.14, -10

It is used to: 

- Count of records (e.g., 5 rows)
- Average/mean of a feature
- Model parameters like intercept (c) in linear regression

## Vector

Numerical value which has both magnitude and direction. 

A vectore is an ordered list of numbers. It can represent a point is space or quantity with both magnitude and direction 

Example : (90,3) IQ=90 study hours = 3 

It is used to: 

- Text processing (NLP)
   - Words/sentences → converted into vectors (embeddings)
- Model calculations
   - Used in operations like dot product, distance, similarity

## Unit Vector 

A unit vector is a vector with magnitude (length) equal to 1, used to show direction only.

If vector **v = (x, y)**, then the unit vector is:

$$\hat{v} = \frac{v}{|v|}$$


        y
        ↑
        |      • (x, y)
        |     /
        |    /   v
        |   /
        |  /
        | /
        |/________→ x
       (0,0)

Unit vector → same direction, length = 1

## Vector Addition (Sum of Two Vectors)

The sum of two vectors is obtained by adding their corresponding components.
