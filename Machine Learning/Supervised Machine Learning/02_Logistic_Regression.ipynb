# Logistic Regression in Machine Learning

------------------------------------------------------------------------

## **Introduction**

Logistic Regression is a **supervised machine learning algorithm** used
for **classification problems**.\
Unlike Linear Regression which predicts continuous values, Logistic
Regression predicts the **probability** that an input belongs to a
specific class.

It is mainly used for **binary classification**, where the output can be
one of two categories such as: 
- Yes/No\
- True/False\
- 0/1

It uses the **sigmoid function** to convert input values into a
probability between **0 and 1**.

------------------------------------------------------------------------

## **Types of Logistic Regression**

### **1. Binomial Logistic Regression**

Used when the dependent variable has only **two categories**.\
Examples: Yes/No, Pass/Fail, 0/1.\
This is the most common form.

### **2. Multinomial Logistic Regression**

Used when the dependent variable has **three or more categories**
**without** order.\
Example: Classifying animals into "cat", "dog", or "sheep".

### **3. Ordinal Logistic Regression**

Used when the dependent variable has **three or more ordered
categories**.\
Example: Ratings like *low*, *medium*, *high*.

------------------------------------------------------------------------

## **Assumptions of Logistic Regression**

1.  **Independent Observations**\
    No two samples should be dependent or correlated.

2.  **Binary Dependent Variable**\
    Logistic regression assumes a binary target (for \>2 categories,
    Softmax is used).

3.  **Linearity with Log-Odds**\
    Predictors must have a linear relationship with the **log-odds** of
    the output.

4.  **No Extreme Outliers**\
    Outliers can distort model coefficients.

5.  **Large Sample Size**\
    A bigger dataset ensures more stable results.

------------------------------------------------------------------------

## **Understanding the Sigmoid Function**

- The sigmoid function is an important component of logistic regression. It converts the raw output (linear combination of features) into a probability value between 0 and 1.

- The function takes any real-valued number and maps it into the range 0 to 1, forming an S-shaped curve, also known as the sigmoid curve or logistic curve. This makes it ideal for classification tasks because probabilities must always lie between 0 and 1.

- In logistic regression, we apply a threshold (commonly 0.5) to the sigmoid output to determine the predicted class:

- If the sigmoid output is greater than or equal to 0.5, the input is classified as Class 1.

- If the sigmoid output is below 0.5, the input is classified as Class 0.

## Mathematical Representation in Logistic Regression

Let:

- `x_i` be the **i-th observation** of input features `X`
- `w = [w1, w2, w3, …, wm]` be the **weights (coefficients)**
- `b` be the **bias term** (also called the intercept)

The linear combination is computed as the **dot product** of weights and features:

\[
z = w \cdot X + b
\]

Here:

- `z` is a **continuous value** from the linear model.
- Logistic Regression then converts this value into a **probability**.

---

## Applying the Sigmoid Function

To convert `z` into a probability between 0 and 1, we use the **sigmoid (logistic) function**:

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

The output `σ(z)` is the **predicted probability (predicted y)**.
