# Introduction to Linear Regression

Linear regression is a type of **supervised machine learning algorithm** used for predicting continuous outcomes. It learns from **labelled datasets** and finds the most optimized **linear function** that maps input features to an output. It assumes a **linear relationship** between input (independent variable) and output (dependent variable), meaning the output changes at a constant rate with respect to the input.

This relationship is visually represented by a **straight line**.

---

## Example: Predicting Exam Scores

Suppose we want to predict a student's exam score based on how many hours they studied. Observations show that as study hours increase, exam scores also tend to increase.

In this case:

* **Independent Variable (Input)**: Hours studied (we control or observe this value)
* **Dependent Variable (Output)**: Exam score (this depends on the input)

Linear regression helps us model this relationship and use it to make predictions for new data.

---

## Why Linear Regression is Important?

Linear regression is one of the foundational techniques in machine learning and statistics. Here’s why it is important:

### **1. Simplicity and Interpretability**

It is easy to understand, visualize, and interpret. This makes it an excellent starting point for beginners.

### **2. Predictive Ability**

It helps predict future outcomes based on historical patterns. Useful in fields like finance, healthcare, and marketing.

### **3. Basis for Other Models**

Many advanced algorithms—such as logistic regression, neural networks, and generalized linear models—are built on top of linear regression concepts.

### **4. Efficiency**

It is computationally lightweight and performs well when the relationship between variables is linear.

### **5. Widely Used**

It is one of the most commonly used algorithms for regression tasks in real-world applications.

### **6. Analytical Insights**

Linear regression provides insights into how input variables influence the output, helping understand patterns and relationships.

## Best Fit Line in Linear Regression

In **linear regression**, the best-fit line is the straight line that most accurately represents the relationship between the **independent variable (input)** and the **dependent variable (output)**.  
It minimizes the difference between the actual data points and the predicted values from the model.

---

## 1 Goal of the Best-Fit Line
- Find a straight line that **minimizes error** between observed data points and predicted values.
- Helps predict the dependent variable for **new, unseen data**.

### Linear Regression in Machine Learning
- **Y** → Dependent / Target variable  
- **X** → Independent / Predictor variable  
- Linear function is the simplest regression function.  
- X can be a **single feature** or **multiple features** representing the problem.




---

## 2 Equation of the Best-Fit Line 

For simple linear regression



Where:
- **y** → Predicted value (dependent variable)  
- **x** → Input (independent variable)  
- **m** → Slope (change in y per unit change in x)  
- **b** → Intercept (value of y when x = 0)  

The best-fit line will be the one that optimizes the values of m (slope) and b (intercept) so that the predicted y values are as close as possible to the actual data points.


--- 

## 3. Minimizing the Error: Least Squares Method
- Method: **Least Squares** → minimizes the sum of squared differences between actual and predicted values.  
- These differences are called **residuals**.



👉 Ensures the line best represents the data by minimizing squared residuals.

---

## 4. Interpretation of the Best-Fit Line

- **Slope (m):** Indicates how much y changes with each unit change in x.  
  - Example: If slope = 5 → For every 1-unit increase in x, y increases by 5 units.
- **Intercept (b):** Predicted value of y when x = 0.  
  - Point where the line crosses the y-axis.

## ⚠️ Limitations of Linear Regression
  - **Assumes Linearity:** Works only if the relationship between variables is linear.  
  - **Sensitivity to Outliers:** Outliers can skew slope and intercept, distorting the best-fit line.

Where:
- **x₁, x₂, ..., xₖ** → Independent variables

---

## 3. 📐 Assumptions of Linear Regression

1. **Linearity** → Relationship between inputs (X) and output (Y) is linear.  

      
2. **Independence of Errors** → Prediction errors should not influence each other.  
3. **Constant Variance (Homoscedasticity)** → Errors should have equal spread across all values of X.  
   - If spread changes (fans out/shrinks), it’s called **heteroscedasticity**.  



4. **Normality of Errors** → Errors should follow a normal (bell-shaped) distribution.  
5. **No Multicollinearity** (for multiple regression) → Input variables shouldn’t be too closely related.  
6. **No Autocorrelation** → Errors shouldn’t show repeating patterns, especially in time-series data.  
7. **Additivity** → Total effect on Y is the sum of effects from each X, with no interaction terms.

# 📊 Types of Linear Regression

Linear regression can be categorized based on the number of independent features:

- **Simple Linear Regression (Univariate Regression):** One independent variable.  
- **Multiple Linear Regression (Multivariate Regression):** More than one independent variable.

---

## 1. 🔹 Simple Linear Regression

Used when we want to predict a target value (dependent variable) using **only one input feature** (independent variable).  
It assumes a **straight-line relationship** between the two.

### Formula:

**Example:** Predicting a person’s salary (y) based on their years of experience (x).

---

## 2. 🔹 Multiple Linear Regression

Involves **more than one independent variable** and one dependent variable.  
The goal is to find the **best-fit line (or hyperplane)** that predicts values based on multiple inputs.

### Formula:


Where:
- **ŷ** → Predicted value  
- **x₁, x₂, …, xₙ** → Independent variables  
- **θ₁, θ₂, …, θₙ** → Coefficients (weights for each predictor)

---

## 3. 🧠 Key Idea of Regression

- Regression uses a dataset with **X (features)** and **Y (target values)**.  
- The algorithm learns a function that maps **X → Y**.  
- Once trained, this function can predict **Y** for new, unseen **X** values.  
- Regression always predicts **continuous values** (not categories).

# Cost Function and Gradient Descent in Linear Regression (Standard Notes)

## 1. Need for a Cost Function

In linear regression, we aim to find the **best-fit line** for a dataset. However, predictions (ŷ) rarely match actual values (y). To measure how far off our predictions are, we use a **cost function** (or loss function). This helps evaluate how well the model performs.

---

## 2. Mean Squared Error (MSE) – The Cost Function

Linear regression uses **Mean Squared Error (MSE)** as its cost function.

### Prediction Equation

```
ŷᵢ = θ₁ + θ₂xᵢ
```

### MSE Formula

```
J(θ₁, θ₂) = (1/n) * Σ (ŷᵢ − yᵢ)²
```

Where:

* **n** = number of data points
* **yᵢ** = actual value
* **ŷᵢ** = predicted value

MSE measures the average squared difference between predictions and actual values. A smaller MSE indicates a better fit.

---

## 3. Gradient Descent for Optimization

After computing the loss using MSE, the next step is to **minimize** it. This is done using **Gradient Descent**, an iterative method for optimization.

### Purpose of Gradient Descent

To find the optimal values of **θ₁** and **θ₂** that minimize the cost function **J**.

### Gradient Descent Update Rule

```
θⱼ := θⱼ - α * (∂J / ∂θⱼ)
```

Where:

* **α** = learning rate
* **∂J / ∂θⱼ** = gradient of the cost function with respect to parameter θⱼ

### How Gradient Descent Works

1. Calculate predictions.
2. Compute the MSE.
3. Calculate gradients.
4. Update θ₁ and θ₂.
5. Repeat until the cost function converges.

---

## 4. Final Outcome

After repeated updates using gradient descent, the parameters converge to values that produce the **minimum MSE**. This results in the **best-fit line** for the given data.

---



# Gradient Descent for Linear Regression

Gradient Descent is an **optimization technique** used to train a linear regression model by minimizing prediction error. It starts with random parameter values and iteratively adjusts them to reduce the difference between predicted and actual values.



---

## ⚙️ How Gradient Descent Works
1. **Initialize parameters** (\(\theta_1\) = intercept, \(\theta_2\) = slope) with random values.  
2. **Calculate error** between predicted values (\(\hat{y}_i\)) and actual values (\(y_i\)).  
3. **Compute gradients** — measure how much each parameter contributes to the error.  
4. **Update parameters** in the direction that reduces error.  
5. **Repeat** until the error converges to the minimum.  

This iterative process helps the model find the **best-fit line** for the data.

J(θ) = (1 / 2m) * Σ (ŷᵢ − yᵢ)² + λ Σ |θⱼ|
```

* First term: Least squares loss (difference between prediction and actual value).
* Second term: L1 penalty (sum of absolute values of coefficients). This can shrink coefficients to zero → feature selection.

---

### 2. Ridge Regression (L2 Regularization)

Ridge Regression adds an L2 penalty, penalizing large coefficients. It is useful when predictors are highly correlated.

**Objective Function:**

```
J(θ) = (1 / 2m) * Σ (ŷᵢ − yᵢ)² + λ Σ θⱼ²
```

* First term: Least squares loss.
* Second term: L2 penalty (sum of squared coefficients). Prevents coefficients from becoming too large but does not set them to zero.

---

### 3. Elastic Net Regression

Elastic Net combines L1 and L2 regularization.

**Objective Function:**

```
J(θ) = (1 / 2m) * Σ (ŷᵢ − yᵢ)² + αλ Σ |θⱼ| + (1/2)(1 − α)λ Σ θⱼ²
```

* First term: Least squares loss.
* Second term: L1 (Lasso) part.
* Third term: L2 (Ridge) part.
* **λ** controls total regularization strength.
* **α** controls L1 vs L2 mix.

  * α = 1 → Lasso
  * α = 0 → Ridge
