# Machine Learning

## What Is Machine Learning?

Machine learning is the science and art of programming computers so that they can learn directly from data, rather than relying on explicitly coded rules


### Why Use Machine Learning? 

Machine learning is incredibly powerful and excels in scenarios where traditional programming falls short It is especially useful for:

1. **Simplifying complex rules**: It can replace long, hard-to-maintain lists of hard-coded rules with simpler models that perform better

2. **Solving problems with no known algorithms**: It can find solutions for highly complex tasks, like speech recognition, where traditional approaches fail

3. **Adapting to change**: Models can be easily retrained on new data to keep up with fluctuating environments

4. **Data mining**: It can analyze massive datasets to uncover hidden patterns and provide valuable insights

## Types of Machine Learning Systems

### 1. Training Supervision 
Systems can be classified by the amount and type of supervision provided during the training phase:

- **Supervised learning**: The training data fed to the algorithm includes the desired solutions, which are known as labels. The two most common supervised tasks are **classification** (e.g., predicting whether an email is spam or ham) and **regression** (predicting a target numeric value, such as a car's price based on its features).

- **Unsupervised learning**: The training data is entirely unlabeled, meaning the system tries to learn the underlying structure without a teacher. Typical tasks include **clustering** (grouping similar instances together), **visualization**, **dimensionality reduction** (simplifying data without losing much information), and **anomaly detection** (detecting rare or unusual instances like credit card fraud).

- **Semi-supervised learning**: These algorithms can handle data that is only partially labeled, which is useful because labeling data is often time-consuming and expensive. They typically combine unsupervised techniques (like clustering) with supervised learning to propagate the few available labels to the rest of the unlabeled data.

- **Self-supervised learning**: This approach involves generating a fully labeled dataset from a fully unlabeled one. For example, a model might be trained to repair randomly masked parts of an image by using the original, unmasked image as the target label.

- **Reinforcement learning**: In this highly distinct approach, a learning system called an agent observes its environment, takes actions, and receives positive rewards or negative penalties in return. Through trial and error, it learns the best strategy—called a policy—to maximize its rewards over time. AlphaGo is a famous example of this type of system

### 2. Batch Versus Online Learning
This criterion focuses on whether the system can learn continuously from a stream of incoming data:

- **Batch learning**: The system is incapable of learning incrementally; it must be trained from scratch using all available data. Because this process is time-consuming and resource-intensive, it is typically done offline. Once launched, the model applies what it has learned but does not continue to learn.

- **Online learning**: The system is trained incrementally by feeding it data instances sequentially, either one by one or in small groups called mini-batches. This is fast and cheap, allowing the system to adapt to new data on the fly. Online learning is particularly useful for systems dealing with rapidly changing data (like the stock market) or for handling massive datasets that cannot fit into a single machine's memory (known as out-of-core learning)

### 3. Instance-Based Versus Model-Based Learning
This classification depends on the approach the system uses to generalize and make predictions on new, unseen examples:

- **Instance-based learning**: The system learns the training examples by heart. When presented with a new case, it uses a similarity measure to compare the new data point to the memorized examples to make a prediction.

- **Model-based learning**: Instead of just comparing new points to old ones, the algorithm detects patterns in the training data and builds a generalized predictive model. You specify a performance measure (such as a cost function to minimize errors) and train the model to find the best parameter values to make accurate predictions on new data


## Main Challenges of Machine Learning

### 1. Challenges Related to Bad Data:

- **Insufficient Quantity of Training Data**: Unlike human toddlers who can learn to recognize an apple from just a few examples, machine learning algorithms are not quite there yet
. Even very simple problems typically require thousands of examples, and complex problems like image or speech recognition may require millions of examples to work properly.

- **Nonrepresentative Training Data**: For a model to generalize well, the training data must perfectly represent the new cases you want to predict. If your sample is too small, you will have sampling noise (nonrepresentative data due to chance), and even large samples can be nonrepresentative if the sampling method is flawed, which is known as sampling bias.

- **Poor-Quality Data**: If your training data is full of errors, outliers, and noise (such as poor-quality measurements), it becomes incredibly difficult for the system to detect the true underlying patterns. Data scientists often spend a significant portion of their time cleaning up training data to discard outliers or fill in missing features.

- **Irrelevant Features**: A system will only be capable of learning if the data contains enough relevant features and not too many irrelevant ones (garbage in, garbage out). You must perform feature engineering to extract and select the most useful features for your model to train on

### 2. Challenges Related to Bad Models:

- **Overfitting the Training Data**: This occurs when a model is too complex relative to the amount and noisiness of the training data. The model performs exceptionally well on the training data by memorizing it (including the noise), but it fails to generalize to new, unseen instances. You can fix overfitting by simplifying the model (a process called regularization), gathering more training data, or reducing the noise in your current data.

- **Underfitting the Training Data**: This is the exact opposite of overfitting. It happens when your model is too simple to learn the underlying structure of the data, resulting in inaccurate predictions even on the training examples. To fix underfitting, you can select a more powerful model with more parameters, engineer better features, or reduce the constraints (regularization) placed on the model


## Testing and Validating

To evaluate how well a machine learning model will perform on new, unseen cases, you must test it before putting it into the real world, as deploying a horribly bad model would quickly lead to user complaints
. Data scientists do this by strategically splitting their data into different sets:
The Training Set and Test Set: The most common approach is to split your available data into two sets, typically using 80% for training and holding out 20% for testing
. You train your model using the training set and evaluate it on the test set to find its generalization error (how well it handles new instances it has never seen before)
. If your model has a low error rate on the training set but a high generalization error on the test set, it means the model is overfitting the training data
.
The Validation Set: When deciding between different models or tweaking settings (hyperparameters), you should never use your test set, because the model will end up adapting to that specific set and will perform poorly on truly new data
. Instead, you use a technique called holdout validation by carving out a validation set (also called a dev set) from your training data
. You train your candidate models on the reduced training set, evaluate them on the validation set to pick the best one, and then retrain the winning model on the full training set
. Finally, you evaluate this model on the test set just once
.
Cross-Validation: If your validation set is too small, your evaluations might be imprecise, but if it is too large, you might not have enough data left to properly train your model
. To solve this, you can use cross-validation, which evaluates the model across many small validation sets and averages the results to give you a much more accurate measure of its performance
.
The Train-Dev Set for Data Mismatch: Sometimes your training data (e.g., millions of web pictures) isn't perfectly representative of the data the model will encounter in production (e.g., pictures taken on a mobile app)
. If your model performs poorly on the validation set, you won't know if it is overfitting the training data or if there is simply a data mismatch
. To diagnose this, you hold back a train-dev set from your abundant training data
. If the model performs poorly on this train-dev set, it is overfitting
. If it performs well on the train-dev set but still fails on the validation set, you know the problem is specifically a data mismatch
