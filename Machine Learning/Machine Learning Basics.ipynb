# Machine Learning Tutorial

Machine learning is a branch of Artificial Intelligence that focuses on developing models and algorithms that let computers learn from data without being explicitly programmed for every task.

## Types of Machine Learning

- Supervised Machine Learning
- Unsupervised Machine Learning
- Reinforcement Learning
- Semi-Supervised Learning 
- Self-Supervised Learning

which combines elements of both supervised and unsupervised learning

### Supervised Machine Learning
Supervised learning is defined as when a model gets trained on a "Labeled Dataset". Labelled datasets have both input and output parameters. In Supervised Learning algorithms learn to map points between inputs and correct outputs. It has both training and validation datasets labelled.

Example: If you train a model using labeled images of cats and dogs, it learns the features of each. When shown a new image, it predicts whether it’s a cat or a dog.

### Unsupervised Machine Learning
Unsupervised Learning works with unlabeled data, meaning there are no predefined outputs. The algorithm finds hidden patterns, groups or relationships within the data on its own. It’s mainly used for clustering, dimensionality reduction and data visualization.

Example: If you have customer data without labels, the algorithm can group similar customers based on purchase behavior useful for segmentation and marketing.


## Reinforcement Learning
Reinforcement learning trains an agent to make a sequence of decisions through trial and error. The agent interacts with the environment, receives feedback in the form of rewards or penalties and learns optimal actions over time.

Example: An AI agent learning to play chess gets positive feedback for good moves and negative for poor ones. Over time, it learns strategies to win more often.

## Semi-Supervised Learning
Semi-Supervised Learning is a type of machine learning where the model is trained on a small amount of labeled data and a large amount of unlabeled data.

## Classifaction vs Regression 

Classification predicts categories or classes, Regression predicts continuous numbers.

Classification output is discrete (e.g., spam/not spam), Regression output is continuous (e.g., house price).

Common algorithms: Classification → Logistic Regression, SVM; Regression → Linear Regression, Random Forest Regressor.

## Overfitting
Overfitting occurs when a machine learning model learns the training data too well, including noise or random fluctuations, which makes it perform poorly on new, unseen data.

## Underfitting

Underfitting happens when a machine learning model is too simple to capture the underlying patterns in the data, resulting in poor performance on both training and test data.

## Overfitting Vs Underfitting 

Overfitting happens when a model learns the training data too well, including noise, so it performs poorly on new data.

Underfitting occurs when a model is too simple to capture patterns, performing poorly on both training and test data.

Overfitting means the model is too complex, while underfitting means the model is too simple.

To fix overfitting, use regularization, more data, or simpler models; to fix underfitting, use a more complex model, add features, or reduce regularization.

## Bis

is the error caused by a model being too simple to capture the underlying patterns in the data.

## Variance

is the error caused by a model being too sensitive to small changes or noise in the training data.

## Bias-Variance tradeoff

Bias-Variance tradeoff is about balancing errors due to oversimplification (bias) and errors due to sensitivity to data (variance).

High bias causes underfitting, where the model is too simple to learn patterns.

High variance causes overfitting, where the model learns noise in the training data.

The goal is to find a model that generalizes well, minimizing both bias and variance.

## Feature Engineering: 

The process of creating, transforming, or selecting features from raw data to improve the predictive performance of machine learning models. Good features can make a simple model perform well.

Scaling:

* Purpose: Ensures all features are on a similar scale so models (like Gradient Descent-based models or distance-based models) converge faster and perform better.
* StandardScaler: Transforms features to have mean = 0 and standard deviation = 1.
* MinMaxScaler: Scales features to a fixed range [0,1], preserving the relationships between values.

Encoding:

* Purpose: Converts categorical data into numerical values so models can use them.
* One-Hot Encoding: Creates binary columns for each category (e.g., “Red, Blue, Green” → three columns).
* Label Encoding: Assigns integer values to categories (e.g., “Red=0, Blue=1, Green=2”), simpler but can imply ordinal relationships.

Feature Selection:

* Purpose: Selects only the most relevant features to reduce overfitting, improve performance, and speed up training.
* Correlation: Removes features that are highly correlated with others to avoid redundancy.
* Mutual Information: Measures dependency between feature and target, keeping the most informative ones.
* Lasso (L1 Regularization): Penalizes less important features, often shrinking some coefficients to zero, effectively selecting features.


## Cross-Validation 

is a technique to evaluate model performance by splitting data into multiple train-test sets.


# Introduction to Linear Regression

Linear regression is a type of **supervised machine learning algorithm** used for predicting continuous outcomes. It learns from **labelled datasets** and finds the most optimized **linear function** that maps input features to an output. It assumes a **linear relationship** between input (independent variable) and output (dependent variable), meaning the output changes at a constant rate with respect to the input.

This relationship is visually represented by a **straight line**.


## Example: Predicting Exam Scores

Suppose we want to predict a student's exam score based on how many hours they studied. Observations show that as study hours increase, exam scores also tend to increase.

## Decision Trees

- **Definition:** A supervised machine learning algorithm that splits data into branches based on feature values to make predictions for classification or regression tasks.

- **How it works:** The model recursively divides the dataset using feature conditions that reduce impurity (such as Gini index, entropy, or variance) until it reaches a decision.

- **Advantages:** Easy to understand and interpret, handles both numerical and categorical data, and requires little data preprocessing.

- **Disadvantages:** Prone to overfitting, especially with deep trees, and can be unstable with small changes in the data.

- **Use Cases:** Credit risk prediction, medical diagnosis, customer segmentation, and decision-making systems.



## Random Forest

- **Definition:** A supervised ensemble learning algorithm that builds **multiple decision trees** and combines their predictions to improve accuracy and reduce overfitting.

- **How it works:** It uses **bagging (bootstrap sampling)** to train many decision trees on different subsets of data and then averages their predictions for regression or uses majority voting for classification.

- **Advantages:** Reduces overfitting compared to a single decision tree, handles large datasets well, and can measure **feature importance**.

- **Disadvantages:** Less interpretable than a single decision tree and can require more computational resources.

- **Use Cases:** Fraud detection, recommendation systems, customer churn prediction, and general classification or regression problems.

## Gradient Boosting

- **Definition:** An ensemble machine learning method that builds multiple decision trees **sequentially**, where each new tree tries to correct the errors made by the previous trees.

- **How it works:** The model trains trees one after another using **boosting**, minimizing a loss function by focusing more on the data points that were previously predicted incorrectly.

- **Advantages:** High predictive accuracy, handles complex patterns well, and works effectively with structured/tabular data.

- **Disadvantages:** More computationally expensive, slower to train, and requires careful tuning to avoid overfitting.

- **Use Cases:** Fraud detection, customer churn prediction, ranking systems, and high-performance predictive modeling tasks (commonly implemented with **XGBoost** and **LightGBM**).

## Decision Tree vs Random Forest vs Gradient Boosting

- **Model Type:** Decision Tree is a **single tree model**, Random Forest is an **ensemble of many trees using bagging**, while Gradient Boosting is an **ensemble method where trees are built sequentially using boosting**.

- **Training Process:** Decision Tree builds **one tree by splitting data**, Random Forest builds **multiple trees in parallel using random samples**, while Gradient Boosting builds **trees sequentially where each tree corrects the errors of the previous one**.

- **Overfitting:** Decision Trees are **prone to overfitting**, Random Forest **reduces overfitting by averaging multiple trees**, while Gradient Boosting **controls errors through sequential learning and regularization**.

- **Performance:** Decision Trees are **simple and interpretable**, Random Forest provides **better accuracy and stability**, while Gradient Boosting (XGBoost, LightGBM) often provides **the highest performance but is more complex and computationally intensive**.

- **Use Cases:** Decision Trees are used for **simple decision rules**, Random Forest for **general classification and regression tasks**, and Gradient Boosting for **high-performance predictive models like fraud detection, ranking, and churn prediction**.

# 🖥️ Support Vector Machine (SVM)

**Support Vector Machine (SVM)** is a **supervised machine learning algorithm** used for **classification** and **regression tasks**.  

It tries to find the **best boundary**, known as a **hyperplane**, that separates different classes in the data.  

It is especially useful for **binary classification**, such as:

- Spam vs. Not Spam  
- Cat vs. Dog  

---

## 🔹 Goal of SVM
The main goal of SVM is to **maximize the margin** between the two classes.  

- **Margin:** Distance between the hyperplane and the nearest data points from each class.  
- A **larger margin** generally leads to **better performance** on new and unseen data.

## K-Means Clustering

- **Definition:** K-Means is an **unsupervised learning algorithm** that groups similar data points into **k clusters** based on their feature similarity.

- **How it works:** The algorithm initializes **k centroids**, assigns each data point to the nearest centroid, and then updates the centroids repeatedly until the clusters stabilize.

- **Advantages:** Simple, fast, and efficient for large datasets, and works well when clusters are clearly separated.

- **Disadvantages:** Requires choosing the value of **k**, sensitive to initial centroid selection, and does not work well with irregular-shaped clusters.

- **Use Cases:** Customer segmentation, document clustering, image compression, and pattern recognition.

### Hierarchical Clustering

- **Definition:** Hierarchical clustering is an **unsupervised learning algorithm** that builds a hierarchy of clusters and represents them in a **tree-like structure called a dendrogram**.

- **How it works:** It groups similar data points step by step using two approaches: **Agglomerative (bottom-up)** where clusters are merged, and **Divisive (top-down)** where clusters are split.

- **Advantages:** Does not require specifying the number of clusters initially and provides a clear visualization of cluster relationships.

- **Disadvantages:** Computationally expensive for large datasets and sensitive to noise and outliers.

- **Use Cases:** Gene sequence analysis, document clustering, and customer segmentation.

### DBSCAN (Density-Based Spatial Clustering)

- **Definition:** DBSCAN is an **unsupervised density-based clustering algorithm** that groups data points that are closely packed together and marks low-density points as noise or outliers.

- **How it works:** It forms clusters based on two parameters: **eps (radius distance)** and **minPts (minimum number of points required to form a dense region)**.

- **Advantages:** Can detect **arbitrary-shaped clusters** and automatically identifies **noise/outliers** without specifying the number of clusters.

- **Disadvantages:** Sensitive to parameter selection and may struggle with datasets that have **varying densities**.

- **Use Cases:** Anomaly detection, spatial data analysis, geographic clustering, and fraud detection.

## Clustering Evaluation Methods

- **Silhouette Score:** Measures how similar a data point is to its own cluster compared to other clusters. Higher value means better clustering.
- **Elbow Method:** Used in K-Means to determine the optimal number of clusters by observing where the improvement in variance reduction slows down.
- **Davies-Bouldin Index:** Measures cluster similarity; **lower values indicate better clustering**.
- **Calinski-Harabasz Score:** Evaluates cluster separation; **higher score indicates better-defined clusters**.

### PCA (Principal Component Analysis)

- **Definition:** PCA is a dimensionality reduction technique that transforms high-dimensional data into a smaller set of new features called **principal components**, while preserving most of the data variance.

- **How it works:** It finds directions (principal components) where the data varies the most and projects the original data onto these directions.

- **Advantages:** Reduces the number of features, speeds up model training, and helps remove multicollinearity between features.

- **Disadvantages:** Principal components are harder to interpret because they are combinations of original features.

- **Use Cases:** Data visualization, noise reduction, feature compression, and preprocessing for machine learning models.

# Machine Learning Interview Answers

## 1. What is Regularization?

**Definition:**  
Regularization is a technique used in machine learning to **prevent overfitting by adding a penalty to the model’s loss function**, which discourages overly complex models with large weights.

**How it works:**
- It modifies the loss function by adding a **regularization term**.
- This forces the model to keep weights smaller and simpler.

**Common Types:**
- **L1 Regularization (Lasso):** Adds absolute value of weights, can reduce some weights to zero (feature selection).
- **L2 Regularization (Ridge):** Adds squared value of weights, reduces weight magnitude but rarely makes them zero.
