# LLM Performance & Optimization

## Latency vs Throughput


Latency: The time it takes for an LLM to generate a response from the moment a prompt is received.

Throughput: The number of requests or tokens the model can process per unit of time.

How It Works

- Latency is critical when users expect real-time responses, such as in chatbots or live assistants.

- Throughput is important when the system needs to handle many requests simultaneously, like batch processing or large-scale API usage.

Trade-offs:

- Optimizing for low latency may reduce throughput (e.g., running a single prompt on a large model for speed).

- Optimizing for high throughput may increase latency (e.g., batching multiple prompts for efficiency).

Importance

- Balancing latency and throughput ensures responsive and scalable AI systems.

- Critical for interactive applications (low latency) vs bulk processing tasks (high throughput).

Pros / Considerations

- ✅ Low latency → better user experience, especially for interactive apps.

- ✅ High throughput → better resource utilization and cost efficiency for large workloads.

- ✅ Systems often need tuning and parallelization to optimize both metrics.

Cons / Challenges

- ⚠️ Reducing latency may increase computational cost.

- ⚠️ Maximizing throughput may delay individual responses.

- ⚠️ Hardware, model size, and prompt complexity affect both metrics.

## Batching


Batching is the process of grouping multiple inputs (prompts) together and processing them simultaneously through an LLM. This improves computational efficiency and throughput while making better use of hardware resources.

How It Works

- Instead of sending each prompt individually, multiple prompts are combined into a single batch.

- The model processes all prompts in parallel, leveraging matrix operations on GPUs or TPUs.

- Once processed, outputs for each prompt are returned to the respective users.

Batching is especially useful in high-throughput systems and during training or fine-tuning, where many examples need to be processed at once.

Importance

Reduces per-prompt overhead and inference cost.

Increases throughput by utilizing parallel computation effectively.

Helps balance latency vs. efficiency, especially for large models.

Pros

- ✅ Improves hardware utilization (GPUs/TPUs).

- ✅ Increases throughput, allowing more requests per second.

- ✅ Reduces cost per request in large-scale deployments.

Cons / Challenges

- ⚠️ Larger batches can increase latency for individual requests.

- ⚠️ Memory constraints may limit batch size for very large models.

- ⚠️ Uneven input lengths may require padding, leading to wasted computation.

Analogy

Batching is like preparing multiple meals at the same time in a restaurant:

- You cook several orders together → more efficient and faster overall.

- Individual orders may wait slightly longer, but the overall kitchen throughput improves.

## KV Caching (Key-Value Caching)


KV caching is an inference optimization technique where the Key (K) and Value (V) tensors from previous tokens are stored (cached) so the model does not recompute attention for earlier tokens when generating new ones.

It significantly reduces latency during autoregressive generation.

Why It’s Needed

LLMs generate text token by token.
Without caching:

- For every new token, the model recomputes attention over all previous tokens.

- This makes inference slower as the sequence grows.

With KV caching:

The model stores past K and V matrices.

- For a new token, it only computes attention against the cached values.

- No need to recompute previous states.

This changes inference complexity from repeatedly recomputing the full sequence to incremental computation.

How It Works (Conceptually)

In the Transformer attention mechanism:

Attention = Softmax(QKᵀ / √d) V

During generation:

- First token → compute Q, K, V → store K and V in cache.

- Second token → compute new Q, K, V → append new K and V to cache.

- Third token → same process.

The cache grows with each generated token.

Importance

- Critical for fast real-time chat applications.

- Reduces redundant computation.

- Improves latency significantly for long responses.

Without KV caching, long conversations would become progressively slower.

Pros

- ✅ Much faster autoregressive decoding

- ✅ Essential for streaming responses

- ✅ Improves user experience in chat systems

Cons / Challenges

- ⚠️ Increases memory usage (cache grows with sequence length)

- ⚠️ Large context windows require large KV memory

- ⚠️ Not helpful during full-sequence training (mostly inference optimization)

Analogy

KV caching is like taking notes during a lecture:

- Without notes → you reread the entire book every time.
 
- With notes → you only add new information and refer to stored notes.

You avoid repeating work.

## Quantization

Quantization is a technique where model weights (and sometimes activations) are stored using lower precision numbers (e.g., 8-bit or 4-bit instead of 16-bit or 32-bit floating point) to reduce memory usage and computation cost.

Why It’s Needed

Large Language Models are huge:

- Billions of parameters

- Each parameter stored as FP32 (4 bytes) or FP16 (2 bytes)

For example:

- 7B parameters in FP16 → ~14 GB memory

- 7B parameters in 8-bit → ~7 GB

- 7B parameters in 4-bit → ~3.5 GB

Quantization makes it possible to:

- Run models on smaller GPUs

- Deploy on consumer hardware

- Reduce inference cost

How It Works (Conceptually)

Instead of storing weights like:

0.123456789 (32-bit float)

We approximate them with fewer bits:

- FP32 → FP16 → INT8 → INT4

The values are scaled and rounded into a smaller numerical range.

Lower precision =

- Less memory

- Faster matrix multiplication

- Slight accuracy trade-off

Pros

- ✅ Massive memory reduction

- ✅ Faster inference

- ✅ Lower deployment cost

- ✅ Enables edge-device deployment

Cons / Trade-offs

- ⚠️ Slight drop in model accuracy

- ⚠️ Numerical instability in extreme low precision

- ⚠️ More engineering complexity
