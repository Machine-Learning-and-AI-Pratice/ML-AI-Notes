## Transformer Architecture (End-to-End)

A Transformer is a neural network architecture designed to handle sequential data efficiently using attention mechanisms, without relying on recurrence (RNNs) or convolutions (CNNs). It consists of an Encoder-Decoder structure (for sequence-to-sequence tasks) or can be encoder-only or decoder-only depending on the application.

## Workflow 
When a sentence enters a Transformer, the first step is tokenization. The text is split into smaller units, called tokens, which could be words, subwords, or even characters. Each token is then transformed into a dense vector representation, known as an embedding, which captures its semantic meaning. Since Transformers do not process sequences sequentially like RNNs, positional encoding is added to these embeddings to retain information about the order of tokens in the sentence.

These embeddings are then passed into the encoder. The encoder is made of multiple stacked layers, each designed to understand the relationships between tokens. Within each layer, self-attention allows every token to “look at” every other token in the sequence, helping the model capture global dependencies. The output of the self-attention is then passed through a feed-forward network to enhance the representation, and residual connections with layer normalization ensure the flow of information remains stable as it moves through the layers. By stacking multiple encoder layers, the model builds deep, contextual representations of the input sequence.

Next, the decoder takes over, generating the output sequence. Each decoder layer starts with masked self-attention, which ensures that each token being generated only looks at previously generated tokens, maintaining the causal property needed for autoregressive generation. The decoder then performs encoder-decoder attention, allowing it to focus on the most relevant parts of the input sequence from the encoder. A feed-forward network further processes these representations, and again residual connections and layer normalization help maintain stability across the layers. Multiple decoder layers are stacked to improve the model’s capacity for generating coherent sequences.

Once the decoder processes the sequence, it produces a vector representation for each position in the output. These vectors pass through a linear layer followed by a softmax, converting them into a probability distribution over the vocabulary. The model selects the most probable token at each step, and this token is fed back into the decoder to predict the next token, continuing until the sequence is complete.

The attention mechanism underpins this entire process. Self-attention captures dependencies within the sequence, while multi-head attention allows the model to focus on different aspects of the sequence simultaneously. The use of scaled dot-product attention ensures this computation is efficient, allowing the model to weigh relationships between tokens accurately and quickly.


1️⃣ Input Representation

- Tokenization: Text is split into tokens (words, subwords, or characters).

- Embedding: Tokens are converted into dense vectors (embeddings).

- Positional Encoding: Since Transformers have no sequential recurrence, positional information is added to embeddings to retain word order.

2️⃣ Encoder

- The encoder processes the input sequence and creates contextual representations.

- Each encoder layer has two main sub-layers:

   - Self-Attention: Each token attends to all other tokens to capture global dependencies.

   - Feed-Forward Network: Applies a non-linear transformation to enhance representation.

   - Residual Connections & Layer Normalization: Help stabilize training and preserve information across layers.

- Multiple encoder layers are stacked to build deep contextual understanding.

3️⃣ Decoder

- The decoder generates output sequences from the encoder’s context (for sequence-to-sequence tasks).

- Each decoder layer has three sub-layers:

  - Masked Self-Attention: Ensures the model only attends to previous tokens when generating text (causal modeling).

  - Encoder-Decoder Attention: Allows the decoder to focus on relevant parts of the input.

  - Feed-Forward Network: Further processes the representations.
  
  - Residual connections & layer normalization are also applied in the decoder.

- Multiple decoder layers are stacked for deeper generation capability.

4️⃣ Output Generation

- The decoder outputs a vector for each position in the sequence.

- Linear + Softmax Layer: Converts the vectors into a probability distribution over the vocabulary.

- Autoregressive Prediction: The next token is predicted based on previous outputs (for causal models).

5️⃣ Attention Mechanism

- Self-Attention: Captures dependencies between tokens in a sequence.

- Multi-Head Attention: Uses multiple attention heads to focus on different aspects of the sequence simultaneously.

- Scaled Dot-Product Attention: Efficiently computes attention scores using queries, keys, and values.

6️⃣ Advantages

- ⚡ Parallelizable → faster training than RNNs.

- 🌐 Captures long-range dependencies easily.

- 🧠 Provides context-aware representations for every token.

- 🔄 Flexible → works for encoder-only, decoder-only, or encoder-decoder architectures.

7️⃣ Common Applications

- Encoder-only: BERT → understanding tasks (classification, Q&A).

- Decoder-only: GPT → generation tasks (text, code).

- Encoder-Decoder: T5, MarianMT → translation, summarization.

## What is self-attention?

Self-attention is a mechanism in Transformers that allows each word (or token) in a sequence to focus on other words in the same sequence to understand context. It answers the question:

“Which words in this sentence are important to understanding this word?”

## How It Works

- Every token is represented as a vector.

- For each token, the model computes three vectors:

  - Query (Q) – what this token is looking for

  - Key (K) – what this token contains

  - Value (V) – the actual information of the token

- The attention score is calculated by comparing the query of one token with the keys of all tokens.

- These scores are normalized using softmax to produce weights.

- Each token’s final representation is the weighted sum of all token values, meaning it now contains information about the words it “attended to.”

## Why It’s Important

🔗 Captures long-range dependencies in a sequence (even far-apart words).

🧠 Creates context-aware representations for every token.

⚡ Enables parallel processing, unlike RNNs which read sequentially.

## Query, Key, Value (QKV) in Self-Attention

In self-attention, each token in a sequence is represented by three vectors: Query (Q), Key (K), and Value (V). These vectors determine how much attention each word should give to other words.

1️⃣ Query (Q)

- Represents what the token is looking for in the sequence.

- Think of it as the token asking:

“Which other words are relevant to me?”

2️⃣ Key (K)

- Represents the identity or content of each token.

- Think of it as a “tag” for every token that can be matched against queries.

3️⃣ Value (V)

- Represents the actual information the token carries.

- This is what gets passed along if the token is attended to.

4️⃣ How Q, K, V Work Together

- Compute attention scores by taking the dot product of Q (query) and K (key) for each token pair.

- Normalize the scores with softmax to get weights.

- Multiply these weights by the V (value) vectors.

- Sum the results → the token now contains a context-aware representation.

## Scaled Dot-Product Attention

Scaled dot-product attention is the core computation of self-attention. It determines how much attention each token should give to other tokens in a sequence by comparing queries (Q) and keys (K) and applying it to the values (V).

1️⃣ How It Works

- Dot Product: Compute the dot product between the Query (Q) of a token and the Key (K) of every token in the sequence.

  - This measures similarity/importance between tokens.

- Scaling: Divide the dot product by √(dimension of K).

  - This prevents large values in the dot product that can make softmax outputs unstable.

- Softmax: Apply softmax to the scaled scores to get attention weights.

  - Weights determine how much each token should contribute to the current token’s new representation.

- Weighted Sum: Multiply the weights by the Value (V) vectors and sum them.

  - Produces a context-aware representation for the token.

2️⃣ Formula

The formula for scaled dot-product attention is:

$$\text{Attention}(Q, K, V) = \text{Softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right) V$$
Where:  
- **Q** = Query matrix  
- **K** = Key matrix  
- **V** = Value matrix  
- **d_k** = Dimension of the key vectors

This is the core computation of **self-attention** in Transformers.

3️⃣ Why Scaling is Important

- Without scaling, large dot-product values can cause softmax gradients to be very small, slowing down training.

- Dividing by √(dimension of K) keeps values in a stable range.


Scaled dot-product attention decides how much weight to give each participant when forming your own understanding

## Multi-Head Attention

Multi-head attention extends self-attention by allowing the model to focus on different aspects of the sequence simultaneously. Instead of using a single attention computation, it uses multiple “heads” in parallel.

1️⃣ How It Works

Linear Projections: The input embeddings are linearly projected into multiple sets of queries (Q), keys (K), and values (V) — one set per head.

Parallel Attention: Each head performs scaled dot-product attention independently.

Concatenation: The outputs of all attention heads are concatenated.

Final Linear Layer: The concatenated output is passed through a linear layer to produce the final multi-head attention representation.

2️⃣ Why It’s Important

🧠 Focus on Multiple Aspects: Different heads can capture different types of relationships between tokens, e.g., syntax, semantics, long-term dependencies.

⚡ Better Representations: Combining multiple heads provides richer, context-aware token embeddings.

🔄 Parallelizable: All heads are computed in parallel, making it efficient.

4️⃣ Formula (Simplified)

$$\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, ..., \text{head}_h) W^O$$
$$\text{where } \text{head}_i = \text{Attention}(QW_i^Q, KW_i^K, VW_i^V)$$
