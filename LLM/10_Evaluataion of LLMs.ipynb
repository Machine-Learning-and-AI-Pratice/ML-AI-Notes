# Evaluation of LLMs

## How Do You Evaluate LLMs?

Evaluating LLMs means measuring how well a model performs in terms of:

- Accuracy

- Reasoning ability

- Safety

- Robustness

- Efficiency

There is no single metric — evaluation is multi-dimensional.

## Automatic vs Human Evaluation

Automatic evaluation uses predefined metrics to measure model performance quickly and objectively. Human evaluation relies on people to judge the quality, relevance, and usefulness of model outputs.

Automatic evaluation is fast, scalable, and inexpensive. Human evaluation is slower, costly, and not easily scalable.

Automatic metrics include BLEU, ROUGE, Perplexity, and Accuracy. Human evaluation considers fluency, coherence, factual correctness, helpfulness, and safety.

Automatic evaluation is consistent and reproducible but may not reflect real-world quality. Human evaluation captures nuance, context, and subtle errors that metrics often miss.

Automatic evaluation is ideal during model training and benchmarking. Human evaluation is essential for final validation, alignment, and user experience assessment.

## Perplexity


Perplexity is a metric used to evaluate how well a language model predicts a sequence of tokens. It measures how “confused” the model is when predicting the next token. Lower perplexity means better performance.

1️⃣ How It Works

- A model assigns probabilities to each token in a sequence.

- Perplexity is derived from cross-entropy loss.

- If the model assigns high probability to correct tokens → low perplexity.

- If probabilities are low/confused → high perplexity.

2️⃣ Interpretation

- Low Perplexity (good): Model predicts text confidently and accurately.

- High Perplexity (bad): Model struggles to predict the next token.

- Perplexity of 1 = perfect prediction (rare in practice).

3️⃣ Why It’s Important

- Standard metric for evaluating language modeling performance.

- Helps compare different models or training checkpoints.

- Used during pretraining and validation.

4️⃣ Limitations

- Does not directly measure coherence, factuality, or reasoning ability.

- Lower perplexity does not always mean better real-world performance.

- Sensitive to vocabulary size and tokenization.

5️⃣ Analogy

Perplexity is like measuring how surprised you are when reading the next word in a sentence:

- If the next word is obvious → low surprise → low perplexity.

- If the word is unexpected → high surprise → high perplexity.

## BLEU vs ROUGE (Limitations Included)


BLEU and ROUGE are evaluation metrics used to measure the quality of generated text by comparing it with a reference (ground truth) text.

BLEU (Bilingual Evaluation Understudy): Primarily used for machine translation.

ROUGE (Recall-Oriented Understudy for Gisting Evaluation): Commonly used for text summarization.

1️⃣ How They Work

- BLEU: Measures precision of n-grams (how many generated phrases match the reference).

- ROUGE: Measures recall of n-grams (how much of the reference is captured in the generated text).

Both rely on overlap between generated text and reference text.

2️⃣ Strengths

- ✅ Simple and fast to compute.

- ✅ Useful for benchmarking and comparing models.

- ✅ Standard metrics in translation and summarization research.

3️⃣ Limitations

- ⚠️ Focus on surface-level n-gram overlap, not meaning.

- ⚠️ Cannot capture paraphrasing or semantic similarity well.

- ⚠️ Sensitive to wording differences even if meaning is correct.

- ⚠️ Poor at evaluating creativity, reasoning, or factual correctness.

- ⚠️ Require reference texts — difficult for open-ended generation tasks.

4️⃣ Practical Issues

- A perfectly valid paraphrase can get a low BLEU/ROUGE score.

- Models can optimize for higher scores without improving true quality.

- Not suitable alone for evaluating LLMs in dialogue or reasoning tasks.

5️⃣ Analogy

BLEU and ROUGE are like grading essays by counting matching phrases:

- If many phrases match → high score.

- But even a well-written, meaningful answer with different wording might score low.

## Faithfulness 

Faithfulness measures whether a model’s output is factually consistent with the source input or reference information.
It checks if the model adds, changes, or invents facts (hallucination).

1️⃣ What It Means

Output should stay grounded in the given input.

- No fabricated facts.

- No contradictions.

- No unsupported claims.

2️⃣ Where It’s Important

- 📄 Summarization – Summary must reflect original document.

- 🔍 Question Answering – Answer must come from provided context.

- 🧠 RAG systems – Generated text must align with retrieved documents.

- ⚖️ Medical / Legal AI – Accuracy is critical.

3️⃣ Faithfulness vs Other Metrics

-Perplexity → Measures prediction confidence.

-BLEU/ROUGE → Measure word overlap.

- Faithfulness → Measures factual correctness relative to source.

A response can have:

- High ROUGE score ❌ but still be unfaithful.

- Low perplexity ❌ but contain hallucinations.

4️⃣ Common Problems

- 🚨 Hallucination (making up facts)

- 🚨 Subtle distortion of meaning

- 🚨 Overgeneralization beyond source

- 🚨 Confident but incorrect answers

5️⃣ Example

Source:
“The company reported a 10% revenue increase in 2023.”

Faithful summary:
“The company’s revenue grew by 10% in 2023.”

Unfaithful summary:
“The company doubled its profits in 2023.” ❌ (Not in source)

6️⃣ Why It’s Hard

- LLMs are trained for fluency, not strict factual grounding.

- They optimize next-token probability, not truth.

- Long context increases risk of subtle drift.

## Coherence

Coherence measures whether a model’s output is logically connected, well-structured, and consistent throughout the text.
It checks if ideas flow naturally from one sentence to the next.

1️⃣ What It Means

- Sentences follow a logical order.

- No contradictions within the response.

- Clear progression of ideas.

- Smooth transitions between points.

2️⃣ Where It Matters

- 📝 Essay writing

- 🤖 Chatbots & dialogue systems

- 📚 Story generation

- 📊 Long-form explanations

Long responses especially require strong coherence.

3️⃣ Coherence vs Other Metrics

- Fluency → Grammatically correct language.

- Faithfulness → Factually consistent with source.

- Coherence → Logical consistency and flow.

A response can be:

- Fluent but incoherent ❌

- Faithful but poorly structured ❌

Coherent and fluent ✅

4️⃣ Common Coherence Issues

- 🔁 Repeating the same idea unnecessarily

- 🔀 Jumping between unrelated topics

- ❌ Contradicting earlier statements

- 🧱 Poor paragraph transitions

5️⃣ Example

- Coherent:
“Transformers use self-attention. This allows them to capture long-range dependencies. As a result, they perform well in NLP tasks.”

- Incoherent:
“Transformers use self-attention. Dogs are common pets. Long-range dependencies are important.”

6️⃣ Why It’s Challenging

- Long context windows increase risk of drift.

- Multi-turn conversations can cause topic shifts.

- Models optimize probability, not logical structure.

## Factual Accuracy

Factual accuracy measures whether a model’s output is objectively correct and aligned with real-world facts.

It checks if the information generated is true, verifiable, and not fabricated.

## Task-Specific Benchmarks

Task-specific benchmarks are standardized datasets designed to evaluate how well a model performs on a particular NLP task (like sentiment analysis, translation, or question answering).

Instead of testing “general intelligence,” they measure performance on specific capabilities.
