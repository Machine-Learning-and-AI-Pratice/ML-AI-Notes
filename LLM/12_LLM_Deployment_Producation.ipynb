## LLM Serving Architecture
In production, LLMs are deployed behind APIs with a serving layer (like FastAPI or gRPC), a model inference engine (e.g., vLLM / TensorRT), GPU workers, and optional components like RAG pipelines, caching, and monitoring. The architecture must handle batching, scaling, and low-latency responses.

## GPU vs CPU Inference
GPUs are preferred for LLM inference due to massive parallel matrix computation, making them much faster for large models. CPUs are cheaper and suitable for small models or low-throughput systems but struggle with large-scale transformer inference.

## Model Sharding
Large models that cannot fit into a single GPU are split across multiple GPUs (tensor parallelism or pipeline parallelism). Sharding enables serving very large LLMs but increases communication overhead between devices.

## Horizontal vs Vertical Scaling
Vertical scaling increases resources on a single machine (more powerful GPU).
Horizontal scaling adds more machines or GPU nodes behind a load balancer.
LLM systems typically prefer horizontal scaling for high availability and traffic handling.

## Cold Start Problem
LLMs have long startup times due to model loading into GPU memory. Cold starts increase latency in serverless or autoscaling setups. Solutions include warm pools, preloading models, or keeping minimum replicas running.

## Rate Limiting
Protects the system from overload by restricting request volume per user or API key. Prevents abuse and ensures stable throughput under high demand.

## Observability & Logging
Production LLM systems require monitoring for:

- Latency

- Throughput

- Token usage

- GPU utilization

- Error rates

- Hallucination patterns

Logging prompts and responses (with privacy controls) is critical for debugging and continuous improvement.
