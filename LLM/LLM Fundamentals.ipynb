# 1️⃣ LLM Fundamentals 

## What is an LLM?

A Large Language Model (LLM) is an advanced AI system designed to understand, generate, and manipulate human language by learning patterns from massive amounts of text data.

## How is an LLM different from traditional ML models?




- LLMs are large, flexible, and general-purpose, designed for language understanding and generation.

- Traditional ML models are smaller, task-specific, and rely heavily on feature engineering.

- LLMs learn patterns automatically from huge data; traditional models need explicit design for each task.

- LLMs produce human-like text or structured responses.Traditional ML models produce labels, probabilities, or numeric predictions.

- LLMs require high computational power for training and inference. Traditional ML models are computationally lighter and faster to train.

## Transformer vs RNN vs CNN

Transformers are highly parallelizable, flexible, and excel at long-range dependencies in sequences. RNNs process sequentially and are better for short-term sequence data. CNNs focus on spatial patterns and are ideal for images and videos.

Transformers capture global context across inputs. RNNs have limited context awareness due to vanishing gradients. CNNs capture local context but struggle with long-range dependencies.

Transformers are widely used in NLP tasks like translation and text generation. RNNs are commonly used for time-series and speech recognition. CNNs are primarily used for image classification, object detection, and segmentation.

Transformers require massive data and high computational power. RNNs require moderate computation but are slower due to sequential processing. CNNs are computationally efficient for image tasks but not suitable for text sequences.

Transformers can handle zero-shot and few-shot learning. RNNs and CNNs usually need task-specific retraining or careful feature design.

## Why transformers work well for language?

- Parallel Processing of Words 🟢

  - Unlike RNNs, Transformers process all words in a sentence at once, making them much faster to train and able to handle long sentences efficiently.

- Attention Mechanism 🎯

  - Transformers use self-attention to weigh the importance of each word relative to others in a sequence.

  - This allows them to capture context and relationships, even between words far apart in a sentence.

- Long-Range Dependency Handling 🌐

  - They excel at understanding connections across long sequences of text, something RNNs struggle with due to vanishing gradients.

- Context Awareness 🧠

  - Every word representation considers the whole sentence or paragraph, giving rich contextual understanding for tasks like translation, summarization, or Q&A.

- Scalability ⚡

  - Transformers can be scaled up with more layers and parameters, learning complex patterns from massive datasets.

  - This is why models like GPT-4 and LLaMA achieve human-like text generation.

- Versatility 🛠️

  - They are task-agnostic, meaning the same architecture works for translation, summarization, code generation, and more.

## What is Self-Attention?

Self-attention is a mechanism used in Transformers that lets a model focus on different parts of the input sequence when encoding a word. It determines which words are important relative to each other.

1️⃣ How It Works

For each word in a sentence, the model computes attention scores with every other word.

These scores tell the model how much each word should “pay attention” to other words.

The word’s representation is then updated as a weighted sum of all words in the sequence.

2️⃣ Why It’s Important

🔗 Captures long-range dependencies between words (even far apart).

🧠 Provides context-aware representations, improving understanding of meaning.

⚡ Allows parallel processing, unlike RNNs which process sequentially.

## Encoder vs Decoder vs Encoder-Decoder

Encoders understand and compress input sequences into contextual embeddings. Decoders generate output sequences using the given context. Encoder-Decoder models combine both to perform end-to-end sequence-to-sequence tasks.

Encoders capture the meaning and relationships within input data. Decoders generate coherent outputs step by step based on context. Encoder-Decoder models maintain both understanding and generation for tasks like translation or summarization.

Encoders are used in tasks like text classification, sentiment analysis, and question-answering. Decoders are used for text generation, dialogue systems, and code completion. Encoder-Decoder models are widely used in machine translation, summarization, and sequence-to-sequence learning.

Encoders require large datasets for effective contextual understanding. Decoders need sufficient context and training for coherent output generation. Encoder-Decoder models require significant computation and data to perform well on complex tasks.

Encoder-only models are optimized for understanding and representation learning. Decoder-only models are optimized for generation and autoregressive tasks. Encoder-Decoder models excel at transforming input sequences into output sequences in an end-to-end manner.

## Autoregressive Models

Autoregressive models generate data step by step, where each output depends on previous outputs. They predict the next element in a sequence based on what has come before.

## Causal vs Bidirectional Models

Causal models predict the next token in a sequence using only past tokens. Bidirectional models use both past and future tokens to understand the full context.

Causal models are autoregressive and generate text step by step. Bidirectional models are designed for understanding and consider the entire sequence simultaneously.

Causal models are widely used in text generation, dialogue systems, and code completion. Bidirectional models are commonly used in sentiment analysis, question-answering, and named entity recognition.

Causal models require sequential processing and rely on previous context for predictions. Bidirectional models require access to the full sequence and use masked language modeling during training.

Causal models are exemplified by GPT series and other autoregressive models. Bidirectional models are exemplified by BERT, RoBERTa, and other understanding-focused models.
