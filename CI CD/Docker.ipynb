# Docker

## What is Docker

Docker is a platform that packages applications and their dependencies into containers.
These containers run consistently across different environments.

## Containers vs Virtual Machines

Containers share the host OS and run lightweight, fast applications in isolated environments.

Virtual Machines include a full OS for each instance, making them heavier and slower but more isolated.

## Docker Architecture

Docker architecture consists of the Docker Client, Docker Daemon, and Docker Registry.
The client sends commands to the daemon, which builds and runs containers using images from registries.

## Docker Installation & Setup 

Install Docker by downloading Docker Desktop or using your OS package manager.
After installation, start Docker and verify with docker --version and docker run hello-world.

# Docker CLI Basics

## Docker run 

docker run creates and starts a new container from an image.    
`
Example: docker run -p 3000:3000 my-app runs the app and maps port 3000.
`

## docker ps

docker ps lists all running containers.
Use docker ps -a to see all containers, including stopped ones.

## Docker Stop 
docker stop gracefully stops a running container.
```Example: docker stop <container_id> stops the specified container.```

## Docker pull 
docker pull downloads an image from a Docker registry (like Docker Hub).    
`Example: docker pull nginx fetches the latest Nginx image.`

## Docker push 

docker push uploads a Docker image to a registry like Docker Hub.
`Example: docker push username/app pushes your image to your repository.`

## Docker images

docker images lists all Docker images stored locally.
It shows details like repository, tag, image ID, and size.

## Interactive Containers (-it)

-it runs a container in interactive mode with a terminal attached.
`Example: docker run -it ubuntu bash lets you interact inside the container.`

## Port Mapping (-p)

-p maps a container port to a host port.
`Example: docker run -p 3000:3000 app exposes container port 3000 on your machine.`

## Container Logs

docker logs shows the output (stdout/stderr) of a container.
`Example: docker logs <container_id> lets you view application logs.`

# Docker Images & Dockerfile

## DockerFile

A Dockerfile is a script containing instructions to build a Docker image.
It defines the base image, dependencies, and commands to run your application.

## Writing Dockerfiles 

means defining step-by-step instructions to build a Docker image.
You use commands like FROM, COPY, RUN, and CMD to set up your application.

## Base Images (FROM) 

FROM specifies the base image to start building your Docker image.
`Example: FROM node:18 uses Node.js as the foundation.`

## Layers & Caching

Docker images are built in layers, where each instruction creates a new layer.
Caching reuses unchanged layers to speed up builds and improve efficiency.

## COPY vs ADD

COPY simply copies files from your system into the container.

ADD can also extract archives and download files from URLs, making it more powerful but less predictable.

## CMD vs ENTRYPOINT

CMD sets default commands that can be overridden at runtime.
ENTRYPOINT defines a fixed command that always runs when the container starts.

## docker build

docker build creates a Docker image from a Dockerfile.
`Example: docker build -t my-app . builds and tags the image.`

## Image Tagging

Image tagging assigns a name and version to a Docker image.
`Example: my-app:1.0 helps identify and manage different image versions.`



# Docker Volumes & Storage

## Volumes

Volumes are used to store data outside the container filesystem.
They persist data even if the container is removed.

## Bind Mounts

Bind mounts link a host machine directory directly to a container.
They allow real-time file sharing between host and container.

## Volume Management

Volume management involves creating, listing, and removing Docker volumes.
Commands like docker volume create, ls, and rm help manage persistent storage.

## Data Persistence

Data persistence ensures that container data is saved even after the container stops or is removed.
It is achieved using volumes or bind mounts.

## Shared Storage

Shared storage allows multiple containers to access the same data.
It is typically implemented using Docker volumes or bind mounts.



# Docker Networking

## Bridge Network

Bridge network is the default Docker network that allows containers to communicate on the same host.
Each container gets its own IP, and they can talk using internal networking.

## Host Network

Host network makes the container share the host’s network stack.
It removes network isolation, so the container uses the host’s IP and ports directly.

## Custom Networks

Custom networks are user-defined Docker networks for better control and isolation.
They allow containers to communicate using names and provide improved networking features.

## Container Communication

Container communication allows containers to interact with each other over a network.
They typically communicate using container names as hostnames within the same network.

## Service Exposure

Service exposure makes a container’s application accessible outside the container.
It is done by mapping container ports to host ports using -p.


# Docker Compose

## Docker Compose

Docker Compose is a tool to define and run multi-container applications.
It uses a docker-compose.yml file to configure and manage services together.

## docker-compose.yml

docker-compose.yml is a configuration file used to define services, networks, and volumes.
It describes how multiple containers should run together in an application.

## Services

Services define individual containers in Docker Compose.
Each service specifies the image, configuration, and how the container should run.

## Compose Networking

Compose networking automatically creates a network for services to communicate.
Containers can reach each other using service names as hostnames.

## Environment Variables

Environment variables are used to pass configuration values into containers.
They allow you to customize application behavior without changing code.


## Service Scaling

Service scaling increases or decreases the number of container instances for a service.
`Example: docker-compose up --scale app=3 runs multiple instances of the app.`




# Environment & Configuration

## .env Files

.env files store environment variables in key-value format.
They are used to manage configuration separately from code.

## Environment Variables

Environment variables are key-value pairs used to configure applications at runtime.
They help keep settings flexible without modifying the code.

## Secrets Management

Secrets management is used to securely store sensitive data like passwords and API keys.
It prevents exposing secrets in code or environment variables.


# Docker Registry & Distribution

## Docker Hub

Docker Hub is a cloud-based registry for storing and sharing Docker images.
You can push, pull, and manage images from repositories.

## Private Registries

Private registries are secure repositories for storing Docker images with restricted access.
They are used to keep proprietary or sensitive images private.

## Image Tagging Strategy

Image tagging strategy defines how you name and version Docker images.
It ensures consistency, traceability, and easier deployment management.


## Versioning

Versioning assigns specific versions to Docker images using tags.
It helps track changes and deploy the correct image in different environments.


# Optimization & Best Practices

## Multi-stage Builds

Multi-stage builds use multiple FROM statements in a Dockerfile.
They help create smaller, optimized images by separating build and runtime environments.


## Image Size Optimization

Image size optimization reduces Docker image size for faster builds and deployments.
Techniques include using lightweight base images and cleaning unnecessary files.

## Layer Caching

Layer caching reuses unchanged image layers during builds.
It speeds up Docker builds and reduces resource usage.

## Security Practices

Security practices in Docker involve running containers with least privileges, scanning images for vulnerabilities, and keeping software up to date.
They help protect the host system and containerized applications from attacks.

## Non-root Containers

Non-root containers run applications as a non-root user inside the container.
This improves security by limiting permissions and reducing potential damage if compromised.



# Docker in Production

## Logging

Logging captures container output and events for monitoring and debugging.
Docker provides docker logs and integration with logging drivers for persistent storage.

## Monitoring

Monitoring tracks the health, performance, and resource usage of containers.
It helps detect issues, optimize performance, and ensure applications run smoothly.

## Restart Policies

Restart policies define when Docker should automatically restart a container.
Examples: no, always, on-failure, or unless-stopped.

## Resource Limits

Resource limits control how much CPU and memory a container can use.
They prevent a container from consuming all host resources and affecting other containers.
