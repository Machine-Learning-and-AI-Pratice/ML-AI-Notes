# Agentic AI

Agentic AI is an AI system that can autonomously plan, make decisions, and take actions to achieve specific goals with minimal human intervention.
It focuses on goal-driven reasoning and can interact with tools and environments to complete complex tasks.


- Traditional AI works by taking a specific input and generating an output based on learned patterns, without going beyond the given prompt.      
- Agentic AI can break a goal into multiple steps, plan the approach, and execute tasks using reasoning and available tools.
- Traditional AI usually works in isolation for each request, while Agentic AI can maintain context and continue working across multiple steps or interactions.   
- Traditional AI depends fully on user prompts for every action, whereas Agentic AI can decide the next actions on its own once a goal is given.  
- Traditional AI focuses on prediction or classification tasks, but Agentic AI focuses on solving end-to-end problems and achieving outcomes. 

## Types of Agent AI 

### 1. Simple Reflex Agents

Simple reflex agents act only on the current input using fixed rules without memory or learning. They respond quickly but cannot handle complex or changing environments.
Example: Traffic light systems.

### 2. Model-Based Reflex Agents

These agents maintain an internal state (memory) of the environment to make better decisions. They handle partial information but still rely mostly on rules.
Example: Robot vacuum cleaners.

### 3. Goal-Based Agents

Goal-based agents choose actions by planning steps to achieve a specific goal. They are more flexible because they think ahead before acting.
Example: GPS navigation systems.

### 4. Utility-Based Agents

Utility-based agents select actions based on the “best outcome” by evaluating multiple factors like cost, risk, and benefit. They focus on optimization.
Example: Stock trading systems.

### 5. Learning Agents

Learning agents improve their performance over time by learning from experience and feedback. They adapt to new environments automatically.
Example: Chatbots that improve responses over time.

### 6. Multi-Agent Systems (MAS)

Multi-agent systems involve multiple AI agents working together or competing to solve problems. They coordinate or compete in shared environments.
Example: Robot fleets in warehouses.

### 7. Hierarchical Agents

Hierarchical agents work in layers where high-level decisions are broken into smaller tasks for lower levels. This makes complex systems easier to manage.
Example: Industrial automation systems.

## Agent2Agent (A2A) 

Agent-to-Agent (A2A) Protocol is an open communication protocol developed by Google that enables AI agents to communicate, collaborate, exchange tasks, and work together across different systems, frameworks, and organizations.

## workflow

The **Agent2Agent (A2A) Protocol** follows a client-server model where one agent sends a task request and another agent executes it. For example, a CustomerBot can request an OrderBot to check product availability and return the result. Agents can switch between client and server roles depending on the interaction.

Each agent exposes an **Agent Card**, a JSON profile containing its identity, capabilities, and security requirements, enabling agent discovery and selection. Work is managed as a **Task**, which progresses through states such as Submitted, Working, Input-Required, Completed, Failed, or Cancelled.

Agents exchange information using structured **Messages** (text, files, forms), and the final output is returned as an **Artifact**, a structured and reusable result such as a report, document, or JSON response.


## Types of Agent Interactions in A2A

### 1. Cooperative Interaction

Agents work together toward a shared goal by sharing data, resources, and strategies.
Example: Supply chain agents coordinating inventory and deliveries.

### 2. Competitive Interaction

Agents have conflicting goals and compete to achieve the best outcome for themselves.
Example: Bidders competing in an online auction.

### 3. Negotiative Interaction

Agents communicate and negotiate to reach a mutually acceptable agreement.
Example: Buyer and seller negotiating price and delivery terms.

### 4. Mediated Communication

A third-party mediator agent helps coordinate communication between agents.
Example: Traffic control system managing multiple vehicle agents.

## Types of Agentic Architectures
There are several types of agentic architectures each with its own strengths and weaknesses, suitable for different tasks and environments. Some common types include:

### 1. Single-agent architecture: A single AI system that functions independently, making decisions and taking actions without the involvement of other agents.

1. Perception
Perception is the ability of an agent to collect and understand information from its environment using inputs like text, images, audio, sensors, APIs, and databases to detect changes and important events.

2. Cognitive Layer
The cognitive layer analyzes the collected information, evaluates possible outcomes, and decides the best action based on goals using methods like rule-based systems, machine learning models, and reinforcement learning.

3. Action and Execution
The action layer is responsible for executing the decisions made by the agent by interacting with the environment through robotics, software automation, APIs, or external systems.

4. Learning and Adaptation
Learning and adaptation allow the agent to improve over time by learning from experience using supervised learning, unsupervised learning, and reinforcement learning techniques.

### 2. Multi-agent architecture: This architecture involves multiple AI systems interacting with each other, collaborating and coordinating their actions to achieve common goals. Its sub types are:

- Vertical architecture: This approach involves agentic AI systems organized in a hierarchical structure with higher-level agents overseeing and guiding the actions of lower-level agents.
- Horizontal architecture: This involves agentic AI systems operating on the same level without any hierarchical structure, communicating and coordinating their actions as needed.
- Hybrid architecture: This involves a combination of different agentic architecture types and using the strengths of each to achieve optimal performance in complex environments.

## Agentic Framework
A structured system that defines how autonomous agents interact, make decisions and execute tasks. It integrates components like planning, memory, tool usage and communication to enable goal-driven behavior.

- Reactive Architectures: It link immediate environmental input to direct responses. These agents operate on instinct, acting solely on current stimuli without using memory or planning. They are unable to learn from the past or anticipate future events.
- Deliberative Architectures: It enable agents to act by reasoning, strategizing and building internal representations of the world. Unlike reactive systems, these agents assess their surroundings, consider possible outcomes and make decisions after thoughtful analysis.
- Cognitive Architectures: It simulate human-like intelligence. They combine perception, reasoning, learning and decision-making abilities, allowing agents to handle complex tasks by thinking and adapting much like people do.

## Model Context Protocol (MCP) 

is a standard framework by Anthropic that allows AI models to securely connect with external tools and data sources. It helps models access real-time information without needing custom integrations. MCP makes system integration easier, faster, and more scalable across different applications and industries.

## MCP (Model Context Protocol) Architecture



### MCP host
The LLM is contained within the MCP host, an AI application or environment such as an AI-powered IDE. This is typically the user's interaction point, where the MCP host uses the LLM to process requests that may require external data or tools.

### MCP client
The MCP client, located within the MCP host, helps the LLM and MCP server communicate with each other. It translates the LLM's requests for the MCP and converts the MCP's replies for the LLM. It also finds and uses available MCP servers.

### MCP server
The MCP server is the external service that provides context, data, or capabilities to the LLM. It helps LLMs by connecting to external systems like databases and web services, translating their responses into a format the LLM can understand which helps developers provide diverse functionalities.

### Transport layer
The transport layer uses JSON-RPC 2.0 messages to communicate between the client and server, mainly through two transport methods:

**Standard input/output (stdio)**: Works well for local resources, offering fast, synchronous message transmission              
**Server-sent events (SSE)**: Preferred for remote resources, allowing efficient, real-time data streaming

### How does the MCP work?
Imagine you ask an AI assistant: "Find the latest sales report in our database and email it to my manager."

**Request and tool discovery**: The LLM understands it cannot access a database or send emails on its own. It uses the MCP client to search for available tools, where it finds two relevant tools registered on MCP servers: a database_query tool and an email_sender tool.   
**Tool invocation**: The LLM generates a structured request to use these tools. First, it calls the database_query tool, specifying the report name. The MCP client then sends this request to the appropriate MCP server.      
**External action and data return**: The MCP server receives the request, translates it into a secure SQL query for the company's database, and retrieves the sales report. It then formats this data and sends it back to the LLM.     
**Second action and response generation**: Now equipped with the report data, the LLM calls the email_sender tool, providing the manager's email address and the report content. After the email is sent, the MCP server confirms the action was completed.         
**Final confirmation**: The LLM provides a final response to you: "I have found the latest sales report and emailed it to your manager."            


## Choosing the right MCP server for your needs

### Remote vs. local servers
MCP servers can be deployed either locally alongside the AI application or remotely on a separate machine or cloud service.

Local servers: Best for tasks requiring high speed and low latency, such as providing context from a local integrated development environment (IDE) or a private file system. They use fast transport methods like standard input/output (stdio) and offer enhanced security as sensitive data does not need to travel over a network. This makes them ideal for offline applications or processing confidential information.

Remote servers: Offer greater flexibility and scalability, allowing LLMs to connect to public web APIs (for example, for weather data or stock prices) or shared enterprise services. Using transport methods like Server-Sent Events (SSE), remote servers enable real-time data streaming and can be accessed by multiple AI applications, making them perfect for building shared, scalable tools.

### Managed vs. self-hosted servers
Beyond the location, you can choose who manages the underlying infrastructure.

Managed servers: Leveraging serverless platforms (like Cloud Run) or container orchestration services (like Google Kubernetes Engine), managed servers abstract away the complexity of infrastructure. They provide auto-scaling, high availability, and built-in security, allowing developers to focus on building tool functionality rather than managing server maintenance and uptime. This is the recommended approach for enterprise-grade, scalable applications.

Self-hosted servers: This approach provides maximum control over the deployment environment. You can run an MCP server on your own on-premises hardware or a custom virtual machine in the cloud. This option is suitable for organizations with highly specific security, compliance, or legacy system integration requirements where a managed environment may not be feasible.


## A2A vs MCP 

- A2A (Agent2Agent) focuses on enabling communication and collaboration between autonomous agents, whereas MCP (Model Context Protocol) focuses on connecting AI models with external tools, APIs, and data sources.
- A2A is created by Google, while MCP is created by Anthropic.
- In A2A, the key concepts include Agent Cards, Tasks, Messages, HTTP/JSON-RPC, and SSE for real-time communication, while in MCP, the main components are Host, Client, Server, Tools, Resources, and Prompts.
- A2A uses task-based, asynchronous communication where agents exchange messages to complete work, whereas MCP uses structured request-response communication using schemas like JSON Schema.
- A2A is mainly used for collaboration between multiple independent agents, while MCP is used to give AI models access to external systems like databases, files, and APIs.

## MCP vs Normal Tool Calling

- MCP (Model Context Protocol) is a structured framework that connects AI models to external tools and data sources using a standardized Host–Client–Server architecture, whereas a normal tool call is a direct function or API call made by the model without any protocol layer.
- MCP supports dynamic tool discovery at runtime, allowing the model to see and select available tools, whereas normal tool calls use predefined and hardcoded functions that the model already knows.
- MCP uses a standardized communication format (like JSON-RPC and schemas) to ensure consistent interaction, while normal tool calls rely on direct function execution or API requests without a shared protocol.
- MCP is loosely coupled and highly scalable, making it easy to add or update tools without changing the model logic, whereas normal tool calls are tightly coupled and require code changes for every new tool.
- MCP manages context, sessions, and orchestration between components, while normal tool calling has minimal or no built-in context management.
- MCP is suitable for complex AI systems with multiple integrated services, whereas normal tool calls are better for simple, fixed-function use cases.

## MCP vs RAG 
- Purpose: MCP enables LLMs to interact with external tools, APIs, and systems, while RAG enhances LLM responses by retrieving relevant information from a knowledge base.
- Functionality: MCP supports both information retrieval and action execution (e.g., updating a CRM or booking a ticket), whereas RAG mainly focuses on retrieving information to improve response accuracy.
- Interaction Model: MCP provides two-way communication between AI models and external systems, while RAG offers one-way retrieval of documents that are added to the prompt.
- Use Cases: MCP is ideal for AI agents that need to perform tasks and integrate with external services, whereas RAG is commonly used for question answering, document search, summarization, and reducing hallucinations.
