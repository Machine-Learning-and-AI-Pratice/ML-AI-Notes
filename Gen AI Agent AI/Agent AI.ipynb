## Agentic AI

Agentic AI is a goal-driven AI system where an LLM acts as the reasoning engine and can dynamically plan tasks, select and invoke tools, observe the results, maintain state or memory, and re-plan until the goal is achieved, rather than simply generating a response.

### Agentic AI vs traditional LLM 

A traditional LLM application is generally request-response based: the user provides a prompt, the LLM generates an answer, and the application returns it. 

Agentic AI is goal-driven and can autonomously decide what steps to take, select tools or skills, execute actions, observe results, maintain state or memory, and re-plan until the objective is completed

### AI Agent vs AI workflow 

An AI workflow follows a predefined sequence of steps, whereas an AI agent can dynamically decide what steps to take based on the goal, current state, and tool results


## AI Agent

The core components of an AI agent are the reasoning model, instructions, state and memory, tools or skills, planning and decision-making, an execution loop, and guardrails. Together they allow the agent to understand a goal, decide what to do, take actions, observe results, and adapt until the task is completed.

### Core components

```text
                 AI AGENT
                    │
       ┌────────────┼────────────┐
       ↓            ↓            ↓
   LLM/Reasoning  State/Memory  Tools/Skills
       │            │            │
       └────────────┼────────────┘
                    ↓
             Planning / Decision
                    ↓
              Execution Loop
                    ↓
              Observe Result
                    ↓
             Re-plan / Continue
                    ↓
               Final Result
```

1. **LLM / Reasoning Engine**
   Understands the request, reasons about the task, and decides the next action.

2. **Instructions / System Prompt**
   Defines the agent's role, objectives, behavioral rules, and constraints.

3. **State**
   Tracks the current execution context—goal, current step, tool results, errors, and intermediate outputs.

4. **Memory**
   Stores information that may be useful later, such as conversation history or previous experiences.

5. **Tools / Skills**
   Give the agent capabilities beyond generating text—APIs, databases, search, code execution, enterprise systems, etc.

6. **Planning & Decision-Making**
   Decomposes complex goals and determines what action should happen next.

7. **Execution Loop**
   Typically follows:

   **Plan → Act → Observe → Update State → Re-plan**

8. **Guardrails & Policies**
   Control what the agent is allowed to do, validate inputs/outputs, enforce permissions, and require human approval for risky actions.

9. **Orchestration / Runtime / Harness**
   Manages execution, retries, timeouts, budgets, tool calls, persistence, and observability.


### Agent decide which tool to use 

An agent typically decides which tool to use by comparing the user's goal with the available tool descriptions, capabilities, input schemas, and current context. The LLM selects the most appropriate tool and generates structured arguments. Before execution, the runtime or harness validates the tool call against permissions and policies.


## Core Classical AI Agents

**Simple reflex agents:** Act directly on the current percept using condition-action rules without maintaining memory.
**Example:** A thermostat turning on heating when temperature drops.

**Model-based reflex agents:** Maintain an internal state/model of the environment to handle information that is not immediately observable.
**Example:** A robot vacuum remembering which areas of a room it has already cleaned.

**Goal-based agents:** Use explicit goals to evaluate possible actions and choose a sequence that achieves the desired outcome.
**Example:** A GPS system finding a route to a destination.

**Utility-based agents:** Evaluate possible outcomes using a utility function and select the action that provides the highest overall benefit or best trade-off.
**Example:** A financial agent balancing risk and return when managing an investment portfolio.

**Learning agents:** Learn from previous experiences and feedback to improve their future decisions and performance.
**Example:** A recommendation system adapting to a user's previous interactions.

## Advanced and System-Level Agents

**Multi-agent systems (MAS):** Use multiple independent or specialized agents that cooperate, communicate, or compete to solve complex problems.
**Example:** Research agent + analysis agent + writing agent working together.

**Hierarchical agents:** Organize agents into multiple levels, such as manager → supervisor → specialist agents, to handle complex workflows.
**Example:** An enterprise AI system with a manager agent coordinating IT, HR, and Finance supervisors.


## Agent State

Agent State is the shared memory structure that stores the current status of an AI Agent during execution. It contains conversation history, task progress, intermediate results, tool outputs, and control information. In frameworks like LangGraph, state is passed between nodes and updated after every step, enabling reliable multi-step agent workflows

## Agent memory 

Agent memory is the mechanism that allows an AI agent to store, retrieve, and use information from previous interactions, actions, and experiences.

## Short-term memory 

is the agent's temporary memory used to maintain context during the current conversation or task. often Redis/checkpoint

## Long-term memory 

Long-term memory allows an AI agent to store important information and retrieve it later, even across different conversations or sessions. used by database/ vector database 

### Episodic memory 

stores an agent's past experiences, including the situation, actions performed, and outcomes. For example, if an IT agent previously resolved a VPN issue by restarting the VPN service, it can remember that episode and use it when a similar issue occurs again. Episodic memory is a type of long-term memory.

### Semantic memory 

stores factual knowledge, concepts, and information that an agent can retrieve and use later.semantic memory remembers general facts. It is commonly implemented using a vector database or structured database.

# Agent Pattern 

## ReAct Pattern (Reason + Act)

The ReAct (Reason + Act) pattern is an AI agent framework where the model reasons about a problem, takes an action (tool call), observes the result, and repeats until it reaches the final answer.

Example: IT troubleshooting, research, or dynamic API workflows.

TradeOff: High latency, Higher cost, Less predictable and Tool-call errors. 

## Plan-and-Execute

The Plan-and-Execute pattern is an AI agent architecture designed to solve complex, multi-step tasks by separating the reasoning process into two phases: 

- Planner – The LLM first creates a high-level plan by breaking the user's goal into smaller steps.

- Executor – The agent then executes those steps one by one, using tools, APIs, or other agents as required.

Example: My VPN is not working. Fix it and create a ticket if necessary.

TradeOff: Higher latency, Higher cost, Plan can become stale if the environment changes during execution, Needs re-planning when a step fails, More complex than a simple ReAct agent

### 1. Self-Correcting Agent

A **self-correcting agent** generates an output, evaluates whether it is correct, and if it finds errors, it **loops back to fix its own mistakes**. The process repeats until the output meets the required quality.

### 2. Iterative Agent

An **iterative agent** repeats a task to **collect more information or improve the result**, rather than fixing mistakes.

It continues until it reaches a stopping condition.

# Multi-Agent 

A **Multi-Agent** system in LangGraph consists of multiple specialized agents that collaborate to solve complex tasks. Each agent focuses on a specific responsibility, such as research, coding, or data analysis, while a supervisor or orchestrator coordinates their work and combines the results into a final response. 

## Supervisor Pattern

A **Supervisor Agent** acts as the **orchestrator**. It does **not** perform the work itself. Instead, it analyzes the user's request, delegates tasks to the appropriate specialized agents, collects their results, and returns the final response.

Example: It is widely used in enterprise AI assistants, financial analysis platforms.

TradeOff: The main trade-off is better coordination and modularity versus increased latency, cost, complexity, and the risk of the supervisor becoming a bottleneck or single point of failure.

# Hierarchical Pattern in LangGraph

The **Hierarchical Pattern** is a multi-agent architecture where there are multiple levels of supervisors. A top-level supervisor delegates tasks to domain-specific supervisors, and each domain supervisor coordinates specialized worker agents. This pattern is commonly used in large enterprise applications because it improves scalability, modularity, and maintainability. It is similar to an organizational hierarchy, where managers oversee teams and report to senior managers.



## Collaborative Pattern

The Collaborative Pattern is a multi-agent architecture where specialized agents work together toward a common goal by sharing information, providing feedback, and building on each other's outputs.

TradeOff: trade-offs are higher cost, latency, orchestration complexity, and the need to manage agent conflicts or collaboration loops.


## The Sequential Pattern 

is a multi-agent workflow where agents execute in a predefined order, and each agent consumes the output of the previous agent. 

Example: It's ideal for dependent, pipeline-based tasks such as document processing, where extraction must happen before analysis and analysis before summarization. 

TradeOff:higher latency and error propagation because a failure in one stage can affect all downstream agents.

## The Parallel Pattern 

is a multi-agent workflow where multiple agents execute simultaneously on independent tasks. After all agents complete their work, their outputs are merged—typically using a reducer—to generate the final response. This will help reducing latency


Example:  Planning a Trip will run Attraction Agent Hotel Agent and Food Agent all in parallel 

TradeOff: increased cost, concurrency management, and complexity in handling conflicting outputs."


## Agent Loop 

An Agent Loop is the repeated cycle an AI agent follows to understand the goal, decide what to do, use tools, observe the result, and continue until the task is completed.

## prevent infinite agent loops

In Agentic AI, we prevent infinite loops by defining explicit stopping conditions and safety limits.

Tech: Maximum iteration limit, Timeout, Tool-call limit, Error/retry limit and Tokens limit

## agent failures are handled 

using error handling, retries, fallbacks, validation, and human escalation so that one failed agent doesn't necessarily bring down the entire workflow.

# Agent tool calling 

A tool is an executable capability provided to an AI agent. The LLM decides when to use the tool, generates the required arguments according to the tool schema, and the application executes the tool and returns the result back to the agent

The LLM doesn't directly execute tools. It looks at the user's request, compares it with the available tool names, descriptions, and input schemas, and selects the tool whose capability best matches the task. It then generates the required arguments in the schema-defined format. The agent framework executes the tool and sends the result back to the LLM.

## validate tool inputs and outputs

I validate tool inputs using strict schemas such as Pydantic or JSON Schema, along with type checks, required fields, allowed values, authorization, and business rules. After the tool executes, 

I validate its output against the expected schema and check for errors, missing fields, and business-rule violations. If validation fails, I use bounded retries, fallback logic, or human escalation rather than trusting the result blindly

## prevent unnecessary tool calls

I prevent unnecessary tool calls by first checking the user's intent and current state, using caching for previously retrieved information, applying conditional routing, providing clear tool descriptions, and enforcing tool-call limits. For predictable requests, I prefer deterministic routing instead of relying entirely on the LLM. I also check whether the required information is already available before making another tool call

## Tool Failure Error Handle 

handle tool failures by first classifying the error. For transient failures such as timeouts, 429s, and 5xx errors, I use bounded retries with exponential backoff. If retries are exhausted, I use a fallback tool or agent. I validate tool outputs against a schema and don't retry permanent errors such as authorization failures. For critical or unrecoverable failures, I use human escalation, along with timeouts, circuit breakers, logging, and monitoring for production reliability

## design 100+ tools without overwhelming the LLM context

I would avoid exposing all 100+ tools to the LLM at once. I would use tool discovery, routing, namespaces, and hierarchical agents so that the LLM only sees the small set of tools relevant to the current task.

# Skills 

## Agentic Skills 

Agentic Skills are reusable capabilities that an AI agent can use to accomplish a particular type of task. A skill is usually higher-level than a single tool: it can combine multiple tools, instructions, reasoning steps, and policies into one reusable capability.

Example: in an enterprise IT agent, troubleshoot_vpn can be a skill that checks VPN status, searches troubleshooting documentation, determines the cause, creates an incident if required, and notifies the user. Individual functions like check_vpn() and create_ticket() are tools, while the agent decides when to invoke the overall skill.”

## Skills vs Agent and Tools 

An agent is the decision-making and orchestration layer. It understands the user's goal, selects the appropriate capability, and decides what to do next.       

A skill is a reusable higher-level capability or workflow that may combine multiple tools to accomplish a specific task.        

A tool is the lowest-level executable operation, such as calling an API, querying a database, searching documents, or restarting a VM.          

## Skills in an Agentic AI architecture

We need Skills because an agent may have hundreds of low-level tools, and asking the LLM to reason directly over every individual tool makes the system harder to manage, less reliable, and more expensive.

## Skills loading to the Agent 

An Agent discovers and selects a Skill using a Skill Registry and a routing layer. The registry stores metadata such as the Skill's name, description, domain, required tools, permissions, and risk level. 

For each user request, a router performs semantic or metadata-based retrieval to identify the most relevant Skills. After authorization and policy filtering, the Agent loads the detailed instructions for the selected Skill and executes it using the required tools. This progressive-discovery approach scales much better than putting hundreds of Skills directly into the LLM context

## Multiple skills can solve the same request        

When multiple Skills can solve the same request, I use a Skill Router or Ranker rather than selecting the first match. I rank candidates based on semantic relevance, specialization, user permissions, availability, risk, and cost. I prefer the most specialized Skill when confidence is high. If confidence is low or two Skills have similar scores, I either ask a clarification question or use a controlled fallback strategy. If the task naturally requires multiple capabilities, I allow Skills to collaborate sequentially or in parallel




## Agent2Agent (A2A) 

Agent-to-Agent (A2A) Protocol is an open communication protocol developed by Google that enables AI agents to communicate, collaborate, exchange tasks, and work together across different systems, frameworks, and organizations.

## workflow

The **Agent2Agent (A2A) Protocol** follows a client-server model where one agent sends a task request and another agent executes it. For example, a CustomerBot can request an OrderBot to check product availability and return the result. Agents can switch between client and server roles depending on the interaction.

Each agent exposes an **Agent Card**, a JSON profile containing its identity, capabilities, and security requirements, enabling agent discovery and selection. Work is managed as a **Task**, which progresses through states such as Submitted, Working, Input-Required, Completed, Failed, or Cancelled.

Agents exchange information using structured **Messages** (text, files, forms), and the final output is returned as an **Artifact**, a structured and reusable result such as a report, document, or JSON response.


## Types of Agent Interactions in A2A

### 1. Cooperative Interaction

Agents work together toward a shared goal by sharing data, resources, and strategies.
Example: Supply chain agents coordinating inventory and deliveries.

### 2. Competitive Interaction

Agents have conflicting goals and compete to achieve the best outcome for themselves.
Example: Bidders competing in an online auction.

### 3. Negotiative Interaction

Agents communicate and negotiate to reach a mutually acceptable agreement.
Example: Buyer and seller negotiating price and delivery terms.

### 4. Mediated Communication

A third-party mediator agent helps coordinate communication between agents.
Example: Traffic control system managing multiple vehicle agents.

## Agent Card 

An Agent Card is a machine-readable metadata document that describes an AI Agent's identity, capabilities, tools, endpoints, authentication, and supported communication protocols. It allows other agents or orchestrators to discover the agent, understand what it can do, and interact with it automatically. It is similar to an OpenAPI specification but designed specifically for AI Agents

## Agent discover another Agent

One agent discovers another by querying an Agent Registry or Directory Service. The registry returns the target agent's Agent Card, which contains its capabilities, endpoint, authentication details, and supported protocols. The requesting agent validates that the discovered agent can perform the required task, authenticates if necessary, and then invokes the agent. This enables dynamic discovery and interoperability in multi-agent systems

## A2A vs API Communication  

API communication is a direct request-response interaction where a client calls a predefined endpoint to perform a specific operation. A2A (Agent-to-Agent) communication enables autonomous collaboration between AI agents. Agents discover each other using Agent Cards, reason about tasks, and delegate work dynamically. APIs are operation-oriented, while A2A is goal-oriented and designed for multi-agent workflows

## Json RPAC 

JSON-RPC (JavaScript Object Notation - Remote Procedure Call) is a lightweight protocol that allows one application to call a function (procedure) on another application over a network using JSON messages.

Think of it as "calling a remote function using JSON."

## Example 

A2A (Agent-to-Agent) enables specialized AI agents to communicate and collaborate to complete complex tasks. For example, an IT support request can be split by an Orchestrator into Device, Email, and Asset Management agents. Each agent performs its own task and returns results through the A2A protocol. The Orchestrator then combines the responses into a single, comprehensive answer for the user.
