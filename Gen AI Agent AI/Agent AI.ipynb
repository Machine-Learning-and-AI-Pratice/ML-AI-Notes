# Agentic AI

## What is Agent 

An Agent is a system that leverages an AI model to interact with its environment in order to achieve a user-defined objective. It combines reasoning, planning, and the execution of actions (often via external tools) to fulfill tasks.

- Understand natural language: Interpret and respond to human instructions in a meaningful way.

- Reason and plan: Analyze information, make decisions, and devise strategies to solve problems.

- Interact with its environment: Gather information, take actions, and observe the results of those actions.

## What are AI Tools?
A Tool is a function given to the LLM. This function should fulfill a clear objective.


- Web Search: Allows the agent to fetch up-to-date information from the internet.
- Image Generation:	Creates images based on text descriptions.
- Retrieval: Retrieves information from an external source.
- API Interface:	Interacts with an external API (GitHub, YouTube, Spotify, etc.)

Large language models (LLMs) don’t actually use tools themselves—they only process and generate text. When we say an “agent” can use tools, what’s really happening is:

The LLM is told what tools exist and how to call them using text.
When a question requires external data (like weather), the LLM outputs a structured text request (e.g., call weather_tool("Paris")) instead of answering directly.
The agent system (not the LLM) detects this request, executes the tool, and gets real data.
That result is fed back into the conversation as additional context.
The LLM then uses this new information to produce a final, natural-language answer.

## Agent Workflow

Agents’ work is a continuous cycle of: thinking (Thought) → acting (Act) and observing (Observe).

## Example:- 

Alfred is an AI agent designed to answer user queries by combining reasoning with tool usage. When a user asks for the current weather in New York, Alfred follows a cyclical process:

- Thought (Reasoning):
Alfred understands the request and decides it needs real-time weather data, which requires calling an external API.
- Action (Tool Use):
It sends a structured request to a weather tool (e.g., get_weather with location = New York).
- Observation (Feedback):
The tool returns live data (e.g., temperature, conditions, humidity), which Alfred receives and processes.
- Updated Thought:
Alfred uses this new information to prepare a complete answer.
- Final Answer:
It responds to the user with accurate, up-to-date weather details.

## Chain-of-Thought (CoT)
Chain-of-Thought (CoT) is a prompting technique that guides a model to think through a problem step-by-step before producing a final answer.

This approach helps the model reason internally, especially for logical or mathematical tasks, without interacting with external tools.

## ReAct: Reasoning + Acting
A key method is the ReAct approach, which combines “Reasoning” (Think) with “Acting” (Act).

ReAct is a prompting technique that encourages the model to think step-by-step and interleave actions (like using tools) between reasoning steps.

## 🤖 Agent Actions

Actions are the concrete steps an AI agent takes to interact with its environment.  
These can include retrieving information, calling APIs, or interacting with systems.

## 🔁 The Stop and Parse Approach

- The LLM communicates actions purely through **text output**  
- It describes:
  - **What action to take**
  - **What parameters to use**
- Once a complete action is generated (JSON, code, or function call), the LLM must **stop generating tokens**

### Why this matters:
- Ensures the output is **structured and parseable**
- Transfers control from the **LLM → Agent system**
- Prevents malformed or incomplete actions


# Observation: Integrating Feedback to Reflect and Adapt

## Overview
Observations are how an agent perceives the outcomes of its actions. They provide essential feedback that informs the agent’s reasoning and helps guide future decisions.

## Key Functions of Observations
- **Collect Feedback**  
  Gather signals from the environment (e.g., API responses, error messages, logs) to determine if an action succeeded or failed.

- **Append Results**  
  Add new information to the agent’s existing context (memory), ensuring updated awareness.

- **Adapt Strategy**  
  Use updated context to refine future thoughts, decisions, and actions.

## Example
- API Response: *“partly cloudy, 15°C, 60% humidity”*  
- This data is appended to memory and used to decide whether further actions are needed or a final answer can be given.

## Types of Observations

| Type                  | Examples                                              |
|-----------------------|-------------------------------------------------------|
| System Feedback       | Error messages, success notifications, status codes   |
| Data Changes          | Database updates, file system changes                 |
| Environmental Data    | Sensor readings, system metrics, resource usage       |
| Response Analysis     | API responses, query results, computation outputs     |
| Time-based Events     | Deadlines reached, scheduled task completion          |
