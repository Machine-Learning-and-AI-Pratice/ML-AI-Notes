## Middleware

| Category | Import / Function | Primary Purpose |
| --- | --- | --- |
| **Context Management** | `SummarizationMiddleware` | Summarizes older message history automatically to prevent context window overflow. |
| **Execution Control** | `HumanInTheLoopMiddleware` | Pauses the agent to require human approval before executing specific or sensitive tools. |
| **Lightweight Hooks** | `before_model`, `after_model`, `wrap_tool_call` | Decorators for injecting simple, fast functions directly into the execution loop without a full class. |

## Dynamic Models:

Instead of using just one AI model, the system evaluates the user's request and automatically routes it to the most efficient model (e.g., a cheap model for simple tasks, a powerful model for coding) or swaps to a fallback if a model fails.  

## Dynamic Tools:

Giving an AI too many tools wastes tokens and causes confusion. The system evaluates the prompt and dynamically filters the tool list to provide only the relevant ones for that specific task.  

## Dynamic Prompting:

Rather than using a static instruction, the system builds prompts at runtime by injecting real-time data, relevant documents (RAG), or personalized examples right before sending it to the model.
