# AI Fundamentals

## What is Generative AI?
Generative AI is a type of artificial intelligence that creates new content such as text, images, audio, or code by learning patterns from existing data.

## What is Discriminative AI?
Discriminative AI is a type of machine learning model that focuses on **separating or classifying different classes or labels in data**.
It learns the relationship between input features and the target output to make predictions.
These models are mainly used for tasks such as classification, regression, and pattern recognition.

## Generative Adversarial Networks (GANs)
GANs are generative models that use two neural networks called a generator and a discriminator that compete with each other during training.
The generator creates fake data while the discriminator tries to separate real data from generated data.
Through this competition, the generator learns to produce realistic data such as images.

## Autoencoder
A neural network that learns to encode input data into a compact representation (latent space) and then decode it to reconstruct the original data, used for compression, denoising, and feature learning.

## Variational Autoencoders (VAEs)
VAEs are generative models that learn a **compressed representation (latent space)** of input data.
The model first **encodes the input data into a probability distribution**, then **decodes it to reconstruct or generate new data**.
By sampling from the learned latent space, VAEs can generate **new data that is similar to the training data**.

## Diffusion Models
Diffusion models generate data by gradually adding noise to training data and then learning how to reverse this process.
During generation, the model starts with random noise and progressively removes noise to create structured data.
These models are widely used for high-quality image generation.

# RAG Fundamentals

## What is Retrieval-Augmented Generation (RAG)?
Retrieval-Augmented Generation (RAG) is a framework that combines information retrieval with generative language models.
Instead of relying solely on a model's pre-trained knowledge, RAG first retrieves relevant documents or context from an external knowledge base and then uses that context to generate accurate and up-to-date responses.

## What are the limitations of RAG?
- RAG relies on high-quality retrieved documents, so irrelevant retrievals reduce answer accuracy
- Higher latency due to retrieval and generation steps
- LLM context window is limited, so large retrieved content may be truncated
- May hallucinate or give wrong answers if retrieved data is noisy or contradictory
- Scaling and storage of embeddings for millions of documents can be costly
- Reasoning is limited to retrieved content, missing knowledge outside it

## Agentic RAG
Agentic RAG is an advanced approach where the LLM acts as an agent that can:
- Decide what to retrieve
- Decide when to retrieve
- Perform multi-step reasoning before generating the final answer

## Self-RAG
Self-RAG is an advanced Retrieval-Augmented Generation framework that uses reflection tokens to enable the model to evaluate its own generated responses, decide whether additional information retrieval is needed, and refine the answer. This self-reflection process improves accuracy, reliability, and factual consistency while reducing hallucinations.

### Reflection Tokens:
Reflection tokens are special tokens used in Self-RAG that allow the model to assess the quality of its generated response, determine whether more information should be retrieved, and guide the refinement of the final answer to improve accuracy and reduce hallucinations.

## Corrective RAG (CRAG)
Corrective Retrieval-Augmented Generation (CRAG) is an enhanced RAG framework that evaluates the quality of retrieved documents and, when the retrieved information is insufficient or unreliable, performs corrective actions such as retrieving additional data or refining the context to improve the accuracy of the generated response.

## How do you manage context windows when retrieved content exceeds LLM limits?
When retrieved content exceeds the LLM context window:
- Select the most relevant top-k chunks based on similarity scores
- Summarize or compress retrieved documents to reduce token usage
- Use reranking, chunk filtering, or sliding window approaches to keep only the most important context

# Prompt Engineering

## What is Prompt Engineering?
Prompt engineering is the process of designing and refining inputs (prompts) to guide a language model to produce desired outputs.

## System vs User vs Assistant Prompts
In conversational LLMs, prompts are categorized to structure interactions and guide model behavior:
- **System Prompts**: Instructions given to the model before the conversation starts to define its role, behavior, or style
- **User Prompts**: Inputs provided by the end user, usually questions, commands, or requests
- **Assistant Prompts**: Previous responses generated by the model itself, used as context in multi-turn conversations

## Chain-of-Thought (CoT) Prompting
Chain-of-Thought prompting encourages the LLM to reason step by step before producing a final answer.
Instead of giving a single-step response, the model breaks down its reasoning, improving accuracy on complex tasks.
Example:
Instead of asking: "What is 23 × 17?"
A CoT prompt: "Let's think step by step: 23 × 17 = ?"
The model generates intermediate steps:
- 23 × 10 = 230
- 23 × 7 = 161
- 230 + 161 = 391
Finally, outputs the answer: 391
CoT can be used in zero-shot or few-shot prompting, often combined with examples that demonstrate step-by-step reasoning.

## Role Prompting
Role prompting assigns the LLM a specific role or persona in the prompt to guide its behavior, tone, and style.

## How do you reduce hallucinations using prompts?
Hallucinations can be reduced by:
- Grounding the model with context
- Instructing it not to guess
- Enforcing structured outputs
- Encouraging step-by-step reasoning

## Zero-shot Prompting
Zero-shot prompting asks a model to perform a task without providing any examples in the prompt.
The model relies only on its pretrained knowledge and the instructions in the prompt to generate the answer.

## Few-shot Prompting
Few-shot prompting provides a few example inputs and outputs in the prompt so the model understands how to perform the task.
Example:
Extract the company name.
Text: Apple released a new iPhone.
Company: Apple
Text: Microsoft announced a new AI service.
Company: Microsoft
Text: Google introduced a new model.
Company:

## Contextual Prompting 

is a prompt engineering technique where you provide the AI with relevant background information, specific instructions, tone and objectives within your prompt. This ensures the AI generates responses that are not just accurate but also tailored to your needs and context.

## ReAct (Reasoning + Acting) 
is a prompting technique where an AI combines step-by-step reasoning with actions, allowing it to think through a problem and take steps to solve it.
Breaks complex problems into steps and decides actions at each stage
Uses results from actions (observations) to refine and improve future decisions
Used in chatbots, robotics, and automation

## Retrieval-Augmented Prompting
Retrieval Augmented Prompting (RAP) improves AI by enabling it to access external information along with its trained knowledge, resulting in more accurate, relevant and up to date responses.

## Tree of Thought (ToT) 
Tree of Thought (ToT) is a reasoning technique where an LLM explores multiple possible solution paths instead of following a single chain of reasoning.

# Embeddings & Document Handling

## What are Embeddings?
Embeddings are numerical vector representations of data such as text, images, or audio.
They capture the semantic meaning and relationships between pieces of data in a high-dimensional space.
Similar items have embeddings that are close to each other in this vector space.

## Common Embedding Models
- **Sentence-BERT (SBERT)**: Widely used for semantic search and similarity tasks
- **OpenAI Embeddings**: Commonly used in RAG applications for high-quality vector representations
- **BERT-based models**: Used for generating contextual word and sentence embeddings
- **MiniLM / all-MiniLM**: Lightweight models for fast and efficient embeddings
- **E5 and Instructor models**: Optimized for retrieval and RAG tasks with strong semantic search performance

## What is Chunking?
Chunking is the process of splitting large documents into smaller, manageable pieces (chunks) before storing them in a vector database for retrieval.
Each chunk is usually a few sentences, paragraphs, or semantically coherent sections.

## Need to chucking

LLM Token Limitations: Long documents may exceed token limits, making direct processing inefficient or requiring techniques like truncation or sliding windows      
Improved Retrieval Accuracy: Smaller segments allow retrieval pipelines to match context more precisely.        
Better Performance: Chunking reduces computation overhead and speeds up embedding searches.     
Context Preservation: Keeps relevant text together, reducing hallucinations and incorrect reasoning.            
Efficient Knowledge Access: Enables document querying without loading entire files into memory.             

## Chunking Strategies
- **Fixed-size chunking**: Fixed-size chunking involves dividing the text into chunks of a predefined size, typically based on the number of characters or tokens. General RAG pipelines, large-scale indexing, quick implementation.
- **Recursive chunking**: Recursive chunking divides the text hierarchically using a set of separators. If the initial chunks are too large, the method recursively splits them until the desired size is achieved. PDFs, reports, books, technical documentation.
- **Sentence-based chunking**: Sentence splitting involves dividing the text into individual sentences using punctuation marks. Q&A systems, summarization, conversational AI.
- **Semantic chunking**: Semantic chunking involves grouping sentences or phrases based on their semantic similarity. This method often uses clustering algorithms or embedding models. High-accuracy RAG, legal documents, research papers, knowledge bases.

## Chunk Overlap
Chunk overlap refers to the technique of including a small portion of text from the end of one chunk at the beginning of the next chunk. This helps maintain continuity between chunks and prevents important information from being lost when text is split. It is especially useful when sentences or ideas span across multiple chunks.

## Selecting Chunk Sizes
Choosing the right chunk size depends on the type of document and the use case. If chunks are too large, the model may include unnecessary data. If chunk is too small, it may lose essential meaning. Some recommended chunk sizes in LangChain are:

300–500 Tokens: Useful for most general documents where moderate context is needed.
600–900 Tokens: Ideal for technical guides and manuals requiring deeper reference context.
100–200 Tokens: Effective for short chats, logs or small knowledge fragments.

## Semantic Similarity
Semantic similarity measures how similar two pieces of text are based on meaning rather than exact words.
It uses embeddings and vector similarity (e.g., cosine similarity) to compare the meaning of sentences or documents.
This helps systems like RAG retrieve relevant information even when the query and document use different words but have the same meaning.

## Cosine Similarity
Cosine similarity measures how similar two vectors are by calculating the angle between them.
Formula: cos(θ) = (A · B) / (|A||B|)
- Value close to 1: vectors are very similar
- Value close to 0: vectors are unrelated
- Value close to -1: vectors are opposite
In RAG systems, query embeddings are compared with document embeddings using cosine similarity.

## Euclidean Distance
Euclidean distance measures the straight-line distance between two vectors in multi-dimensional space.
- Smaller distance → vectors are more similar
- Larger distance → vectors are less similar
Cosine similarity is often preferred for text because it focuses on the angle (semantic similarity) rather than magnitude.

# algorithms for finding the Nearest Neighbors (NN)

## HNSW (Hierarchical Navigable Small World) 

is an Approximate Nearest Neighbor (ANN) algorithm used for fast vector similarity search. It builds a multi-layer graph where each vector connects to nearby neighbors. During search, it starts from the top layer and progressively moves to lower layers to quickly locate the closest vectors. HNSW offers O(log N) average search time, high recall, and is widely used in vector databases like Azure AI Search, FAISS, Pinecone, and Qdrant.

## IVF (Inverted File Index)

IVF (Inverted File Index) is an Approximate Nearest Neighbor (ANN) algorithm that speeds up vector search by grouping similar vectors into clusters. During a search, it first identifies the closest cluster(s) and then searches only within those clusters instead of the entire dataset. This significantly reduces search time while maintaining good accuracy. IVF is commonly used in FAISS and Milvus, and is often combined with Product Quantization (IVF-PQ) for large-scale vector search.

##  Product Quantization (PQ)

Product Quantization (PQ) is an Approximate Nearest Neighbor (ANN) technique that compresses high-dimensional vectors into compact codes to reduce memory usage. It splits each vector into smaller subvectors, creates codebooks using clustering, and stores only the codebook indices instead of the original values. During search, approximate distances are computed using these codes. PQ is widely used with IVF (IVF-PQ) in FAISS and Milvus for large-scale vector search.

## Brute Force

Brute Force is an Exact Nearest Neighbor (ENN) search algorithm that compares the query vector with every vector in the database. It computes a similarity metric such as Cosine Similarity, Euclidean Distance, or Dot Product, and returns the closest vectors. It guarantees 100% accuracy, but has O(N) search complexity, making it unsuitable for very large datasets.

## LSH (Locality Sensitive Hashing)

Locality Sensitive Hashing (LSH) is an Approximate Nearest Neighbor (ANN) algorithm that hashes similar vectors into the same bucket using locality-sensitive hash functions. During a search, only the vectors in the matching bucket are compared instead of the entire dataset, making retrieval much faster. LSH offers sub-linear search time but may sacrifice some accuracy compared to graph-based methods like HNSW.

# Retrieval Techniques

## BM25
BM25 is a probabilistic ranking function widely used in keyword-based search engines to rank documents based on their relevance to a query.
It's an improvement over TF-IDF that considers term frequency saturation and document length normalization.

## ANN search
Approximate Nearest Neighbor (ANN) search finds vectors that are **closest to a query vector in high-dimensional space**. It is much faster than exact nearest neighbor search. ANN is commonly used for embeddings-based retrieval in LLM systems.


## KNN Seach 

vector search, KNN is used to retrieve the top K most similar embeddings to a query vector using similarity metrics like cosine similarity. It is a core component of RAG systems for finding relevant context.”


## RRF
Reciprocal Rank Fusion (RRF) combines the results of keyword and vector searches. It does this by **adding the reciprocal of each result's rank**, producing a more balanced final ranking. This improves retrieval accuracy for hybrid search systems.


## tiktoken
`tiktoken` is **OpenAI’s tokenizer library**. It counts the number of tokens in a string to ensure that prompts **fit within the model’s context window**. Proper token counting helps avoid truncation or errors when sending requests to the LLM.

## Reranking
Reranking is the process of reordering retrieved documents based on relevance to a query, after an initial retrieval step.
**Why it's used:**
- **Improve relevance**: Initial retrieval (BM25 or vector search) may return less relevant documents
- **Better generation**: Providing the LLM with the most relevant chunks reduces hallucinations
- **Hybrid optimization**: Works especially well combining keyword and embedding-based retrieval

## Top-K Retrieval
Top-k retrieval retrieves the k most relevant document chunks for a query based on similarity scores.
The vector database ranks all documents and returns only the top k results (e.g., top-3 or top-5).
These retrieved chunks are sent to the LLM as context to generate a more accurate answer.

## Query Rewriting
Query rewriting modifies or expands a user query to make it clearer and more effective for retrieval.
The system rewrites the query using conversation context or additional keywords to improve search results.

## Query Decomposition
Query decomposition breaks a complex question into smaller sub-questions.
Each sub-question is retrieved and answered separately using relevant documents.
The system then combines the results to generate a complete and accurate final answer.

## Multi-Query Retrieval
Multi-query retrieval transforms a single user query into multiple related queries to retrieve a broader and more relevant set of documents.
Each sub-query targets a different aspect or interpretation of the original question.

# Data Storage & Indexing


### FAISS (Facebook AI Similarity Search)
Open-source vector search library from Meta used for fast similarity search on local systems.
Supports Flat, IVF, HNSW, and Product Quantization (PQ) indexes with CPU and GPU acceleration, making it ideal for research and prototype RAG applications.

### Pinecone
Fully managed cloud-native vector database built specifically for AI and RAG workloads.
Automatically handles indexing, scaling, metadata filtering, and infrastructure management, making it suitable for production enterprise applications.

### Azure AI Search
Enterprise search service that combines keyword, vector, and hybrid search capabilities in a single platform.
Uses BM25, HNSW, and Semantic Ranker to improve retrieval quality and integrates seamlessly with Azure OpenAI for RAG solutions.

### ChromaDB
Open-source vector database designed specifically for LLM and RAG applications with simple deployment.
Uses HNSW for similarity search and integrates easily with frameworks like LangChain, making it popular for prototyping and small-scale AI projects.


## Indexing in Vector Databases
Indexing organizes and stores embeddings so that similarity search can be performed quickly and efficiently.
When a document is converted into an embedding, it is added to the vector index.
Common indexing methods include **HNSW**, **IVF**, and **PQ**, which help find nearest neighbors without scanning every vector.

## HNSW 
is a graph-based Approximate Nearest Neighbor algorithm that organizes vectors into hierarchical layers and navigates through connected neighbors to quickly find the most similar vectors without scanning the entire dataset.

## How does metadata filtering work in RAG?
Metadata filtering uses additional information about document chunks (author, date, category, language, source) to narrow down retrieval candidates.
When a query is made, filters can be applied (e.g., only recent documents, a specific topic) to improve relevance.

## How do you update or refresh the document index in real-time?
- New documents are processed through the indexing pipeline (cleaning, chunking, embedding generation)
- Generated embeddings are added to the existing vector database without rebuilding the entire index
- Many vector stores support incremental updates and real-time indexing

Use multiple indexes when there are different schemas, security boundaries, business units, retention policies, or very large datasets. For most RAG applications, a single index with metadata filtering is preferred because it is easier to manage and allows searching across all documents.

## How do you handle stale or outdated data?
- Regularly update or refresh the index by reprocessing documents and regenerating embeddings
- Use metadata like timestamps or versioning to identify and filter outdated information
- Implement periodic re-indexing or document replacement

# Performance & Optimization

## How do you reduce latency in RAG pipelines?
- Use efficient vector databases and **ANN (Approximate Nearest Neighbor)** search
- Cache embeddings and frequent query results
- Reduce the number of retrieved chunks (top-k) and optimize chunk size
- Use smaller or optimized embedding models and LLMs

## Caching Strategies
- **Embedding caching**: Store embeddings of documents and queries so they don't need recomputation
- **Query result caching**: Cache results of frequently asked queries with their retrieved chunks
- **Response caching**: Store final LLM-generated answers for common queries for instant results

## How do you balance accuracy vs speed?
- Use smaller or optimized models to improve speed while maintaining accuracy
- Adjust the number of retrieved documents (top-k) to balance quality and processing time
- Apply reranking or filtering to keep only the most relevant chunks
- Use caching and efficient vector search

# Security & Compliance

## How do you handle sensitive or private data in RAG?
- Apply access control and authentication so only authorized users can retrieve sensitive documents
- Encrypt data and embeddings during storage and transmission
- Filter or mask sensitive information before indexing
- Use secure deployment and monitoring to prevent data leakage

## How do you prevent prompt injection attacks?
- Validate and sanitize inputs to detect and block malicious prompts
- Use strict system prompts and guardrails so the model ignores unsafe instructions
- Filter retrieved documents to prevent malicious content from influencing the model
- Implement access control and monitoring to detect suspicious queries

## How do you ensure data privacy in cloud deployments?
- Use data encryption for data at rest and in transit
- Implement strong access control and authentication
- Use secure cloud configurations and compliance policies (IAM roles, network security)
- Regularly monitor, audit logs, and update security settings

## How do you prevent data leakage when integrating with LLMs?
- Filter or mask sensitive information before sending data to the LLM
- Use access controls and role-based permissions
- Implement output filtering and monitoring to prevent the model from exposing confidential information
- Use secure APIs and encrypted communication
