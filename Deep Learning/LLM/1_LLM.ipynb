# 1️⃣ LLM Fundamentals 

## What is an LLM?

A Large Language Model (LLM) is an advanced AI system designed to understand, generate, and manipulate human language by learning patterns from massive amounts of text data.

## How is an LLM different from traditional ML models?




- LLMs are large, flexible, and general-purpose, designed for language understanding and generation.

- Traditional ML models are smaller, task-specific, and rely heavily on feature engineering.

- LLMs learn patterns automatically from huge data; traditional models need explicit design for each task.

- LLMs produce human-like text or structured responses.Traditional ML models produce labels, probabilities, or numeric predictions.

- LLMs require high computational power for training and inference. Traditional ML models are computationally lighter and faster to train.

## Transformer vs RNN vs CNN

RNN processes data sequentially one step at a time, CNN captures local patterns from small regions of the input using convolution filters, while Transformers process all tokens in parallel using self-attention.

RNN maintains context through hidden states passed across time steps, CNN focuses on detecting local patterns, whereas Transformers capture relationships between all tokens in the sequence.

RNN struggles with long-range dependencies due to vanishing or exploding gradient problems, CNN requires deeper layers to capture long-distance relationships, while Transformers handle long-range dependencies effectively using attention mechanisms.

RNN training is slower because computation is sequential, CNN training is faster due to parallel convolution operations, while Transformers train faster with full parallel processing on GPUs.

In terms of time complexity, RNN has approximately O(n) sequential computation over the sequence length, CNN has around O(n·k) where k is the kernel size capturing local patterns, while Transformers have O(n²) complexity because self-attention computes relationships between every pair of tokens.

## Why transformers work well for language?

Transformers work well for language because they use self-attention, which allows the model to understand relationships between all words in a sentence. This helps capture context and long-range dependencies effectively. They also support parallel processing, which makes training faster and scalable. Because of these advantages, Transformers perform very well on modern NLP tasks.


## What is Encoder? 

An encoder is a neural network component that processes input data and converts it into contextual representations that capture the meaning and relationships within the input sequence. In Transformers, the encoder uses self-attention and feed-forward layers to generate these representations.

## What is Decoder? 

decoder is a neural network module that generates the output sequence step by step using the encoded representation from the encoder and previously generated tokens, typically using masked self-attention and encoder–decoder attention in Transformer architectures.

## Encoder vs Decoder 

Encoders are designed to understand and represent input sequences, while decoders are designed to generate output sequences from learned patterns and context.

Encoders convert input text into contextual embeddings that capture relationships and meaning, whereas decoders use context and previously generated tokens to predict and generate the next word step by step.

Encoders process the entire input sequence simultaneously to capture full context, whereas decoders generate tokens sequentially, one token at a time.

Encoders are commonly used in tasks like text classification, sentiment analysis, named entity recognition, and question answering, while decoders are used in text generation, chatbots, dialogue systems, story generation, and code completion.

## Autoregressive Models

Autoregressive models are models that generate output sequentially by predicting the next token based on previously generated tokens. At each step, the model uses the earlier tokens as context and predicts the most probable next token. This process continues until the full sequence is generated.

## Causal vs Bidirectional Models

Causal models predict the next token in a sequence using only past tokens. Bidirectional models use both past and future tokens to understand the full context.

Causal models are autoregressive and generate text step by step. Bidirectional models are designed for understanding and consider the entire sequence simultaneously.

Causal models require sequential processing and rely on previous context for predictions. Bidirectional models require access to the full sequence and use masked language modeling during training.

Causal models are exemplified by GPT series and other autoregressive models. Bidirectional models are exemplified by BERT, RoBERTa, and other understanding-focused models.
