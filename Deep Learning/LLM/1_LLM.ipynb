# LLM Fundamentals 

## What is an LLM?

A Large Language Model (LLM) is an advanced AI system designed to understand, generate, and manipulate human language by learning patterns from massive amounts of text data.

## How is an LLM different from traditional ML models?




- LLMs are large, flexible, and general-purpose, designed for language understanding and generation.

- Traditional ML models are smaller, task-specific, and rely heavily on feature engineering.

- LLMs learn patterns automatically from huge data; traditional models need explicit design for each task.

- LLMs produce human-like text or structured responses.Traditional ML models produce labels, probabilities, or numeric predictions.

- LLMs require high computational power for training and inference. Traditional ML models are computationally lighter and faster to train.

## Transformer vs RNN
Transformer processes entire sequences in parallel using self-attention; RNN processes one token sequentially through hidden states.

Transformer captures long-range dependencies perfectly across any distance; RNN suffers vanishing gradients and forgets distant context.

Transformer adds positional encodings explicitly to understand sequence order; RNN learns sequence order naturally through recurrent loops.

Transformer scales to billions of parameters with GPU-parallel training; RNN limited to small models due to sequential training bottlenecks.

In terms of time complexity, RNN has approximately O(n) sequential computation over the sequence length, CNN has around O(n·k) where k is the kernel size capturing local patterns, while Transformers have O(n²) complexity because self-attention computes relationships between every pair of tokens.

## Why transformers work well for language?

Transformers work well for language because they use self-attention, which allows the model to understand relationships between all words in a sentence. This helps capture context and long-range dependencies effectively. They also support parallel processing, which makes training faster and scalable. Because of these advantages, Transformers perform very well on modern NLP tasks.


## What is Encoder? 

An encoder is a neural network component that processes input data and converts it into contextual representations that capture the meaning and relationships within the input sequence. In Transformers, the encoder uses self-attention and feed-forward layers to generate these representations.

## What is Decoder? 

decoder is a neural network module that generates the output sequence step by step using the encoded representation from the encoder and previously generated tokens, typically using masked self-attention and encoder–decoder attention in Transformer architectures.

## Encoder vs Decoder 

Encoders are designed to understand and represent input sequences, while decoders are designed to generate output sequences from learned patterns and context.

Encoders convert input text into contextual embeddings that capture relationships and meaning, whereas decoders use context and previously generated tokens to predict and generate the next word step by step.

Encoders process the entire input sequence simultaneously to capture full context, whereas decoders generate tokens sequentially, one token at a time.

Encoders are commonly used in tasks like text classification, sentiment analysis, named entity recognition, and question answering, while decoders are used in text generation, chatbots, dialogue systems, story generation, and code completion.

## Encoder–Decoder Architecture in Transformers 

The encoder–decoder architecture in Transformers is used for sequence-to-sequence tasks, where an input sequence is converted into an output sequence.

The encoder processes the input text and creates contextual representations using self-attention mechanisms, which help the model understand relationships between words in the input.

The decoder then uses these encoded representations along with previously generated tokens to generate the output sequence step by step. It also uses masked self-attention to ensure it only looks at past tokens during generation.

This architecture is widely used in tasks like machine translation, text summarization, and question answering. Popular models that use this architecture include T5, BART, and the original Transformer model. 

## Sequence-to-Sequence (Seq2Seq)

A sequence-to-sequence (Seq2Seq) model is a neural network architecture that takes a sequence of inputs and generates another sequence as output.

It is commonly used when the input and output are both sequences and may have different lengths.

Seq2Seq models typically use an encoder to process the input sequence and a decoder to generate the output sequence step by step.

They are widely used in tasks such as machine translation, text summarization, and speech recognition.

## Autoregressive Models

Autoregressive models are models that generate output sequentially by predicting the next token based on previously generated tokens. At each step, the model uses the earlier tokens as context and predicts the most probable next token. This process continues until the full sequence is generated.

## Masked Language Model (MLM)
A Masked Language Model (MLM) is a training technique used in some Transformer models where certain words in a sentence are hidden (masked), and the model learns to predict the missing words using the surrounding context.

## Causal vs Bidirectional Models

Causal models generate text by predicting the next token using only previous tokens, while bidirectional models understand text using both left and right context of a word.

During training, causal models learn next-token prediction, whereas bidirectional models use Masked Language Modeling (MLM) where some words are masked and predicted.

Causal models are mainly used for text generation tasks such as chatbots, story generation, and code generation, whereas bidirectional models are used for language understanding tasks.

Causal models are commonly used in models like GPT, LLaMA, and Claude, while bidirectional models are used in models like BERT, RoBERTa, and ALBERT.


# Transformer Architecture (End-to-End)


## Transformer

A Transformer is a neural network architecture designed to handle sequential data efficiently using attention mechanisms, without relying on recurrence (RNNs) or convolutions (CNNs).

It allows the model to analyze all tokens in a sequence simultaneously, which makes training faster and helps capture long-range dependencies between words.

The Transformer architecture mainly consists of encoder and decoder layers, which use self-attention and feed-forward neural networks to understand and generate sequences.



## Workflow 




When an input sentence enters a Transformer, it is first tokenized into tokens and converted into token IDs. These IDs pass through the Embedding Layer, which maps each token to a dense vector representation. Since Transformers process tokens in parallel, Positional Encoding is added to the embeddings to inject information about the position of each token in the sequence.

The resulting embeddings are then passed to the Encoder Stack, which contains multiple Encoder Layers. In each encoder layer, the input first goes through Multi-Head Self-Attention. In this step, the model creates Query (Q), Key (K), and Value (V) vectors by multiplying the input embeddings with learned weight matrices. Then, it calculates attention scores using the scaled dot-product attention formula.This process allows each token in the sequence to attend to every other token, enabling the model to capture contextual relationships between words in the sentence.

The output from the Multi-Head Self-Attention layer is added to the original input using a Residual Connection, and then Layer Normalization is applied to make training more stable. The result is then passed to a Position-wise Feed-Forward Network (FFN), which has two linear layers with a non-linear activation such as ReLU or GELU which helps the model learn more complex patterns and improve the representation of each token. After this, another Residual Connection and Layer Normalization step is applied. This set of operations forms one encoder layer, and many such layers are stacked together in the encoder stack to produce richer and more contextual representations of the input tokens.

The encoder output is then passed to the Decoder Stack, which contains multiple decoder layers. Each decoder layer first applies Masked Multi-Head Self-Attention, where the model computes attention while masking future tokens. This masking ensures that during sequence generation, the model can only attend to previously generated tokens and not future ones.

Next, the decoder performs Encoder–Decoder Attention (Cross Attention), where the decoder uses the encoder outputs to gather important information from the input sequence. In this step, the decoder queries attend to the encoder’s keys and values to connect the generated tokens with the original input. The output is then passed through an Add & Layer Normalization step using a residual connection. After that, it goes through a Feed-Forward Network, which helps the model learn more complex patterns.After this, another Residual Connection and Layer Normalization step is applied to complete the decoder layer.

Finally, the decoder output is passed through a Linear layer, which converts it into scores (logits) for each word in the vocabulary. A Softmax layer then converts these scores into a probability distribution. The word with the highest probability is selected as the next token. This token is then fed back into the decoder, and the process repeats to generate the sequence one token at a time, until an end-of-sequence token is produced.

## What is Self-Attention?

Self-attention is a mechanism that allows each token in a sentence to attend every other tokens in the same sentence to understand its context and relationships.

In a Transformer, each token is first converted into an embedding vector $X$. From this embedding, three vectors are created using learned weight matrices:

$$
Q = XW^Q
$$

$$
K = XW^K
$$

$$
V = XW^V
$$

Where:

- **Q (Query)** represents what the token is looking for  
- **K (Key)** represents what each token contains  
- **V (Value)** represents the information of each token  

The model then computes attention scores by comparing the Query of one token with the Keys of all tokens:

$$
score(i, j) = Q_i \cdot K_j^T
$$

These scores are scaled and passed through a softmax function to obtain attention weights:

$$
Attention(Q, K, V) =
softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$

Finally, these weights are used to compute a **weighted sum of the Value vectors**, producing a new representation for each token that captures information from the entire sentence. This allows the model to understand contextual relationships between words.


## Why Scaling is Important

- Without scaling, large dot-product values can cause softmax gradients to be very small, slowing down training.

- Dividing by √(dimension of K) keeps values in a stable range.



## Multi-Head Attention

Multi-Head Attention is a mechanism in Transformers where the model performs multiple self-attention operations in parallel. Each attention head learns to focus on different relationships between words in a sentence. The outputs from all heads are then concatenated and passed through a linear layer to produce the final representation. This helps the model capture richer contextual information.

$$
MultiHead(Q,K,V) = Concat(head_1, ..., head_h)W^O\

$$
$$\text{where } \text{head}_i = \text{Attention}(QW_i^Q, KW_i^K, VW_i^V)$$


## Positional Encoding

Since Transformers process all tokens in parallel, they have no inherent sense of word order. Positional encoding is added to token embeddings to inject information about the position of each token in the sequence.

1️⃣ How It Works

Each token embedding is combined with a positional vector that encodes its position in the sequence.

The standard method uses sinusoidal functions:

$$PE_{(pos, 2i)} = \sin\left(\frac{pos}{10000^{2i/d_{model}}}\right)$$
$$PE_{(pos, 2i+1)} = \cos\left(\frac{pos}{10000^{2i/d_{model}}}\right)$$

| Variable | Definition | Role in the Model |
| :--- | :--- | :--- |
| **$pos$** | **Token Position** | The index of the word (e.g., in "The cat sat," *The*=0, *cat*=1, *sat*=2). It represents the **row** in the encoding matrix. |
| **$i$** | **Dimension Index** | The specific feature index within the embedding (from $0$ to $d_{model}/2$). It determines the **wavelength** of the sine/cosine wave. |
| **$d_{model}$** | **Model Depth** | The total size of the vector representing a word (e.g., 512). This determines the **width** of the encoding vector. |


##  RoPE — Rotary Positional Embeddings

Rotary Positional Embedding (RoPE) is a technique for encoding token positions in transformers by rotating the Query (Q) and Key (K) vectors according to their position.

## problem with Positional Embedding 

Standard positional encodings add absolute position vectors, so models only learn positions seen during training. At inference on longer sequences, unseen positions cause performance to drop.

## how it works 

RoPE rotates the Query and Key vectors according to each token’s position.          
This makes the attention score depend on the relative distance between tokens.          
As a result, it generalizes to longer sequences beyond training length.     


## Layer normalization 

Layer normalization normalizes each token’s hidden vector across features to keep the scale consistent and stabilize training. It computes the mean and variance, then applies learnable scale and shift parameters (γ and β). LayerNorm is applied before or after self-attention and feed-forward layers, helping prevent gradient issues and improving training speed.

## Residual connections

Residual connections (or skip connections) are a technique in deep neural networks where the input of a layer is added directly to its output. In Transformers, they help the model retain information from earlier layers and stabilize training.


## Feed-forward networks

In Transformers, a **Feed-Forward Network (FFN)** is a fully connected neural network applied **independently to each token after the attention layer**. It consists of **two linear layers with a nonlinear activation function (ReLU or GELU) in between**.
$$
FFN(x) = \max(0, xW_1 + b_1)W_2 + b_2
$$

### Explanation

- The **first linear layer** expands the embedding dimension, allowing the model to learn richer features.
- The **activation function** introduces non-linearity.
- The **second linear layer** projects the representation back to the original embedding size.

Unlike attention, the **FFN processes each token independently**, meaning it **does not mix information between tokens**, but instead **refines each token’s representation** before passing it to the next layer.

## Masked Multi-Head Attention

Masked Multi-Head Attention is a self-attention mechanism used in the decoder of a Transformer. It works like normal multi-head attention but adds a mask to block future tokens. This ensures that when generating a sequence, the model only attends to tokens that have already been generated, maintaining the correct autoregressive behavior.


$$
Attention(Q,K,V) =
softmax\left(\frac{QK^T}{\sqrt{d_k}} + M\right)V
$$

Where **M** is the **mask matrix** that blocks attention to future tokens.


## Cross Attention

**Cross Attention** is an attention mechanism used in the **decoder of a Transformer** that allows the decoder to access information from the encoder output. It helps the model connect the **input sequence (encoder)** with the **generated sequence (decoder)**.


In Cross Attention, the **Query (Q)** comes from the decoder, while the **Key (K)** and **Value (V)** come from the encoder outputs.

$$
Q = X_{decoder} W^Q
$$

$$
K = X_{encoder} W^K
$$

$$
V = X_{encoder} W^V
$$

The model then computes **scaled dot-product attention**:

$$
Attention(Q, K, V) =
softmax\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$



## Softmax Function

The **Softmax function** is used in machine learning to convert a set of numbers (called **logits**) into a **probability distribution**. It ensures that all output values are **between 0 and 1** and that their **sum equals 1**.

### Formula

$$
Softmax(z_i) = \frac{e^{z_i}}{\sum_{j=1}^{n} e^{z_j}}
$$

Where:

- \(z_i\) = input value (logit)  
- \(e^{z_i}\) = exponential of the input  
- The denominator is the **sum of exponentials of all inputs**


## Flash Attention

Flash Attention is a memory- and compute-efficient algorithm for transformer models that calculates the attention scores without storing the full attention matrix, allowing transformers to handle longer sequences faster.
### Problem with standard attention
Standard attention computes the full N×N attention matrix where N = sequence length:
- For N = 32K tokens → 32,000 × 32,000 matrix = ~4GB just for attention weights
- Memory complexity: **O(N²)**
- For long contexts this causes OOM (Out of Memory) errors on GPUs

### How Flash Attention solves it
Instead of computing and storing the full attention matrix, Flash Attention divides Q, K, V into small **tiles** and computes attention block by block, keeping only a running summary in memory.

### Key properties
| Property | Standard Attention | Flash Attention |
|----------|--------------------|----------------|
| **Memory** | O(N²) | O(N) |
| **Speed** | Slower (memory bound) | Faster (IO optimized) |
| **Accuracy** | Exact | Exact (mathematically identical) |
| **Long context** | OOM at 32K+ | Handles 128K+ |

### Why it matters for LLMs
- Enables 128K+ context windows that would be impossible with standard attention
- Used in GPT-4, LLaMA 2/3, Mistral, Claude
- Flash Attention 2 & 3 improved further on parallelism

### Simple analogy
> **Standard attention** = Read entire book into RAM, then answer questions.
> **Flash Attention** = Read one chapter at a time, keep notes, answer questions. Same answer, much less memory.


## Grouped Query Attention

Grouped Query Attention (GQA): Queries are grouped to share key and value heads, reducing memory and computation.

## Multi-Query Attention

Multi-Query Attention (MQA): All queries share a single key and value head, improving inference speed and efficiency.

## 2. Mixture of Experts (MoE)

A model architecture where only a subset of specialized “expert” networks are activated for each input, improving efficiency and capacity without increasing computation for every input.

### How it works
1. Input token arrives at an MoE layer
2. A **Router network** scores each expert for that token
3. Top-K experts (usually Top-2) are selected
4. Only those K experts process the token
5. Their outputs are weighted and combined as the final result

### Why it's cheaper than dense models
- A dense 70B model activates **ALL** 70B parameters for every token
- A MoE 140B model with Top-2 routing activates only ~14B parameters per token
- **Same total parameters, much less compute per token**
- This is called **sparse activation**

### Load balancing problem
The router tends to collapse — it keeps sending tokens to the same 1–2 experts, leaving others unused. Fixed by adding an **auxiliary load balancing loss** that penalizes uneven expert usage during training.

### Models that use MoE
- **Mixtral 8x7B** — 8 experts, Top-2 selected per token
- **GPT-4** (rumored to use MoE)
- **Switch Transformer** (Google)

### Trade-offs
| Advantage | Disadvantage |
|-----------|-------------|
| More parameters, same compute | High memory — all experts must fit in GPU |
| Better performance per FLOP | Complex training stability |
| Scales efficiently | Communication overhead in distributed setups |


## Why Transformers Scale Well

Transformers scale well because they process all tokens in a sequence **in parallel**, allowing efficient use of GPUs and TPUs and enabling faster training compared to sequential models like RNNs.

The **self-attention mechanism** lets each token attend to every other token in the sequence, helping the model capture **long-range dependencies and rich contextual relationships**.

Transformers use a **modular architecture** with repeated layers of attention and feed-forward networks, making it easy to scale the model by increasing layers, hidden size, or parameters.

**Multi-head attention** allows the model to learn different types of relationships between tokens simultaneously, improving its ability to capture complex patterns.

Most computations in Transformers rely on **matrix multiplications**, which are highly optimized on modern hardware, enabling efficient training of very large models.

# Token


## Token limit 

Token limit is the maximum number of tokens (input + output) a language model can process in a single request           
On average across normal English text, 1 token ≈ 0.75 words 

## Why Token Limits Exist:

Memory: Models can only remember a limited number of tokens at once.

Computation: More tokens = more calculations, which slows the model down.

Efficiency: Limits prevent the model from crashing or getting too slow.

## Context Window in LLMs

The context window is the maximum number of tokens a language model can “see” or remember at once.

It includes all the input you give plus the output the model generates.

Any text beyond this window is ignored by the model.

## Token Limits vs Context Window

Token Limit: It is the maximum number of tokens a model can process in one request, including input and output.

Context Window: It is the number of tokens the model can “see” or remember at a time while generating output.

Difference: Token limit refers to the overall capacity, while context window refers to the model’s active memory of the text.

Similarity: In most LLMs, the context window size is equal to the token limit.

## Why does token count affect cost in LLMs?

Token count affects cost because each token requires computation and memory to process. More tokens take longer to generate due to self-attention scaling, and most LLM APIs charge per token (input + output). Efficient prompts and chunking can help reduce token usage and cost.

## Context length extension 

allows LLMs to process longer inputs by increasing the number of tokens they can handle. Models like LLaMA achieve 128K+ context using techniques such as RoPE scaling, efficient attention algorithms, KV caching, and training with longer sequences, which reduce memory usage and maintain performance on long texts.

#  Inference & Decoding

# What is Inference

In LLMs, an interface is a mechanism that allows users or applications to interact with the model, sending inputs and receiving outputs without accessing the internal workings.

## What happens during inference?

During inference, the model processes input text (tokens) and generates output based on what it has learned during training without updating its weights.

Process:

Input text is tokenized into words, subwords, or bytes.

Tokens are converted into embeddings, with positional encodings added if using Transformers.

Embeddings pass through model layers (self-attention + feed-forward networks) to compute context-aware representations.

The model predicts next token(s) step by step using a decoding strategy (greedy, beam search, sampling, top-k/top-p).

Tokens are detokenized into human-readable text as the final output.

## what is parameters in LLM?

Parameters in LLMs are the learned weights and biases inside the neural network that store the knowledge gained during training. They determine how input tokens are processed and how predictions are generated. These parameters are learned during training and remain fixed during inference."

Key parameters in LLMs include token embeddings, attention weights (Q, K, V), feed-forward weights and biases, layer normalization parameters, and output projection weights.

## Hyperparameters 
Hyperparameters are settings chosen by developers before training or inference that control how the model learns or generates outputs.

Examples include learning rate, batch size, temperature, top-k, and top-p

## Decoding Strategy

Decoding strategy is how a language model chooses the next token when generating text during inference. It affects creativity, diversity, and accuracy of the output.

## Greedy Decoding

Greedy decoding is a strategy where the model always selects the token with the highest probability at each step during text generation.    
Produces fast and deterministic outputs, but can be repetitive or generic.

## Beam Searh

Beam Search is a decoding strategy where the model tracks multiple high-probability sequences (beams) at each step instead of just the single most likely token.      
Produces more accurate and fluent outputs than greedy decoding, but it is slower and may still miss very creative sequences."

## Sampling 

Sampling is a decoding strategy where the model randomly picks the next token based on its probability distribution, using methods like top-k, top-p, or temperature to generate diverse and creative outputs.

## TOP-K Sampling 

Top-k sampling is a decoding method where the model first selects the top K tokens with the highest probabilities instead of considering the entire vocabulary. The next token is then randomly chosen from these top K options, which allows some randomness while avoiding very unlikely words. This helps the model generate more natural and creative text while still keeping the output meaningful.


## Top-p (nucleus) sampling

Top-p (nucleus) sampling is a decoding strategy where the model selects the smallest set of tokens whose cumulative probability reaches a threshold p (e.g., 0.9). The next token is then randomly chosen from this set, instead of considering all tokens. This allows the model to generate more natural and diverse text while avoiding very unlikely words.

## Temerature 

Temperature is a parameter used during text generation that controls the randomness of the model’s output. A low temperature makes the model choose more likely tokens and produce safer, more predictable text, while a high temperature increases randomness and creativity in the generated text. It adjusts the probability distribution before selecting the next token.

## Repetition Penalty

Repetition penalty is a mechanism during decoding that discourages the model from repeating the same token or phrase multiple times. It is commonly used with sampling methods to make generated text more coherent and natural.


# Training

## LLMs are trained in three main stages.

First, the model undergoes pretraining, where it is trained on a very large dataset of text to learn language patterns by predicting the next token in a sentence.

Second, the model is fine-tuned on smaller, task-specific datasets so it can perform particular tasks such as question answering or summarization.

Finally, many LLMs use reinforcement learning from human feedback (RLHF), where human feedback is used to improve the quality, safety, and alignment of the model’s responses.

## What is Pretraining? 

Pretraining is the process of training a model on a large general dataset to learn basic patterns by predicting the next token in a sentence before fine-tuning it for a specific task.

## What is Self-supervised Learning? 

Self-supervised learning is a training approach where the model learns from unlabeled data by generating its own training signals. In LLMs, the model learns by predicting missing or next tokens in large text datasets. This allows the model to learn language patterns without manual labeling.

## Transfer Learning

Transfer learning is a technique where a pretrained model is reused on a different but related task. Instead of training from scratch, the model transfers knowledge learned from one dataset to improve performance on another.

## Fine-tuning

Fine-tuning is the process of further training a pretrained model on a smaller, task-specific dataset to improve its performance for a particular task. It helps the model adapt its general knowledge to specific applications such as question answering, summarization, or classification

## Full Fine-Tuning 

Full Fine-Tuning is a training method where all the parameters of a pretrained model are updated using a new task-specific dataset. This allows the model to fully adapt its knowledge to the target task. However, it requires high computational resources and memory because the entire model is being trained.

Full fine-tuning has risks such as high computational cost, overfitting on small datasets, and catastrophic forgetting where the model may lose some of its pretrained knowledge.

## Instruction Tuning

Instruction tuning is a process where a pretrained LLM is fine-tuned on datasets of task instructions paired with expected outputs, enabling the model to follow natural language prompts more accurately.

## Supervised Fine-Tuning (SFT)

Supervised Fine-Tuning is a method where a pretrained LLM is further trained on labeled datasets for a specific task using standard supervised learning. The model learns to map inputs to correct outputs.

## Preference Tuning


Preference tuning is a method to align an LLM with human preferences by fine-tuning it using ranked or rated outputs. Unlike full fine-tuning, it focuses on adjusting the model to produce outputs humans prefer, rather than strictly learning task-specific labels.


## RLHF (Reinforcement Learning from Human Feedback) 

RLHF (Reinforcement Learning from Human Feedback) is a training method used to align large language models with human preferences by using human feedback to guide the model toward producing helpful, safe, and accurate responses.

## Workflow
The process starts with a pretrained LLM that already understands language patterns from large text datasets. Next, the model is fine-tuned using supervised fine-tuning (SFT) on instruction–response examples so it can follow prompts better. Then the model generates multiple responses to prompts, and human reviewers rank or score these responses based on quality, safety, and usefulness. This feedback is used to train a reward model that learns which outputs humans prefer. Finally, the original model is optimized using reinforcement learning (commonly PPO) to maximize the reward score, helping the model produce responses that better align with human preferences.

## DPO (Direct Preference Optimization)

DPO is an alignment technique that trains a language model to prefer better responses using human preference data, without using reinforcement learning or a reward model.

First, a set of prompts is collected and the model generates multiple responses for each prompt.
Human annotators compare the responses and select the better one while marking the other as rejected.
These preference pairs are stored as (prompt, chosen response, rejected response).
The model is then trained to increase the probability of the chosen response and decrease the rejected one.
After training, the model produces responses that better align with human preferences.

## What is a reward model?

A reward model is a model used in RLHF that learns from human feedback to score or rank the outputs of an LLM, helping guide the model to generate responses that are more helpful, safe, and aligned with human preferences.

## Fine-tuning VS RAG

Fine-tuning improves an LLM by training it on specific datasets and updating the model’s parameters.

RAG (Retrieval-Augmented Generation) improves responses by retrieving relevant information from external data sources during generation.

Fine-tuning is useful for specializing the model for a specific domain or task, while RAG helps access large or up-to-date knowledge bases.

Fine-tuning changes the model itself, whereas RAG keeps the model unchanged and uses an external retrieval system.

Fine-tuning usually has higher compute cost because it requires model training, while RAG generally has lower compute cost since it mainly retrieves information.

# LLM Optimization & Efficiency


## What is Quantization? 

Quantization is the process of reducing the numerical precision of model weights and activations (for example from FP32 or FP16 to INT8 or 4-bit) to reduce memory usage and speed up inference while trying to maintain similar model performance.

The main risk is potential accuracy loss because lower precision can reduce the model’s ability to represent detailed weight information.

## Model Distillation

Model distillation is a technique where a smaller, simpler model (student) learns to mimic the behavior of a larger, more complex model (teacher). The goal is to get a lighter model that performs almost as well as the original but is faster and easier to deploy.

Risks include potential accuracy loss and some knowledge not transferring fully.

## Pruning 

Pruning is a model optimization technique that removes less important weights, neurons, or connections from a neural network. The goal is to make the model smaller, faster, and more efficient without significantly affecting accuracy. It works by identifying parameters that contribute little to predictions and eliminating them, thereby reducing memory usage and computational cost.

risks include accuracy loss, hardware inefficiency for unstructured pruning, the need for careful tuning, and extra fine-tuning to maintain performance

## Mixed precision training

Mixed precision training combines 16-bit and 32-bit computations to make model training faster and more memory-efficient. Most calculations use FP16, while weights and sensitive operations stay in FP32 for stability. Loss scaling prevents gradient underflow, allowing larger batch sizes and faster training without losing accuracy.

## PEFT

Parameter-Efficient Fine-Tuning (PEFT) helps prevent catastrophic
forgetting in LLMs by updating only a small set of task-specific
parameters, while keeping most of the model's original parameters
frozen. This approach allows the model to adapt to new tasks without
overwriting previously learned knowledge, ensuring it retains core
capabilities while learning new information efficiently.


the risks are that it may provide limited adaptation for highly specialized tasks, can be sensitive to hyperparameters, and may sometimes yield slightly lower accuracy compared to full fine-tuning.

## What is LoRA?

LoRA (Low-Rank Adaptation) is a parameter-efficient fine-tuning (PEFT) method used to adapt large pre-trained models, especially large language models (LLMs), to specific tasks without updating all model parameters.


This drastically reduces the number of trainable parameters, making fine-tuning faster, memory-efficient, and feasible on smaller hardware.

## work flow 


LoRA starts with a pre-trained model and freezes all original weights W to preserve prior knowledge. Introduces small trainable low-rank matrices A and B at selected layers. The updated weights become:

\[
W' = W + A.B
\]

Fine-tuning is performed only on A and B, keeping W frozen. During training, the model combines frozen weights with LoRA updates, allowing efficient task-specific adaptation without updating the full model. At deployment, the model uses frozen original weights plus the trained LoRA matrices, enabling memory-efficient and high-performance adaptation.

## Risk 
LoRA may provide limited adaptation for highly complex tasks, and its performance is sensitive to hyperparameters and layer selection.


## QLoRA

QLoRA (Quantized Low-Rank Adaptation) is an extension of LoRA that combines low-rank adaptation with model quantization to reduce memory usage. The pre-trained model is first quantized to 4-bit or 8-bit precision, and LoRA modules (A and B) are added to selected layers. During training, only the LoRA matrices are fine-tuned while the quantized base weights remain frozen. This allows efficient fine-tuning of very large models on limited hardware while maintaining performance. The final model combines frozen quantized weights with trained LoRA matrices for task-spec

# LoRA vs QLoRA Differences

1. **Model Precision:** LoRA uses full precision (FP16/FP32) weights, while QLoRA uses quantized 4-bit or 8-bit weights.
2. **Memory Efficiency:** LoRA is moderately memory-efficient, whereas QLoRA is highly memory-efficient and can handle very large models on limited hardware.
3. **Use Case:** LoRA is ideal for fine-tuning moderately sized LLMs, while QLoRA is designed for very large LLMs with limited resources, trading a small amount of accuracy for efficiency.
4. **Hardware Requirements:** LoRA needs GPUs with decent memory, while QLoRA can run on smaller or less powerful GPUs.
5. **Performance Trade-off:** LoRA generally preserves accuracy, whereas QLoRA may have slight accuracy loss due to quantization.



## Catastrophic Forgetting

### What it is
When a pretrained LLM is fine-tuned on new task-specific data, it overwrites the weights that stored pretrained knowledge, causing the model to "forget" what it learned during pretraining.

### Why it happens
Full fine-tuning changes all model weights, and learning a new task can make the model forget what it learned before.

### Example
> A GPT model pretrained on general text, then fine-tuned on medical Q&A → may forget how to do general reasoning, write code, or answer non-medical questions.

### How LoRA/PEFT prevent it
- Original pretrained weights `W` are **frozen** — never modified
- Only small adapter matrices `A` and `B` are trained
- Pretrained knowledge is preserved in `W`, new task knowledge in `A·B`
- Result: model retains general ability + gains task-specific ability

### Other Mitigation Strategies
| Strategy | How it works |
|----------|-------------|
| **Elastic Weight Consolidation (EWC)** | Adds a penalty to the loss that discourages changing weights important for previous tasks |
| **Replay methods** | Mix old training data with new task data during fine-tuning |
| **Lower learning rate** | Smaller updates reduce how aggressively old weights are overwritten |


# Latency vs Throughput

## Definition

* **Latency:** The time it takes for a model to generate a response from the moment a prompt is received.
* **Throughput:** The number of requests or tokens the model can process per unit of time.

## 4 Major Differences

1. **Measurement:** Latency measures the time per individual request, while throughput measures the total requests or tokens processed per unit time.
2. **Focus:** Latency focuses on response speed, whereas throughput focuses on processing capacity.
3. **Critical Use Case:** Latency is crucial for real-time, interactive applications, while throughput is key for batch or large-scale processing.
4. **Optimization Trade-off:** Reducing latency may decrease throughput, while increasing throughput may increase latency.


## Batching


Batching is the process of grouping multiple inputs (prompts) together and processing them simultaneously through an LLM. This improves computational efficiency and throughput while making better use of hardware resources.

## KV Caching (Key-Value Caching)

### What it is
KV Caching stores the Key (K) and Value (V) matrices computed from previous tokens so the model doesn't recompute them at every generation step.

### Why it's needed
In autoregressive generation, the model generates one token at a time. Without caching, at each step it recomputes attention for ALL previous tokens from scratch — extremely wasteful.

### How it works step by step
| Step | What happens |
|------|-------------|
| Step 1 | Token 1 processed → K1, V1 computed and **cached** |
| Step 2 | Token 2 processed → K2, V2 computed and cached. K1, V1 **reused from cache** |
| Step 3 | Token 3 → only new K3, V3 computed. Previous K1,V1,K2,V2 reused |

- **Without cache:** Recompute everything every step → O(N²) cost
- **With cache:** Only compute new token's K, V → O(N) cost

### Memory cost
KV cache size grows with sequence length and batch size. For long contexts (128K tokens) it can consume GBs of GPU memory — this is why Flash Attention and GQA/MQA exist, to reduce this cost.

### Trade-off
> Speed ↑ but GPU memory ↑. Larger batch size = larger cache = risk of OOM.



# Cost Optimization Strategies

Cost optimization in LLM systems focuses on reducing training, inference, infrastructure, and operational costs while maintaining performance, latency, and quality.

## Most Important Points

1. **Model-Level Optimization:** Use quantization (INT8/4-bit), distillation, pruning, and mixed precision to reduce memory, GPU usage, and compute costs.
2. **Inference Optimization:** Apply KV caching, batching, efficient decoding, and shorter prompts to improve latency, throughput, and token efficiency.
3. **Architectural Decisions:** Prefer smaller specialized models and retrieval-based approaches (RAG) instead of retraining large models.
4. **Infrastructure Optimization:** Use autoscaling, spot/reserved instances, GPU monitoring, and model parallelism to optimize resources and costs.
5. **Token & API Cost Optimization:** Minimize prompt size, compress conversation history, and avoid unnecessary token usage to reduce API costs.


## who do you Reducing Inference Latency 

Inference latency can be reduced by optimizing both the model and system. Key strategies include using KV caching to avoid recomputation, employing smaller or quantized models, applying efficient decoding methods, shortening prompts, batching requests carefully, utilizing high-performance hardware, and distributing large models via parallelism or sharding. These approaches help deliver faster responses while maintaining model performance.

# Alignment & Safety

## What is Alignment?
Alignment in LLMs refers to the process of ensuring that the model’s outputs match human intentions, values, and expectations. A well-aligned model produces useful, safe, and contextually appropriate responses.


## Constitutional AI

Constitutional AI (Anthropic) is an alignment method that uses a written "constitution" — a set of principles — to train the model to be helpful, harmless, and honest, reducing reliance on expensive human feedback.

### Two phases

#### Phase 1 — SL-CAI (Supervised Learning)
1. Model generates a response to a harmful prompt
2. Model is asked to **critique its own response** against the constitution (e.g., *"Is this response harmful?"*)
3. Model **revises** its response based on the critique
4. The revised, improved response is used as supervised training data
5. This is called **self-critique and revision** — the model improves itself

#### Phase 2 — RL-CAI (Reinforcement Learning)
1. Model generates pairs of responses
2. Instead of human raters, an **AI model** (trained on the constitution) scores which response is better
3. This AI feedback is used to train a preference model
4. The preference model is used like a reward model in standard RLHF
5. Final model is trained with RL against this AI preference model

### Key difference from RLHF
| Aspect | RLHF | Constitutional AI |
|--------|------|-------------------|
| Feedback source | Human raters | AI model guided by constitution |
| Cost | Very expensive | Much cheaper |
| Scalability | Limited by human bandwidth | Scales automatically |
| Transparency | Implicit human preferences | Explicit written principles |

### Why it matters
Reduces human labeling cost, makes alignment principles explicit and auditable, and allows the model to self-improve through critique.


## Toxicity & Safety Controls in LLMs


Toxicity in LLMs refers to outputs that are offensive, harmful, or unsafe, including hate speech, harassment, misinformation, or inappropriate content. Safety controls are techniques to prevent or minimize toxic outputs, ensuring models are safe and reliable for users.

How Toxicity Arises

Toxicity often originates from biased, unfiltered, or harmful content in training data. LLMs, being statistical learners, can replicate these patterns. Fine-tuning, instruction tuning, or reinforcement learning without careful oversight can also amplify toxicity.

## Guardrails in LLMs

Guardrails are predefined rules, constraints, or safety mechanisms applied to LLMs to ensure outputs are safe, ethical, and aligned with user expectations. They act as preventive layers that restrict harmful, toxic, or undesired behavior.

# Multimodal 

Multimodal LLMs are large language models that can process and generate information from multiple data types (e.g., text, images, audio, video) by encoding each modality, fusing the representations, and producing unified outputs, typically text. They enable tasks like visual question answering, image captioning, and multimodal reasoning.

## Open-Source LLMs vs OpenAI Models
Open-source models (e.g., LLaMA-based variants, Mistral, Falcon) provide access to model weights and allow full customization and on-prem deployment.
Closed models (e.g., OpenAI GPT models) are accessed via API and are managed, optimized, and maintained by the provider.

## Trade-offs
Open-source offers flexibility, cost control at scale, and transparency but requires infrastructure and optimization effort.
Closed models offer high performance, reliability, and managed scaling but with limited customization and recurring API costs.

## Fine-Tuning Flexibility
Open-source models allow full fine-tuning (LoRA, QLoRA, full fine-tune) and architecture modifications.
Closed models typically allow limited fine-tuning or instruction tuning via API but not weight-level control.

## Data Privacy Concerns
Open-source models can run on-prem or in private VPCs, ensuring sensitive data never leaves the organization.
Closed models require sending data to external APIs (though providers offer enterprise privacy guarantees).

## Licensing Issues
Open-source models may have restrictive licenses (e.g., non-commercial clauses or usage limits).
Closed models operate under commercial API agreements with defined usage policies.

## Model Size Selection
Open-source models require careful selection based on GPU memory and latency constraints (7B, 13B, 70B, etc.).
Closed models abstract this complexity; users choose tiers (e.g., standard vs advanced models) without managing infrastructure.
