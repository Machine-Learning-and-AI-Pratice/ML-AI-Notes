# Tokens & Tokenization

## What Are Tokens?


Tokens are the smallest units of text that a language model can understand. They can be words, subwords, characters, or symbols, depending on the tokenization method used.

1️⃣ How Tokens Work

- Sentences or documents are split into tokens by a tokenizer.

- Each token is then converted into a numerical representation (embedding) for processing by a model.

Example: "I love AI" → ["I", "love", "AI"] (word-level) or ["I", "lov", "e", "A", "I"] (subword-level).

2️⃣ Why Tokens Are Important

- ⚡ Model Input: Tokens are the basic units fed into NLP models.

- 🧠 Contextual Understanding: Each token can carry meaning and interact with other tokens via attention.

- 🎯 Flexibility: Subword tokens handle unknown words or rare vocabulary efficiently.

3️⃣ Analogy

- Tokens are like LEGO blocks:

- Each block is a unit of meaning.

- Combining blocks builds words, sentences, and paragraphs that the model can understand.

4️⃣ Token Types

- Word-level: Each word is a token.

- Subword-level: Frequent subword pieces (Byte-Pair Encoding, SentencePiece).

- Character-level: Each character is a token, useful for highly flexible or multilingual models.

## Word vs Subword vs Byte Tokenization

Word tokenization splits text by spaces or punctuation into full words. Subword tokenization breaks words into smaller, frequent units to handle rare or unknown words efficiently. Byte tokenization splits text into bytes, making it fully language-agnostic and capable of handling any symbol.

Word tokenization is simple and intuitive, but struggles with rare words and large vocabularies. Subword tokenization balances vocabulary size and flexibility, while byte tokenization handles all languages and symbols without unknown tokens.

Word tokenization is widely used in simple NLP tasks and search engines. Subword tokenization is preferred in most modern LLMs for robust handling of unknown words. Byte tokenization is used for multilingual and highly flexible models.

Word tokenization requires a smaller computational overhead but is less robust. Subword tokenization increases sequence length slightly but improves coverage. Byte tokenization produces long sequences but ensures no token is ever unknown.

Word tokenization is easy to implement and fast. Subword tokenization supports models like BERT and GPT for large-scale language understanding. Byte tokenization supports models like GPT-NeoX and multilingual Transformers for universal text handling.

## BPE vs WordPiece vs SentencePiece

BPE (Byte-Pair Encoding) merges the most frequent pairs of characters or subwords to create a compact vocabulary. WordPiece merges subwords based on maximizing likelihood in the training corpus. SentencePiece is a language-agnostic tokenizer that can operate at the byte or character level and doesn’t require pre-tokenization.

BPE is simple and frequency-based, WordPiece is probabilistic and more accurate for rare words, while SentencePiece handles any language and removes dependency on whitespace.

BPE is used in models like GPT-2 and GPT-3. WordPiece is widely used in BERT and RoBERTa. SentencePiece is used in T5, mBART, and multilingual models.

BPE requires pre-tokenization and may struggle with languages without spaces. WordPiece is slower to train but produces more consistent subwords. SentencePiece is fully language-agnostic and can handle multilingual and unseen scripts efficiently.

BPE is fast and simple for English or space-separated languages. WordPiece improves accuracy for rare words and model performance. SentencePiece provides universal tokenization for multilingual or byte-level text processing.

## Token Limits vs Context Window

Token limits define the maximum number of tokens a model can process in a single input, while the context window refers to the span of text the model can “see” at once for understanding and generating outputs.

Token limits are set by model architecture and memory constraints. Context windows determine how far back the model can attend to previous tokens to capture dependencies.

Exceeding the token limit may result in truncated input or errors. A larger context window allows the model to handle longer dependencies, improving coherence in text generation.

Models with small token limits are faster and require less memory but struggle with long documents. Models with large context windows are more capable of reasoning over extended text but require more computational resources.

Token limits and context windows are crucial in tasks like summarization, question-answering, or dialogue, where retaining long-term information improves performance.

## Tokenization Issues (OOV Words)

OOV (Out-of-Vocabulary) words are words not present in the model’s vocabulary. Tokenization can struggle with these words, which affects embeddings and model performance.

1️⃣ How It Happens

- Word-level tokenizers cannot recognize rare or new words.

- Misspellings, slang, or foreign words may not exist in the vocabulary.

- Example: "cryptomarket" → OOV if only "crypto" and "market" exist separately.

2️⃣ Effects on Models

- OOV words may be mapped to an “unknown” token, losing semantic meaning.

- Embeddings for these words are unavailable → poor understanding or prediction.

- Reduces the effectiveness of NLP tasks like classification, translation, or question-answering.

3️⃣ Solutions

- Subword Tokenization: Break words into smaller units (BPE, WordPiece, SentencePiece) → reduces OOV occurrences.

- Byte-Level Tokenization: Works at byte/character level → no unknown tokens.

- Spell Correction / Normalization: Fixes misspelled words before tokenization.

4️⃣ Pros & Cons

- Word-level tokenization:

  - ✅ Simple, fast

  - ⚠️ High OOV rate for rare or new words

- Subword/Byte tokenization:

  - ✅ Reduces OOV words, better coverage

  - ⚠️ Slightly longer sequences, more computation

## Why token count affects cost

In LLMs, tokens are the basic units of text processed by the model. The number of tokens directly impacts compute, memory, and API usage, which in turn determines cost.

1️⃣ Compute Requirements

- Each token passes through multiple Transformer layers, each performing attention and feed-forward computations.

- More tokens → more matrix multiplications and operations → higher compute.

2️⃣ Memory Usage

- Attention mechanism stores intermediate activations for each token.

- Longer sequences (more tokens) require more GPU/TPU memory.

3️⃣ Inference Time

- More tokens → slower processing, as self-attention scales roughly quadratically with sequence length.

- Longer texts take more time to generate predictions → higher operational cost.

4️⃣ API Pricing

- Most LLM services (OpenAI, Hugging Face, etc.) charge per token:

- Input tokens (prompt) + output tokens (generated text) → total tokens billed.

- Larger token counts → higher cost.

## Tokenization Mismatch Problems


Tokenization mismatch occurs when the tokenization strategy used at training differs from the one used at inference, or when different tokenizers split the same text differently. This can lead to misalignment, OOV words, or poor model performance.

1️⃣ How It Happens

- Training used subword tokenization (e.g., BPE), but inference uses a different tokenizer.

- Special characters, punctuation, or whitespace handled differently.

Example: "AI-powered" →

- Training tokenizer: ["AI", "-", "powered"]

- Inference tokenizer: ["AI", "power", "ed"]

2️⃣ Effects on Models

- Token embeddings may not align → degraded semantic understanding.

- Output probabilities may be inaccurate.

- Model may generate broken or unexpected tokens, affecting quality of generated text.

3️⃣ Common Causes

- Using different tokenizers between pretraining and fine-tuning.

- Handling of punctuation, emojis, or whitespace inconsistently.

- Changes in subword vocabulary across versions.

4️⃣ Solutions

- Always use the same tokenizer for training and inference.

- Preprocess text consistently (e.g., lowercasing, punctuation handling).

- For multilingual models, use language-agnostic tokenizers like SentencePiece or byte-level tokenization.
