# Multimodal LLMs

What is a Multimodal LLM?

A multimodal LLM is a large language model that can understand and process multiple types of data (modalities) such as text, images, audio, or video — instead of only text.

Traditional LLMs work purely on text input and text output. Multimodal LLMs extend this by converting other data types (like images or speech) into representations that the model can process alongside text.

How It Works (Technical View)

- Input Encoding

    - Text → Token embeddings

    - Image → Processed through a Vision Encoder (like a CNN or Vision Transformer)

    - Audio → Processed through an audio encoder

- Shared Representation Space
    - All modalities are converted into vector embeddings and mapped into a shared embedding space.

- Fusion Mechanism
    - The model combines these embeddings using attention mechanisms so it can reason across modalities.

- Generation
    - The model produces text output (or sometimes images/audio in advanced systems).

Example Use Cases

- Image captioning

- Visual question answering

- Document understanding (text + layout + images)

- Voice assistants

- Video summarization

## Text + Image Models (Vision-Language Models)

Text + Image models are multimodal systems that can understand and reason over both visual and textual inputs together.

Unlike traditional LLMs (text-only), these models combine:

- A Vision Encoder (to process images)

- A Language Model (to process and generate text)

- A Fusion mechanism (usually attention-based)

🔎 How They Work (Architecture Flow)

1️⃣ Image Encoding
The input image is passed through a vision encoder (like a Vision Transformer or CNN).
It converts the image into a sequence of visual embeddings (vector representations).

2️⃣ Text Encoding
Text input is tokenized and converted into embeddings using the language model’s embedding layer.

3️⃣ Cross-Modal Fusion
The visual embeddings and text embeddings are combined.
This is typically done using:

Cross-attention layers

Or projecting image embeddings into the LLM’s embedding space

4️⃣ Reasoning & Generation
The transformer processes both modalities together and generates a text output.

🧠 What Makes Them Powerful?

Text + image models can:

- Answer questions about images

- Generate image captions

- Understand charts and documents

- Perform visual reasoning

- Follow instructions grounded in images

They align visual and textual information into a shared representation space.

🆚 Compared to Text-Only LLMs

- Text-only LLMs operate purely on token sequences.
- Text + image models integrate visual features, allowing grounded reasoning beyond language patterns.

## Vision Transformers (ViT)

Vision Transformers (ViT) apply the Transformer architecture — originally designed for NLP — directly to images by treating an image as a sequence of patches instead of pixels.

🧠 Core Idea

Instead of using convolutional filters (like CNNs), ViT:

- Splits an image into fixed-size patches

- Converts each patch into an embedding (like a token)

- Processes the sequence of patch embeddings using self-attention

This allows the model to capture global relationships across the entire image.

🔄 How Vision Transformer Works (Flow)
1️⃣ Patch Extraction

An image is divided into small patches (e.g., 16×16).
Each patch is flattened into a vector.

2️⃣ Patch Embedding

Each flattened patch is passed through a linear layer to create a dense embedding.

A special [CLS] token is added to represent the whole image.

3️⃣ Positional Encoding

Since Transformers don’t understand spatial order, positional embeddings are added to retain patch location information.

4️⃣ Transformer Encoder

The sequence of patch embeddings passes through stacked Transformer encoder layers:

Multi-head self-attention (captures global context)

Feed-forward networks

Residual connections + layer normalization

5️⃣ Classification Head

The final representation of the [CLS] token is passed to a linear layer for classification.

🆚 ViT vs CNN

- ViT captures global dependencies directly using attention.
CNN captures local spatial patterns using convolution filters.

- ViT scales better with large datasets.
CNN works better on smaller datasets without heavy pretraining.

✅ Why Vision Transformers Are Powerful

- Capture long-range spatial relationships

- Highly parallelizable

- Scale well with data and compute

- Used in many multimodal and large vision-language models

🎯 Interview-Ready Definition

“A Vision Transformer is a model that processes images as sequences of patches and applies transformer self-attention to capture global spatial relationships, replacing convolutional operations used in CNNs.”

## Multimodal Tokenization

Multimodal tokenization is the process of converting different data types (text, images, audio, video) into a unified representation that a transformer model can process.

In text, tokenization splits sentences into words or subwords.
In images, the image is divided into patches (like in Vision Transformers), and each patch is converted into a visual embedding (image tokens).
In audio, the waveform is converted into spectrogram features and then embedded as audio tokens.

All these modality-specific tokens are projected into a shared embedding space, allowing the transformer to apply attention across modalities.

In text + image models, image tokens are often inserted before or between text tokens so the model can jointly reason over both.

## Use Cases of Multimodal Models

Multimodal models enable reasoning beyond pure text.

They are used in:

- Visual Question Answering (answer questions about an image)

- Image captioning

- Document understanding (text + layout + figures)

- Medical imaging analysis with reports

- Video summarization

- Autonomous driving perception systems

- Voice assistants combining speech + context

They are especially powerful where context from multiple data types improves decision-making.

## Challenges in Multimodal Models

Multimodal systems face several technical challenges.

First, alignment problem — ensuring text and image embeddings correspond correctly in shared space.

Second, data scarcity — high-quality paired multimodal datasets are expensive and limited.

Third, modality imbalance — one modality (often text) can dominate learning.

Fourth, high computational cost — processing images/videos along with text significantly increases memory and training cost.

Fifth, hallucination & grounding issues — the model may generate text not actually supported by visual input.

Sixth, evaluation complexity — measuring cross-modal reasoning quality is harder than text-only evaluation.
