# Deep Learning – Most Important Interview Questions & Answers

## Deep Learning Fundamentals

### What is Deep Learning?
Deep Learning is a subset of machine learning that uses **neural networks with multiple layers** to automatically learn complex patterns and representations from large amounts of data.

### How is Deep Learning different from Machine Learning?
Machine Learning often requires **manual feature engineering**, while Deep Learning **automatically learns features using deep neural networks** and usually requires larger datasets and more computational power (GPUs).

### What is a Neural Network?
A Neural Network is a computational model inspired by the human brain, consisting of **layers of interconnected neurons that process input data and learn patterns through weights and activation functions**.

### What is a Perceptron?
A Perceptron is the **simplest neural network model**, consisting of a single neuron that takes input features, multiplies them by weights, adds bias, and applies an activation function to produce an output.

### What are Layers in a Neural Network?
Neural networks consist of three main types of layers:
- **Input Layer** – receives input data
- **Hidden Layers** – perform computations and feature extraction
- **Output Layer** – produces the final prediction

### What are Weights and Biases?
- **Weights** control the importance of input features.
- **Bias** shifts the output to help the model fit data better.

### What is an Activation Function?
An activation function introduces **non-linearity into the neural network**, allowing it to learn complex relationships.

Common activation functions:
- ReLU
- Sigmoid
- Tanh
- Softmax

### Why do we need Non-Linear Activation Functions?
Without non-linear activation functions, the neural network would behave like a **linear model**, regardless of the number of layers, limiting its ability to learn complex patterns.

### What is Forward Propagation?
Forward propagation is the process where **input data passes through the network layers to produce predictions**.

### What is Backpropagation?
Backpropagation is the process of **calculating gradients of the loss function and updating weights using gradient descent to minimize prediction error**.

# Training & Optimization

### What is Gradient Descent?
Gradient Descent is an optimization algorithm used to **minimize the loss function by updating model parameters in the direction of the negative gradient**.

### What is Stochastic Gradient Descent (SGD)?
SGD updates model weights **using one training example at a time**, which makes training faster but noisier.

### What is Mini-Batch Gradient Descent?
Mini-batch gradient descent updates weights using **small batches of training data**, balancing stability and speed.

### What is Learning Rate?
The learning rate is a **hyperparameter that controls the size of weight updates during training**.

### What happens if the Learning Rate is too high or too low?

**Too High**
- Model may overshoot minimum
- Training becomes unstable

**Too Low**
- Training becomes very slow
- Model may get stuck in local minima

### What is the Vanishing Gradient Problem?
In deep networks, gradients become **very small during backpropagation**, preventing early layers from learning effectively.

### What is the Exploding Gradient Problem?
Gradients become **very large during backpropagation**, causing unstable training and large weight updates.

### What is Weight Initialization?
Weight initialization is the process of **setting initial values of neural network weights before training**, which affects convergence speed and training stability.

### What is Regularization in Deep Learning?
Regularization techniques reduce **overfitting by limiting model complexity**.

Examples:
- L1 / L2 Regularization
- Dropout
- Early stopping

### What is Dropout?
Dropout randomly **disables a percentage of neurons during training**, forcing the network to learn more robust features and preventing overfitting.

### What is Batch Normalization?
Batch normalization normalizes layer inputs during training to **improve training stability and accelerate convergence**.

### What is Early Stopping?
Early stopping stops training when **validation loss stops improving**, preventing overfitting.

# Neural Network Architectures

### What is a Feedforward Neural Network?
A feedforward neural network is a basic neural network where **data flows in one direction from input to output without cycles**.

### What is a Convolutional Neural Network (CNN)?
CNN is a deep learning architecture designed for **image and spatial data processing**, using convolution layers to automatically learn spatial features.

### What is Pooling in CNN?
Pooling reduces the **spatial dimensions of feature maps**, decreasing computation and helping extract dominant features.

Common types:
- Max Pooling
- Average Pooling

### What is Padding and Stride?

**Padding**
- Adds extra pixels around input images
- Prevents shrinking of feature maps

**Stride**
- Controls how much the convolution filter moves across the image

### What is a Recurrent Neural Network (RNN)?
RNN is a neural network designed for **sequential data**, where outputs depend on previous inputs.

Examples:
- Text
- Speech
- Time series

### What is LSTM?
LSTM (Long Short-Term Memory) is a type of RNN designed to **solve the vanishing gradient problem and capture long-term dependencies in sequence data**.

### What is GRU?
GRU (Gated Recurrent Unit) is a **simplified version of LSTM** with fewer parameters but similar performance for sequence modeling.

### What is a Sequence-to-Sequence Model?
A Seq2Seq model uses **an encoder-decoder architecture to transform one sequence into another**.

Examples:
- Machine translation
- Text summarization

### What is an Autoencoder?
An autoencoder is a neural network used for **unsupervised learning that compresses input data into a lower-dimensional representation and reconstructs it**.

### What is a GAN?
GAN (Generative Adversarial Network) consists of **two neural networks: a generator and a discriminator that compete with each other to generate realistic data**.

Generator → Creates fake data  
Discriminator → Detects real vs fake data

Applications:
- Image generation
- Deepfakes
- Data augmentation
- Image super-resolution

**Definition → How it works → Example → Advantage**

This structure makes answers **clear, technical, and interview-ready**.
