# Machine Learning

## What is Machine Learning?

Machine Learning (ML) is a branch of Artificial Intelligence that enables computers to learn patterns and make predictions from data without being explicitly programmed for every task.

ML models identify relationships in data and improve their performance automatically as they see more data.

## What is the difference between AI, ML, and Deep Learning?

Here’s a clear distinction between AI, ML, and Deep Learning:

Artificial Intelligence (AI):

The broad field of making machines intelligent—able to perform tasks that normally require human intelligence.

Includes rule-based systems, robotics, NLP, computer vision, and ML.

Machine Learning (ML):

A subset of AI where machines learn patterns from data without explicit programming.

Focuses on training models to make predictions or decisions.

Deep Learning (DL):

A subset of ML that uses deep neural networks with multiple layers to learn complex patterns.

Excels at tasks like image recognition, speech recognition, and natural language processing.

💡 Key idea:
AI ⊃ ML ⊃ DL — AI is the broad goal, ML is learning from data, and DL is using deep neural networks to achieve ML tasks.

## What is training data and test data?

Training Data: The portion of the dataset used to train the machine learning model.

The model learns patterns, relationships, and parameters from this data.

Test Data: The portion of the dataset kept separate from training to evaluate the model’s performance.

Helps check how well the model generalizes to unseen data.

💡 Key idea: Training data is for learning, and test data is for evaluation.

## What is the bias–variance tradeoff?

The bias–variance tradeoff is a fundamental concept in machine learning that describes the balance between underfitting and overfitting:

Bias: Error due to overly simple models that cannot capture underlying patterns (underfitting).

Variance: Error due to overly complex models that fit noise in the training data (overfitting).

💡 Tradeoff:

High bias → low variance: Model is simple, stable, but inaccurate.

High variance → low bias: Model is complex, accurate on training data, but unstable on new data.

Goal: Find the optimal balance where the model generalizes well to unseen data.

## What is overfitting and underfitting?

Overfitting: Occurs when a model learns the training data too well, including noise or irrelevant patterns.

Performs very well on training data but poorly on unseen/test data.

Fixes: Regularization, dropout, more training data.

Underfitting: Occurs when a model is too simple to capture patterns in the data.

Performs poorly on both training and test data.

Fixes: Increase model complexity, add features, train longer.

💡 Key idea:

Overfitting = memorizes training data

Underfitting = fails to learn patterns

## What is cross-validation?'

Cross-validation is a technique used to evaluate a machine learning model’s performance more reliably by testing it on multiple subsets of data.

The dataset is split into k folds (commonly 5 or 10).

The model is trained on k−1 folds and tested on the remaining fold.

This process is repeated k times, each time using a different fold as the test set.

The final performance is the average across all folds, reducing bias from a single train-test split.

💡 Key idea: Cross-validation provides a robust estimate of how well the model generalizes to unseen data.

## What is feature engineering?

Feature engineering is the process of creating, transforming, or selecting input variables (features) from raw data to improve a machine learning model’s performance.

It involves extracting meaningful information from raw data, such as converting timestamps to day/month, encoding categories, or normalizing numerical values.

Good feature engineering can make patterns more apparent for the model, improving accuracy and reducing training time.

Techniques include scaling, encoding, aggregation, dimensionality reduction, and creating interaction features.

💡 Key idea: Feature engineering turns raw data into meaningful inputs that help models learn better.

## What is feature selection?

Feature selection is the process of choosing the most relevant input features from a dataset to use in a machine learning model.

The goal is to remove irrelevant, redundant, or noisy features that do not help the model learn.

Benefits:

Improves model performance and accuracy

Reduces overfitting

Decreases training time and complexity

Methods:

Filter methods: Select features based on statistical measures (e.g., correlation, chi-square).

Wrapper methods: Evaluate subsets of features using model performance (e.g., forward/backward selection).

Embedded methods: Feature selection occurs during model training (e.g., Lasso regression).

💡 Key idea: Feature selection helps the model focus on the most important variables and ignore irrelevant ones.

## What is data preprocessing?

Data preprocessing is the process of cleaning and transforming raw data into a format suitable for machine learning models.

It ensures that data is consistent, complete, and properly formatted for training.

Common steps include:

Handling missing values (e.g., imputation)

Encoding categorical variables (e.g., one-hot encoding)

Scaling or normalizing numerical features

Removing duplicates or outliers

Splitting data into training, validation, and test sets

💡 Key idea: Proper data preprocessing improves model performance, reduces errors, and speeds up training.

# Supervised Learning

## What is supervised learning?

Supervised learning is a type of machine learning where the model learns from labeled data—data that includes both inputs and the correct outputs.

The model uses this data to learn the relationship between inputs and outputs.

After training, it can predict outputs for new, unseen inputs.

Common tasks:

Regression: Predicting continuous values (e.g., house prices).

Classification: Predicting categories or labels (e.g., spam detection, image recognition).

## What is the difference between classification and regression?

Classification:

Predicts discrete categories or labels.

Example: Email → Spam or Not Spam, Image → Cat, Dog, or Bird.

Output is categorical.

Regression:

Predicts continuous numeric values.

Example: Predicting house prices, temperature, or stock prices.

Output is real-valued.

## What are common classification algorithms?

Common classification algorithms in machine learning include:

Logistic Regression: Predicts probabilities for binary or multi-class outcomes.

Decision Trees: Splits data based on features to classify samples.

Random Forest: Ensemble of decision trees for more robust classification.

Support Vector Machines (SVM): Finds the optimal hyperplane that separates classes.

K-Nearest Neighbors (KNN): Classifies based on the majority class of nearest neighbors.

Naive Bayes: Probabilistic classifier based on Bayes’ theorem, assumes feature independence.

Gradient Boosting (XGBoost, LightGBM, CatBoost): Combines weak learners sequentially to improve accuracy.

Neural Networks: Deep learning models capable of handling complex, high-dimensional data for classification tasks.

## What are common regression algorithms?

Common regression algorithms in machine learning include:

Linear Regression: Models the relationship between input features and a continuous output using a straight line.

Polynomial Regression: Extends linear regression by fitting a polynomial curve to capture non-linear relationships.

Ridge and Lasso Regression: Linear models with L2 (Ridge) or L1 (Lasso) regularization to prevent overfitting.

Decision Tree Regression: Uses a tree structure to predict continuous values by splitting data into regions.

Random Forest Regression: Ensemble of decision trees to improve prediction accuracy and reduce variance.

Support Vector Regression (SVR): Uses support vector concepts to predict continuous values with a margin of tolerance.

Gradient Boosting Regression (XGBoost, LightGBM, CatBoost): Sequentially builds weak learners to minimize error and improve performance.

Neural Networks: Deep learning models for complex regression tasks, especially with large or high-dimensional data.

## What is linear regression?

Linear regression is a supervised machine learning algorithm used to predict a continuous numeric value based on one or more input features.

It assumes a linear relationship between the input(s) and the output.

The model finds the best-fitting line (y = mx + b for one feature) that minimizes the difference between predicted and actual values (usually using Mean Squared Error).

Examples:

Predicting house prices based on size and location.

Forecasting sales based on advertising spend.

💡 Key idea: Linear regression models relationships with a straight line and is simple, interpretable, and widely used for prediction tasks.

## What is logistic regression?

Logistic regression is a supervised machine learning algorithm used for classification tasks, not regression.

It predicts the probability of an input belonging to a particular class using the logistic (sigmoid) function, which maps values to a range between 0 and 1.

Based on the predicted probability, the model assigns a class label (e.g., 0 or 1).

Examples:

Email → Spam or Not Spam

Customer → Will churn or not

💡 Key idea: Logistic regression models probabilities for classification using a linear combination of features passed through a sigmoid function.

## What is decision tree?

A Decision Tree is a supervised machine learning algorithm used for classification and regression tasks.

It splits the data into branches based on feature values, forming a tree-like structure with:

Root node: Represents the entire dataset.

Decision nodes: Points where data is split based on a feature.

Leaf nodes: Represent the final prediction (class label or numeric value).

The tree follows simple “if-then” rules, making it easy to interpret.

Examples:

Classifying whether a loan application should be approved.

Predicting house prices based on features like size and location.

💡 Key idea: Decision trees split data recursively to make predictions using interpretable rules.

## What is random forest?

A Random Forest is an ensemble learning algorithm that combines multiple decision trees to improve prediction accuracy and reduce overfitting.

Each tree is trained on a random subset of the data (bootstrapping) and a random subset of features.

For classification, the forest outputs the majority vote of all trees.

For regression, it outputs the average prediction of all trees.

Advantages:

Reduces overfitting compared to a single decision tree.

Handles high-dimensional data well.

Robust to noise and missing data.

💡 Key idea: Random Forest aggregates multiple trees to make more accurate and stable predictions.

## What is gradient boosting?

Gradient Boosting is an ensemble machine learning technique used for regression and classification tasks.

It builds a series of weak learners (usually decision trees) sequentially, where each new tree corrects the errors of the previous trees.

The model optimizes a loss function using gradient descent, hence the name “gradient boosting.”

Advantages:

High accuracy and performance on complex datasets

Can handle heterogeneous features

Popular implementations: XGBoost, LightGBM, CatBoost

💡 Key idea: Gradient Boosting combines many weak models to create a strong predictive model by focusing on the mistakes of prior models.

## What is support vector machine (SVM)?

A Support Vector Machine (SVM) is a supervised machine learning algorithm used for classification and regression tasks.

It works by finding the optimal hyperplane that best separates data points of different classes in a high-dimensional space.

The support vectors are the data points closest to the hyperplane, which determine its position and orientation.

For non-linearly separable data, SVM uses kernel functions (like RBF, polynomial) to map data into higher dimensions where a separating hyperplane can be found.

Advantages:

Effective in high-dimensional spaces

Robust to overfitting, especially in smaller datasets

💡 Key idea: SVM maximizes the margin between classes, making it a powerful and precise classifier.

# Unsupervised Learning

## What is unsupervised learning?

Unsupervised learning is a type of machine learning where the model learns patterns and structures from unlabeled data—data without predefined outputs.

The algorithm identifies relationships, groupings, or patterns in the input data.

Common tasks include:

Clustering: Grouping similar data points together (e.g., customer segmentation with K-Means).

Dimensionality reduction: Reducing the number of features while preserving important information (e.g., PCA).

Anomaly detection: Identifying unusual data points (e.g., fraud detection).

💡 Key idea: Unsupervised learning finds hidden patterns in data without needing labeled examples.

## What is clustering?

Clustering is an unsupervised learning technique that groups similar data points together based on their features.

The goal is to maximize similarity within clusters and minimize similarity between clusters.

Common clustering algorithms:

K-Means: Partitions data into k clusters by minimizing distances to cluster centroids.

Hierarchical Clustering: Builds a tree of clusters by merging or splitting clusters iteratively.

DBSCAN: Groups points based on density, allowing detection of clusters with arbitrary shapes.

Applications: Customer segmentation, image segmentation, anomaly detection

💡 Key idea: Clustering discovers natural groupings in unlabeled data.

## What is K-Means clustering?

K-Means clustering is an unsupervised learning algorithm used to partition data into K distinct clusters.

The algorithm works by:

Initializing K cluster centroids randomly.

Assigning each data point to the nearest centroid.

Recalculating centroids as the mean of all points in the cluster.

Repeating steps 2–3 until cluster assignments stabilize.

Goal: Minimize the sum of squared distances between points and their cluster centroids.

Applications: Customer segmentation, market analysis, image compression

💡 Key idea: K-Means groups similar data points together based on distance to cluster centers.

## What is Hierarchical clustering?

Hierarchical clustering is an unsupervised learning algorithm that builds a hierarchy of clusters in a tree-like structure called a dendrogram.

Two approaches:

Agglomerative (bottom-up): Start with each data point as its own cluster, then merge the closest clusters iteratively.

Divisive (top-down): Start with all data points in one cluster, then split clusters iteratively.

Does not require specifying the number of clusters in advance (you can choose by cutting the dendrogram at a desired level).

Applications: Gene expression analysis, document clustering, social network analysis

💡 Key idea: Hierarchical clustering groups data step by step, producing a tree of nested clusters that shows relationships at multiple levels.

## What is DBSCAN?

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is an unsupervised clustering algorithm that groups data points based on density.

It defines clusters as areas of high point density separated by areas of low density.

Key concepts:

Epsilon (ε): Maximum distance between two points to be considered neighbors.

MinPts: Minimum number of points required to form a dense region (cluster).

Core points: Points with at least MinPts neighbors within ε.

Border points: Points within ε of a core point but with fewer than MinPts neighbors.

Noise points: Points that do not belong to any cluster.

Advantages:

Can find clusters of arbitrary shape.

Does not require specifying the number of clusters in advance.

Applications: Anomaly detection, geospatial clustering, pattern recognition

💡 Key idea: DBSCAN detects dense regions in data and separates them from sparse/noisy regions.

## What is dimensionality reduction?

Dimensionality reduction is a technique in machine learning that reduces the number of input features (dimensions) while retaining the most important information.

Purpose:

Simplify the model and reduce computational cost.

Remove redundant or irrelevant features.

Help with visualization of high-dimensional data.

Common techniques:

PCA (Principal Component Analysis): Projects data onto directions of maximum variance.

t-SNE: Preserves local structure for visualization in 2D or 3D.

UMAP: Maintains global and local structure for visualization and clustering.

Applications: Data compression, noise reduction, feature extraction, visualization

💡 Key idea: Dimensionality reduction simplifies complex data while keeping the most meaningful patterns.

## What is PCA (Principal Component Analysis)?

PCA (Principal Component Analysis) is a dimensionality reduction technique that transforms high-dimensional data into a smaller set of uncorrelated variables called principal components.

How it works:

Finds directions (principal components) where the data varies the most.

Projects the data onto these directions, reducing dimensions while preserving most of the variance.

Benefits:

Reduces computational cost

Removes redundant features

Helps with visualization of high-dimensional data

Applications: Data compression, noise reduction, feature extraction, visualization

💡 Key idea: PCA compresses data by keeping the most important patterns (variance) and discarding less important information.

## What is t-SNE?

t-SNE (t-Distributed Stochastic Neighbor Embedding) is a dimensionality reduction and visualization technique used to map high-dimensional data into 2D or 3D space.

How it works:

Preserves local structure, meaning points that are close in high-dimensional space remain close in the low-dimensional map.

Converts pairwise similarities into probabilities and minimizes the difference between high- and low-dimensional distributions.

Applications:

Visualizing clusters in complex data (e.g., embeddings from NLP or images)

Exploring patterns in high-dimensional datasets

Limitation: Computationally intensive for very large datasets and mainly used for visualization, not modeling.

💡 Key idea: t-SNE turns complex high-dimensional data into an easy-to-visualize map, showing clusters and relationships clearly.

# Model Evaluation

## What are evaluation metrics for classification?

Common evaluation metrics for classification measure how well a model predicts the correct class labels:

Accuracy: Fraction of correctly predicted samples out of total samples.

Good for balanced datasets, but can be misleading for imbalanced classes.

Precision: Fraction of true positives out of all predicted positives.

Measures how many predicted positives are actually correct.

Recall (Sensitivity): Fraction of true positives out of all actual positives.

Measures how many actual positives the model correctly identifies.

F1-Score: Harmonic mean of precision and recall.

Balances precision and recall, useful for imbalanced datasets.

ROC-AUC (Receiver Operating Characteristic – Area Under Curve): Measures the model’s ability to distinguish between classes across all thresholds.

💡 Key idea: Choose metrics based on class balance and the cost of false positives vs false negatives.

## What are evaluation metrics for regression?

Common evaluation metrics for regression measure how well a model predicts continuous values:

Mean Absolute Error (MAE): Average of the absolute differences between predicted and actual values.

Easy to interpret, less sensitive to outliers.

Mean Squared Error (MSE): Average of the squared differences between predicted and actual values.

Penalizes large errors more heavily.

Root Mean Squared Error (RMSE): Square root of MSE.

Same units as the target variable, highlights large errors.

R-squared (R²): Proportion of variance in the target variable explained by the model.

Ranges from 0 to 1, higher is better.

Adjusted R-squared: Adjusted for the number of features, prevents inflated R² when adding irrelevant variables.

💡 Key idea: Regression metrics quantify prediction errors and help compare model performance.

# Model Optimization

## What is hyperparameter tuning?

Hyperparameter tuning is the process of choosing the best set of hyperparameters for a machine learning model to maximize its performance.

Hyperparameters are settings not learned from data, such as:

Learning rate, number of layers, number of neurons

Number of trees in Random Forest or depth of trees

Regularization strength (L1/L2), dropout rate

Common tuning methods:

Grid Search: Try all combinations of hyperparameters.

Random Search: Randomly sample combinations.

Bayesian Optimization: Uses past results to choose promising hyperparameters.

Goal: Improve model accuracy, reduce overfitting, and optimize training.

💡 Key idea: Hyperparameter tuning finds the optimal settings to make a model perform its best on unseen data.

## What is Grid Search vs Random Search?

Grid Search and Random Search are two common methods for hyperparameter tuning:

Grid Search:

Tries all possible combinations of specified hyperparameter values.

Guarantees finding the best combination in the search space.

Can be computationally expensive if there are many parameters or values.

Random Search:

Randomly samples a fixed number of combinations from the hyperparameter space.

More efficient for large search spaces.

May find good solutions faster, but not guaranteed to find the absolute best.

💡 Key idea:

Grid Search = exhaustive search

Random Search = sampling-based search, faster for large or complex spaces.

## What is regularization?

Regularization is a technique in machine learning and deep learning used to prevent overfitting and improve a model’s generalization to new data.

It works by adding a penalty to the model’s complexity, discouraging overly large or unnecessary weights.

Common types:

L1 Regularization (Lasso): Adds the sum of absolute weights as a penalty → can make some weights zero (feature selection).

L2 Regularization (Ridge): Adds the sum of squared weights as a penalty → reduces large weights but keeps all features.

Dropout: Randomly ignores neurons during training to reduce dependence on specific paths.

Early Stopping: Stops training before overfitting occurs.

💡 Key idea: Regularization simplifies the model to improve performance on unseen data and reduce overfitting.

## What is the difference between L1 and L2 regularization?

The difference between L1 and L2 regularization lies in how they penalize model weights:

L1 Regularization (Lasso):

Adds the sum of the absolute values of weights to the loss function.

Can shrink some weights exactly to zero, effectively performing feature selection.

Encourages sparse models.

L2 Regularization (Ridge):

Adds the sum of the squared values of weights to the loss function.

Shrinks weights gradually but rarely to zero, keeping all features but reducing their impact.

Encourages small, smooth weights.

💡 Key idea:

L1 → sparsity, feature selection

L2 → smooth weights, prevents large values

# Data Handling

## How do you handle missing data?

Handling missing data is an important part of data preprocessing. Common strategies include:

Removal:

Delete rows or columns with missing values if they are few and won’t affect the dataset significantly.

Imputation:

Mean/Median/Mode Imputation: Replace missing numeric values with the mean or median, and categorical values with the mode.

Forward/Backward Fill: Use previous or next values in sequential data.

K-Nearest Neighbors (KNN) Imputation: Replace missing values based on similar samples.

Predictive Imputation: Use a model to predict missing values based on other features.

Indicator Method:

Create a binary flag column indicating whether a value was missing.

💡 Key idea: The method depends on data type, amount of missing data, and its importance—imputation preserves data, deletion avoids bias from poor guesses.

## How do you handle imbalanced datasets?

Handling imbalanced datasets is crucial because models trained on them may favor the majority class and perform poorly on the minority class. Common strategies include:

Resampling Techniques:

Oversampling: Increase minority class samples (e.g., SMOTE – Synthetic Minority Oversampling Technique).

Undersampling: Reduce majority class samples to balance the dataset.

Class Weighting:

Assign higher weights to minority class samples during training so the model pays more attention to them.

## What is one-hot encoding?

One-hot encoding is a technique used to convert categorical variables into numeric form so that machine learning models can process them.

Each category is represented as a binary vector where:

The length of the vector = number of unique categories

1 indicates the presence of a category

0 indicates absence

Example:

Categories: Red, Green, Blue

One-hot encoding:

Red → [1, 0, 0]

Green → [0, 1, 0]

Blue → [0, 0, 1]

💡 Key idea: One-hot encoding transforms categorical data into a machine-readable numeric format without implying any order between categories.

##  What is label encoding?

Label encoding is a technique used to convert categorical variables into numeric form by assigning each unique category a unique integer value.

Example:

Categories: Red, Green, Blue

Label encoding:

Red → 0

Green → 1

Blue → 2

Use case: Works well for ordinal categories where there is an inherent order (e.g., Low → 0, Medium → 1, High → 2).

Caution: For nominal categories (no order), label encoding may introduce unintended ordinal relationships. In such cases, one-hot encoding is preferred.

💡 Key idea: Label encoding maps categories to integers so models can process them numerically.
